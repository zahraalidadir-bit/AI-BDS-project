```python
#load the brent and wti data

ro.r('''
library(readxl)

# EIA source URLs (same as the Python EIA_URLS dict, using a named list in R)
EIA_URLS <- list(
  WTI   = "https://www.eia.gov/dnav/pet/hist_xls/RWTCd.xls",
  Brent = "https://www.eia.gov/dnav/pet/hist_xls/RBRTEd.xls"
)

# Downloads a file from a URL into a temp path (auto-cleaned up), returns the path
download_eia_file <- function(url) {
  tmp <- tempfile(fileext = ".xls")
  download.file(url, destfile = tmp, mode = "wb", quiet = TRUE)
  return(tmp)
}

# Reads the "Data 1" sheet, renames columns, converts Date to Date type
load_eia_series <- function(xls_path, value_col_name) {
  df <- read_excel(xls_path, sheet = "Data 1", skip = 2)
  colnames(df) <- c("Date", value_col_name)
  df$Date <- as.Date(df$Date)
  return(as.data.frame(df))
}

# Download both series
wti_file   <- download_eia_file(EIA_URLS$WTI)
brent_file <- download_eia_file(EIA_URLS$Brent)

# Parse both series
wti   <- load_eia_series(wti_file,   "WTI_USD_per_bbl")
brent <- load_eia_series(brent_file, "Brent_USD_per_bbl")

''')
```

```python
# to extract gold price

ro.r('''
# install.packages("jsonlite")   # run once per Colab session
library(jsonlite)

# LBMA source URLs (same pattern as the EIA_URLS named list)
LBMA_URLS <- list(
  Gold_PM = "https://prices.lbma.org.uk/json/gold_pm.json",
  Gold_AM = "https://prices.lbma.org.uk/json/gold_am.json"
)

# Downloads a file from a URL into a temp path (auto-cleaned up), returns the path
download_lbma_file <- function(url) {
  tmp <- tempfile(fileext = ".json")
  download.file(url, destfile = tmp, mode = "wb", quiet = TRUE)
  return(tmp)
}

# Parses the LBMA JSON: records are {"d": "YYYY-MM-DD", "v": [USD, GBP, EUR]}
# EUR is NA before 1999, so the currency index must be selected explicitly.
load_lbma_series <- function(json_path, value_col_name,
                            currency = c("USD", "GBP", "EUR")) {
  currency <- match.arg(currency)
  idx <- switch(currency, USD = 1L, GBP = 2L, EUR = 3L)

  raw <- fromJSON(json_path, simplifyVector = FALSE)

  dates <- as.Date(vapply(raw, function(x) x$d, character(1)))
  vals  <- vapply(raw, function(x) {
    v <- x$v[[idx]]
    if (is.null(v)) NA_real_ else as.numeric(v)
  }, numeric(1))

  df <- data.frame(Date = dates, vals, stringsAsFactors = FALSE)
  colnames(df) <- c("Date", value_col_name)

  # drop non-quoted days, sort, remove any duplicate dates
  df <- df[!is.na(df[[value_col_name]]), ]
  df <- df[order(df$Date), ]
  df <- df[!duplicated(df$Date), ]
  rownames(df) <- NULL
  return(as.data.frame(df))
}

# Download and parse the PM auction series in USD
gold_file <- download_lbma_file(LBMA_URLS$Gold_PM)
gold      <- load_lbma_series(gold_file, "Gold_USD_per_oz", currency = "USD")

# Restrict to the free-market era: gold was administratively pegged until 1971
# and US private ownership was only relegalised in Dec 1974.
gold <- gold[gold$Date >= as.Date("1975-01-02"), ]
rownames(gold) <- NULL

# Administrative break: LBMA Gold Price replaced the London Gold Fix on 2015-03-20
gold$LBMA_electronic <- as.integer(gold$Date >= as.Date("2015-03-20"))

cat("[check] Gold PM obs:", nrow(gold),
    "| range:", format(min(gold$Date)), "to", format(max(gold$Date)), "\n")
cat("[check] non-positive prices:", sum(gold$Gold_USD_per_oz <= 0), "\n")
cat("[check] post-2015 methodology days:", sum(gold$LBMA_electronic), "\n")

''')
```

```python
%%R
```
```python
library(quantmod)

# Yahoo Finance continuous front-month futures tickers
# ZW=F -> CBOT Soft Red Winter Wheat
# KE=F -> KC Hard Red Winter Wheat (formerly KCBT)

get_wheat_series <- function(ticker, value_col_name) {
  getSymbols(ticker, src = "yahoo", auto.assign = FALSE,from = "1990-01-01") -> xts_data
  # yahoo data starts from 2000
  df <- data.frame(Date = index(xts_data), Close = as.numeric(Cl(xts_data)))
  colnames(df) <- c("Date", value_col_name)
  return(df)
}

srw_wheat <- get_wheat_series("ZW=F", "SRW_Wheat_USD_per_bu")
hrw_wheat <- get_wheat_series("KE=F", "HRW_Wheat_USD_per_bu")

```

```python
%%R
```
```python
library(zoo)

# --- Fetching CPI data to remove the inflation effect, so that price can be used as a factor in duration prediction
# without inflation trend confounding the effect of price level
cpi <- read.csv("https://fred.stlouisfed.org/graph/fredgraph.csv?id=CPIAUCSL")
colnames(cpi) <- c("Date", "CPI")
cpi$Date <- as.Date(cpi$Date)
cpi$YearMonth <- format(cpi$Date, "%Y-%m-01")

add_real_price <- function(df, price_col) {

  real_col <- paste0(sub("_USD_per_bbl", "", price_col), "_real")
  df[[real_col]] <- NULL
  df$CPI <- NULL
  df$YearMonth <- NULL

  df$YearMonth <- format(df$Date, "%Y-%m-01")
  df <- merge(df, cpi[, c("YearMonth", "CPI")], by = "YearMonth", all.x = TRUE)
  df <- df[order(df$Date), ]

  #the base for cpi is the last month with data , also as cpi is calculation is published after the month is ended, so for the current month
  #(only the last month in the dataset), we use the month before data
  df$CPI <- na.locf(df$CPI, na.rm = FALSE)

  base_cpi <- tail(df$CPI[!is.na(df$CPI)], 1)
  df[[real_col]] <- df[[price_col]] * (base_cpi / df$CPI)

  df$YearMonth <- NULL
  return(df)
}

wti   <- add_real_price(wti,   "WTI_USD_per_bbl")
brent <- add_real_price(brent, "Brent_USD_per_bbl")
gold <- add_real_price(gold, "Gold_USD_per_oz")
srw_wheat <- add_real_price(srw_wheat,   "srw_USD_per_bu")
hrw_wheat <- add_real_price(hrw_wheat, "hrw_USD_per_bu")
head(wti)
head(brent)

```

```python
# now let's calculate the return of nominal price for MSGarch and Garch family, as they need to use return for their modeling process.

ro.r('''
add_log_return <- function(df, price_col) {
  return_col <- paste0(tolower(sub("_USD_per_.*", "", price_col)), "_return")
  df[[return_col]] <- NULL  # safe to re-run

  df <- df[order(df$Date), ]
  df[[return_col]] <- c(NA, diff(log(df[[price_col]]))) * 100
  return(df)
}

wti   <- add_log_return(wti,   "WTI_USD_per_bbl")
brent <- add_log_return(brent, "Brent_USD_per_bbl")
gold <- add_log_return(gold, "Gold_USD_per_oz")
srw_wheat <- add_log_return(srw_wheat,   "srw_USD_per_bu")
hrw_wheat <- add_log_return(hrw_wheat, "hrw_USD_per_bu")


''')
```

```python
# ── Fix the analysis window here, BEFORE any diagnostics run ────────────
# Keep a full-history copy of each series first, in case a future full-
# sample test (e.g. Zivot-Andrews for structural breaks) ever needs it.
# Everything from this cell down uses the trimmed window only.
ro.r('''
wti_full       <- wti
brent_full     <- brent
gold_full      <- gold
hrw_wheat_full <- hrw_wheat
srw_wheat_full <- srw_wheat

start_date <- as.Date("2010-01-01")
end_date   <- as.Date("2026-07-01")

cat("Fixed date range:", format(start_date), "to", format(end_date), "\n")

wti       <- wti[wti$Date             >= start_date & wti$Date             <= end_date, ]
brent     <- brent[brent$Date         >= start_date & brent$Date           <= end_date, ]
gold      <- gold[gold$Date           >= start_date & gold$Date            <= end_date, ]
hrw_wheat <- hrw_wheat[hrw_wheat$Date >= start_date & hrw_wheat$Date       <= end_date, ]
srw_wheat <- srw_wheat[srw_wheat$Date >= start_date & srw_wheat$Date       <= end_date, ]

cat("Rows after trim -- WTI:", nrow(wti), "| Brent:", nrow(brent),
    "| Gold:", nrow(gold), "| HRW:", nrow(hrw_wheat), "| SRW:", nrow(srw_wheat), "\n")
''')
```

```python
%%R
if (!requireNamespace("tseries", quietly = TRUE)) {
  install.packages("tseries", repos = "https://cloud.r-project.org", quiet = TRUE)
}
suppressPackageStartupMessages(library(tseries))

run_stationarity_tests <- function(df, return_col, label) {
  cat("\n#########################################\n")
  cat("###", label, "\n")
  cat("#########################################\n")

  if (!(return_col %in% colnames(df))) {
    cat("Column", return_col, "not found in", label, "- skipping.\n")
    return(invisible(NULL))
  }

  ret <- na.omit(df[[return_col]])

  cat("=== ADF test (H0: non-stationary) ===\n")
  print(suppressWarnings(adf.test(ret)))

  cat("\n=== KPSS test (H0: stationary) ===\n")
  print(suppressWarnings(kpss.test(ret)))
}

# List each series with its dataframe and its return column name
series_list <- list(
  list(df = wti,       return_col = "wti_return",   label = "WTI"),
  list(df = brent,     return_col = "brent_return",  label = "Brent"),
  list(df = gold,      return_col = "gold_return",   label = "Gold"),
  list(df = srw_wheat, return_col = "srw_return",    label = "SRW Wheat"),
  list(df = hrw_wheat, return_col = "hrw_return",    label = "HRW Wheat")
)

for (s in series_list) {
  run_stationarity_tests(s$df, s$return_col, s$label)
}
```

ADF: p-value should be smaller than the threshold (typically < 0.05) → reject H0(non-stationary) → series is stationary.
Here for all 5 indices the p-value is  < 0.05, so we reject H0→ stationary. ✓

KPSS: p-value should be larger than the threshold (typically > 0.05) → fail to reject H0(stationary) → consistent with the series being stationary.
Here, for all 5 indices the p-value is  > 0.05, so we fail to reject H0
​→ stationary. ✓

```python
# Check for Autocorrelation (Box test)
import rpy2.robjects as ro
ro.r('''
box_ljung <- function(ret, label) {
  bt <- Box.test(as.numeric(ret), lag = 10, type = "Ljung-Box")
  cat("===", label, "===\n")
  cat("Test statistic:", round(bt$statistic, 3), "\n")
  cat("p-value:       ", format.pval(bt$p.value, digits = 4), "\n")
  cat("Lag:           ", bt$parameter, "\n\n")
}

box_ljung(na.omit(wti$wti_return),     "WTI Returns")
box_ljung(na.omit(brent$brent_return), "Brent Returns")
box_ljung(na.omit(gold$gold_return),     "gold Returns")
box_ljung(na.omit(hrw_wheat$hrw_return), "hrw Returns")
box_ljung(na.omit(srw_wheat$srw_return), "srw Returns")

''')
# recommended lag is ln(duration), so from = "2010-01-01" to   = "2024-12-31" is about 14 * 365
```

To check whether the return series contains any predictable autocorrelation structure that a constant-mean model would miss, we applied the Ljung-Box test at lag 10. For WTI and Brent  the p-value is less than 5%, the test rejected the null hypothesis of no autocorrelation, indicating some residual serial dependence in daily returns.

```python
 # but how practically meaningful (not just statistically significant) is this value.
ro.r('''
lag1_acf <- function(ret, label) {
  ac <- acf(as.numeric(ret), plot = FALSE)$acf[2]  # [2] = lag 1 (index 1 is lag 0, always 1.0)
  cat(label, "lag-1 autocorrelation:", round(ac, 4), "\n")
}

lag1_acf(na.omit(wti$wti_return),     "WTI")
lag1_acf(na.omit(brent$brent_return), "Brent")

''')
```

While the Ljung-Box test rejected the null of no autocorrelation for WTI, Brent, and Gold, the effect size is negligible — a statistical artifact of the large sample size rather than a practically exploitable structure. We therefore adopt a constant mean specification and proceed to test directly for conditional heteroskedasticity (ARCH effects), which is the primary motivation for GARCH-family modeling.

```python
##Arch test
ro.r('''
if (!requireNamespace("FinTS", quietly = TRUE)) {
  install.packages("FinTS", repos = "https://cloud.r-project.org", quiet = TRUE)
}
suppressPackageStartupMessages(library(FinTS))

arch_lm_test <- function(ret, label) {
  cat("===", label, "===\n")
  print(suppressWarnings(ArchTest(as.numeric(ret), lags = 12)))
  cat("\n")
}

arch_lm_test(na.omit(wti$wti_return),       "WTI ARCH-LM test (H0: no ARCH effects)")
arch_lm_test(na.omit(brent$brent_return),   "Brent ARCH-LM test (H0: no ARCH effects)")
arch_lm_test(na.omit(gold$gold_return),     "Gold ARCH-LM test (H0: no ARCH effects)")
arch_lm_test(na.omit(hrw_wheat$hrw_return), "HRW ARCH-LM test (H0: no ARCH effects)")
arch_lm_test(na.omit(srw_wheat$srw_return), "SRW ARCH-LM test (H0: no ARCH effects)")
# you might think why all th ep-value are the same, no woriies this number is the least number that R can proceed. So , there are in fact different but R can"t comprehend it.
''')
```

The ARCH-LM test rejects the null of no ARCH effects for all five series (df = 12, p < 2.2e-16, R's floating-point display limit). Test statistics range from ~245 (HRW/Gold) to 1107.5 (WTI), with Brent (804.73) and SRW (623.23) in between. WTI and Brent show the most pronounced volatility clustering on this window; Gold and HRW show comparatively weak clustering, essentially tied with each other. All five series still reject the null, so GARCH-family models remain appropriate across the board.

We wanted to check whether WTI volatility has a memory — meaning does a turbulent day tend to be followed by more turbulent days, and does a calm day tend to be followed by more calm days?
The answer was a very strong yes. The test gave us a chi-squared value of 1107.5 (df = 12) and a p-value so close to zero it is practically impossible to get this result by chance. This tells us that volatility in WTI markets does not jump around randomly — it clusters together in time. Bad periods stick around, and good periods stick around too.

```python
# disturbution tests
ro.r('''
if (!requireNamespace("moments", quietly = TRUE)) {
  install.packages("moments", repos = "https://cloud.r-project.org", quiet = TRUE)
}
suppressPackageStartupMessages(library(tseries))
suppressPackageStartupMessages(library(moments))

distribution_summary <- function(ret, label) {
  ret <- na.omit(as.numeric(ret))
  jb <- jarque.bera.test(ret)

  cat("===", label, "===\n")
  cat("N obs:            ", length(ret), "\n")
  cat("Mean:             ", round(mean(ret), 4), "\n")
  cat("Std Dev:          ", round(sd(ret), 4), "\n")
  cat("Min / Max:        ", round(min(ret), 2), "/", round(max(ret), 2), "\n")
  cat("Skewness:         ", round(skewness(ret), 4), "\n")
  cat("Excess kurtosis:  ", round(kurtosis(ret) - 3, 4), "\n")
  cat("Jarque-Bera stat: ", round(jb$statistic, 2), "\n")
  cat("Jarque-Bera p-val:", format.pval(jb$p.value, digits = 3), "\n\n")
}

distribution_summary(wti$wti_return,     "WTI Returns")
distribution_summary(brent$brent_return, "Brent Returns")
distribution_summary(gold$gold_return, "gold Returns")
distribution_summary(hrw_wheat$hrw_return, "hrw Returns")
distribution_summary(srw_wheat$srw_return, "srw Returns")
''')
```

Skewness no longer tells a single "oil is left-skewed" story now that WTI and Brent point in opposite directions. Brent's strong negative skew (-2.602) fits the classic pattern — large down-days pulling the tail left — which is the usual motivation for an asymmetric variance model like eGARCH or gjrGARCH. WTI, however, is now positively skewed (0.861), the opposite sign. That's worth being cautious about: return skewness and *volatility* asymmetry (how variance responds to positive vs. negative shocks) aren't the same thing, so WTI's positive return skew doesn't automatically rule out a leverage effect in its variance process — but it does mean we can't lean on "left skew → use an asymmetric variance model" as the justification for WTI the way we can for Brent. Before drawing conclusions for either series, we check the sign-bias test next, since that's the test that directly asks whether variance responds asymmetrically to shock direction, rather than inferring it indirectly from the shape of the unconditional return distribution.

Also as the skewness shows negative or left skew >> EGarch or  might be a better fit for oil but before that let's check with sign-bias test because negative or left skewness might appear due to other reasons.

```python
ro.r('''
library(MSGARCH)

sign_bias_test <- function(ret, label) {
  spec <- CreateSpec(variance.spec = list(model = "sGARCH"),
                      distribution.spec = list(distribution = "norm"),
                      switch.spec = list(do.mix = FALSE, K = 1))    #we fit a normal GARCH(1,1) but on MSGARCH library (K=1) so we have 1 regime.
  fit <- FitML(spec = spec, data = ret)
  sigma <- as.numeric(Volatility(fit))     # fitted conditional sd
  e <- as.numeric(ret) - mean(ret)         # demeaned residual
  z <- e / sigma                            # standardized residual
  z2 <- z^2

  e_lag <- c(NA, e[-length(e)])            # previous day's shock
  neg_dummy <- as.numeric(e_lag < 0)       # 1 if yesterday's shock was negative

  df_test <- na.omit(data.frame(
    z2 = z2,
    neg_dummy = neg_dummy,
    neg_size  = neg_dummy * e_lag,
    pos_size  = (1 - neg_dummy) * e_lag
  ))

  sign_bias     <- lm(z2 ~ neg_dummy, data = df_test)
  neg_size_bias <- lm(z2 ~ neg_size,  data = df_test)
  pos_size_bias <- lm(z2 ~ pos_size,  data = df_test)
  joint         <- lm(z2 ~ neg_dummy + neg_size + pos_size, data = df_test)

  cat("===", label, "===\n")
  cat("Sign Bias t-stat:      ", round(summary(sign_bias)$coefficients["neg_dummy","t value"], 3),
      " p-value:", round(summary(sign_bias)$coefficients["neg_dummy","Pr(>|t|)"], 4), "\n")
  cat("Negative Size Bias t: ", round(summary(neg_size_bias)$coefficients["neg_size","t value"], 3),
      " p-value:", round(summary(neg_size_bias)$coefficients["neg_size","Pr(>|t|)"], 4), "\n")
  cat("Positive Size Bias t: ", round(summary(pos_size_bias)$coefficients["pos_size","t value"], 3),
      " p-value:", round(summary(pos_size_bias)$coefficients["pos_size","Pr(>|t|)"], 4), "\n")
  jf <- summary(joint)$fstatistic
  cat("Joint F-stat:          ", round(jf[1], 3),
      " p-value:", format.pval(pf(jf[1], jf[2], jf[3], lower.tail = FALSE), digits = 4), "\n\n")

}

sign_bias_test(na.omit(wti$wti_return),     "WTI")
sign_bias_test(na.omit(brent$brent_return), "Brent")
sign_bias_test(na.omit(gold$gold_return), "gold")
sign_bias_test(na.omit(hrw_wheat$hrw_return), "hrw")
sign_bias_test(na.omit(srw_wheat$srw_return), "srw‹")
''')
```

To determine whether volatility responds symmetrically to positive and negative return shocks, Engle's sign bias test was applied to each series. For WTI and Brent, the joint test rejected the null of no asymmetry (p = 0.0013 and p = 0.0005, respectively), and both series now show significance across all three individual bias components (sign bias, negative size bias, positive size bias) — indicating a pronounced and well-defined leverage effect in both oil benchmarks, not just Brent. Gold showed no evidence of asymmetry (joint p = 0.576), consistent with a symmetric volatility response to shocks. HRW wheat showed no asymmetry at any level — joint test (p = 0.641) and all three individual components are comfortably insignificant, indicating a genuinely symmetric response. SRW wheat is more mixed: the joint test still does not reject symmetry overall (p = 0.123), but its sign-bias (p = 0.025) and positive size-bias (p = 0.041) components are individually significant, pointing to a weak but real asymmetric tendency that the joint test isn't picking up cleanly. On this basis, an asymmetric GARCH specification (e.g. GJR-GARCH or EGARCH) is used for WTI and Brent, a symmetric GARCH is used for Gold and HRW, and both specifications are worth testing for SRW given its partial, component-level evidence of asymmetry.

Why are some of the return values omitted ?

The MSGARCH model cannot be estimated on non-positive prices, since log returns are mathematically undefined in that case; any such value would break the likelihood-based parameter estimation. This becomes relevant for our data because of a single anomalous event: on April 20, 2020, WTI recorded a negative spot price of −$36.98 per barrel, caused by an unprecedented supply-demand imbalance during the COVID-19 pandemic. Aside from this one event, the raw EIA data contained no missing observations for either series (10,200 daily observations for WTI, 9,932 for Brent). Because the negative price makes the log return undefined both on that day and the following one, two consecutive WTI return observations (April 20 and 21, 2020) were excluded prior to model estimation. Brent was not affected by this issue.



# Now we can run our MSGarh

CUSUM → after MSGARCH is fit (as a validation/robustness check)
CUSUM needs a fitted model's residuals (or a mean-model relationship) to test for breaks in, and its purpose here is to independently verify the instability points your MSGARCH model already identified. So it comes chronologically after MSGARCH fitting and instability-point extraction — as a "does an independent method agree with what MSGARCH found?" sanity check, not as an input into the MSGARCH model itself.


don"t forget to run it after MSGARCH.

```python
ro.r('''
```
```python
fit_and_report <- function(returns, label, model_name, distr_name) {
  spec <- CreateSpec(variance.spec = list(model = model_name),
                      distribution.spec = list(distribution = distr_name),
                      switch.spec = list(K = 2))
  fit <- tryCatch(FitML(spec = spec, data = returns), error = function(e) NULL)
  if (is.null(fit)) {
    cat(sprintf("%-8s | %-10s-%-6s | FAILED TO CONVERGE\n", label, model_name, distr_name))
    return(invisible(NULL))
  }
  ll  <- as.numeric(logLik(fit))
  aic <- -2*ll + 2*length(fit$par)
  bic <- -2*ll + log(length(returns))*length(fit$par)
  cat(sprintf("%-8s | %-10s-%-6s | LogLik=%.2f  AIC=%.2f  BIC=%.2f\n", label, model_name, distr_name, ll, aic, bic))
  return(invisible(list(model=model_name, distr=distr_name, ll=ll, aic=aic, bic=bic)))
}

wti_ret   <- as.numeric(na.omit(wti$wti_return))
brent_ret <- as.numeric(na.omit(brent$brent_return))
gold_ret <- as.numeric(na.omit(gold$gold_return))
hrw_ret <- as.numeric(na.omit(hrw_wheat$hrw_return))
srw_ret <- as.numeric(na.omit(srw_wheat$srw_return))

cat("=== WTI ===\n")
invisible(fit_and_report(wti_ret, "WTI", "sGARCH", "std"))
invisible(fit_and_report(wti_ret, "WTI", "sGARCH", "sstd"))
invisible(fit_and_report(wti_ret, "WTI", "sGARCH", "ged"))
invisible(fit_and_report(wti_ret, "WTI", "eGARCH", "std"))
invisible(fit_and_report(wti_ret, "WTI", "eGARCH", "sstd"))
invisible(fit_and_report(wti_ret, "WTI", "eGARCH", "ged"))
invisible(fit_and_report(wti_ret, "WTI", "gjrGARCH", "std"))
invisible(fit_and_report(wti_ret, "WTI", "gjrGARCH", "sstd"))
invisible(fit_and_report(wti_ret, "WTI", "gjrGARCH", "ged"))

cat("\n=== Brent ===\n")
invisible(fit_and_report(brent_ret, "Brent", "sGARCH", "std"))
invisible(fit_and_report(brent_ret, "Brent", "sGARCH", "sstd"))
invisible(fit_and_report(brent_ret, "Brent", "sGARCH", "ged"))
invisible(fit_and_report(brent_ret, "Brent", "eGARCH", "std"))
invisible(fit_and_report(brent_ret, "Brent", "eGARCH", "sstd"))
invisible(fit_and_report(brent_ret, "Brent", "eGARCH", "ged"))
invisible(fit_and_report(brent_ret, "Brent", "gjrGARCH", "std"))
invisible(fit_and_report(brent_ret, "Brent", "gjrGARCH", "sstd"))
invisible(fit_and_report(brent_ret, "Brent", "gjrGARCH", "ged"))

cat("\n=== gold ===\n")
invisible(fit_and_report(gold_ret, "gold", "sGARCH", "std"))
invisible(fit_and_report(gold_ret, "gold", "sGARCH", "sstd"))
invisible(fit_and_report(gold_ret, "gold", "sGARCH", "ged"))
invisible(fit_and_report(gold_ret, "gold", "eGARCH", "std"))
invisible(fit_and_report(gold_ret, "gold", "eGARCH", "sstd"))
invisible(fit_and_report(gold_ret, "gold", "eGARCH", "ged"))
invisible(fit_and_report(gold_ret, "gold", "gjrGARCH", "std"))
invisible(fit_and_report(gold_ret, "gold", "gjrGARCH", "sstd"))
invisible(fit_and_report(gold_ret, "gold", "gjrGARCH", "ged"))

cat("\n=== hrw ===\n")
invisible(fit_and_report(hrw_ret, "hrw", "sGARCH", "std"))
invisible(fit_and_report(hrw_ret, "hrw", "sGARCH", "sstd"))
invisible(fit_and_report(hrw_ret, "hrw", "sGARCH", "ged"))
invisible(fit_and_report(hrw_ret, "hrw", "eGARCH", "std"))
invisible(fit_and_report(hrw_ret, "hrw", "eGARCH", "sstd"))
invisible(fit_and_report(hrw_ret, "hrw", "eGARCH", "ged"))
invisible(fit_and_report(hrw_ret, "hrw", "gjrGARCH", "std"))
invisible(fit_and_report(hrw_ret, "hrw", "gjrGARCH", "sstd"))
invisible(fit_and_report(hrw_ret, "hrw", "gjrGARCH", "ged"))

cat("\n=== srw ===\n")
invisible(fit_and_report(srw_ret, "srw", "sGARCH", "std"))
invisible(fit_and_report(srw_ret, "srw", "sGARCH", "sstd"))
invisible(fit_and_report(srw_ret, "srw", "sGARCH", "ged"))
invisible(fit_and_report(srw_ret, "srw", "eGARCH", "std"))
invisible(fit_and_report(srw_ret, "srw", "eGARCH", "sstd"))
invisible(fit_and_report(srw_ret, "srw", "eGARCH", "ged"))
invisible(fit_and_report(srw_ret, "srw", "gjrGARCH", "std"))
invisible(fit_and_report(srw_ret, "srw", "gjrGARCH", "sstd"))
invisible(fit_and_report(srw_ret, "srw", "gjrGARCH", "ged"))
''')
```

We compared models using BIC rather than AIC or raw log-likelihood, for a simple reason: log-likelihood always goes up when you add parameters, so it can't tell you whether the extra complexity is actually worth it. AIC penalizes added parameters by a fixed 2k, but that penalty doesn't scale with sample size, and in large datasets like ours it tends to favor models with too many regimes (Akaike, 1974; Psaradakis and Spagnolo, 2003). BIC penalizes each parameter by k·ln(n) instead, which is much stricter once n is in the thousands, and it has a cleaner interpretation: it approximates a Bayes factor, so a BIC gap can be read as actual evidence for one model over another rather than just a smaller error number (Schwarz, 1978; Kass and Raftery, 1995). Using their rough guide, a BIC gap under 2 barely counts as evidence, 2–6 is "positive," 6–10 is "strong," and above 10 is "very strong." This is also the standard used in the Markov-switching GARCH literature we're drawing on (Haas, Mittnik and Paolella, 2004; Ardia et al., 2019), so it keeps our model choice consistent with the framework itself.

```python
%%R
grid_results <- list()

fit_and_report <- function(returns, label, model_name, distr_name, K) {
  spec <- CreateSpec(variance.spec = list(model = model_name),
                      distribution.spec = list(distribution = distr_name),
                      switch.spec = list(K = K))
  fit <- tryCatch(FitML(spec = spec, data = returns), error = function(e) NULL)
  key <- paste(label, model_name, distr_name, K, sep = "_")

  if (is.null(fit)) {
    cat(sprintf("%-8s | K=%d | %-10s-%-6s | FAILED TO CONVERGE\n", label, K, model_name, distr_name))
    return(invisible(NULL))
  }

  ll  <- as.numeric(logLik(fit))
  aic <- AIC(fit); bic <- BIC(fit)
  cat(sprintf("%-8s | K=%d | %-10s-%-6s | LogLik=%.2f AIC=%.2f BIC=%.2f\n",
              label, K, model_name, distr_name, ll, aic, bic))

  grid_results[[key]] <<- list(label = label, model = model_name, distr = distr_name, K = K,
                                ll = ll, aic = aic, bic = bic, fit = fit)
  invisible(NULL)
}

wti_ret   <- as.numeric(na.omit(wti$wti_return))
brent_ret <- as.numeric(na.omit(brent$brent_return))

models <- c("sGARCH", "eGARCH", "gjrGARCH")
distrs <- c("std", "sstd", "ged")

cat("=== WTI ===\n")
for (K in c(1, 2)) for (m in models) for (d in distrs) fit_and_report(wti_ret, "WTI", m, d, K)

cat("\n=== Brent ===\n")
for (K in c(1, 2)) for (m in models) for (d in distrs) fit_and_report(brent_ret, "Brent", m, d, K)

cat("=== gold ===\n")
for (K in c(1, 2)) for (m in models) for (d in distrs) fit_and_report(gold_ret, "gold", m, d, K)

cat("\n=== hrw ===\n")
for (K in c(1, 2)) for (m in models) for (d in distrs) fit_and_report(hrw_ret, "hrw", m, d, K)

cat("\n=== srw ===\n")
for (K in c(1, 2)) for (m in models) for (d in distrs) fit_and_report(srw_ret, "srw", m, d, K)
```

```python
ro.r('''
rank_all_by_bic <- function(label) {
  # 1. Filter elements directly (works even if grid_results lacks names)
  matching <- Filter(function(x) !is.null(x$label) && x$label == label, grid_results)
  
  # 2. Return an empty data frame if no matches exist
  if (length(matching) == 0) {
    cat(sprintf("Warning: No entries found for label '%s'\n", label))
    return(data.frame(model = character(), K = integer(), BIC = numeric(), AIC = numeric(), delta_BIC = numeric()))
  }
  
  # 3. Combine matching list elements into a data frame
  tbl <- do.call(rbind, lapply(matching, function(r) {
    data.frame(model = paste0(r$model, "-", r$distr), K = r$K,
               BIC = round(r$bic, 2), AIC = round(r$aic, 2),
               stringsAsFactors = FALSE)
  }))
  
  tbl <- tbl[order(tbl$BIC), ]
  tbl$delta_BIC <- round(tbl$BIC - min(tbl$BIC), 2)
  rownames(tbl) <- NULL
  tbl
}

cat("=== WTI: ranked by BIC ===\n"); print(rank_all_by_bic("WTI"))
cat("\n=== Brent: ranked by BIC ===\n"); print(rank_all_by_bic("Brent"))
cat("\n=== gold: ranked by BIC ===\n"); print(rank_all_by_bic("gold"))
cat("\n=== hrw: ranked by BIC ===\n"); print(rank_all_by_bic("hrw"))
cat("\n=== srw: ranked by BIC ===\n"); print(rank_all_by_bic("srw"))
''')
```

```python
ro.r('''
# single ranked table sorted purely by BIC, combining all K=1 and K=2 results together
rank_all_by_bic <- function(label) {
  keys <- names(grid_results)[sapply(grid_results, function(x) x$label == label)]
  tbl <- do.call(rbind, lapply(keys, function(k) {
    r <- grid_results[[k]]
    data.frame(model = paste0(r$model, "-", r$distr), K = r$K,
               BIC = round(r$bic, 2), AIC = round(r$aic, 2))
  }))
  tbl <- tbl[order(tbl$BIC), ]
  tbl$delta_BIC <- round(tbl$BIC - min(tbl$BIC), 2)
  rownames(tbl) <- NULL
  tbl
}

cat("=== WTI: ranked by BIC ===\n"); print(rank_all_by_bic("WTI"))
cat("\n=== Brent: ranked by BIC ===\n"); print(rank_all_by_bic("Brent"))
cat("\n=== gold: ranked by BIC ===\n"); print(rank_all_by_bic("gold"))
cat("\n=== hrw: ranked by BIC ===\n"); print(rank_all_by_bic("hrw"))
cat("\n=== srw: ranked by BIC ===\n"); print(rank_all_by_bic("srw"))
''')
```

To check whether adding a second regime was actually worth the extra complexity, we ranked every fitted model (single-regime and two-regime, across all model-distribution combinations) together by BIC, rather than comparing regime and non-regime models separately.

On the current 2010–2026 window, every one of the five series now has a single-regime (K=1) model at the top of its ranking . For WTI, the best model is an eGARCH with skewed Student-t errors (K=1); its two-regime counterpart is only third-ranked, 10.34 BIC points behind, and the second-place model (gjrGARCH-sstd, K=1, Δ=5.92) is also single-regime — so the top two spots both belong to K=1 specifications. For Brent, the case against a second regime is even sharper: the top two models are both single-regime eGARCH specifications (skewed-t and plain-t), and the first two-regime model to appear at all — gjrGARCH-sstd, K=2 — trails by 19.21 BIC points, which on the Kass-Raftery scale we're using (>10 = "very strong") counts as very strong evidence against the extra complexity. Gold, HRW, and SRW show the same pattern even more decisively, with their best two-regime models trailing the single-regime winner by 19 to 42 BIC points.

On raw BIC alone, none of the five series currently justify a two-regime specification. That doesn't necessarily settle the question, though: a single-regime model can still leave volatility clustering un-modeled (an ARCH-LM failure) even when it wins on BIC — a real cost that BIC's parameter penalty doesn't directly capture. The next step, as before, is to fit each series' K=1 winner and check its residual diagnostics (mean-equation, ARCH-LM, sign-bias), escalating to K=2 only where the single-regime model actually fails those checks.


```python
ro.r('''
mean_resid_check <- function(fit, returns, label = "") {
  vol <- Volatility(fit)
  std_resid <- returns / vol
  bt <- Box.test(std_resid, lag = 10, type = "Ljung-Box")
  cat(sprintf("--- %s ---\n", label))
  cat("Ljung-Box on standardized residuals: stat =", round(bt$statistic, 2),
      " p =", format.pval(bt$p.value, digits = 4), "\n\n")
}
''')
```

```python
# 1) Mean-equation check (Ljung-Box on residuals themselves)
ro.r('''
mean_resid_check(grid_results[["WTI_eGARCH_sstd_1"]]$fit,   wti_ret,   "WTI - eGARCH-sstd (K=1)")
mean_resid_check(grid_results[["Brent_eGARCH_sstd_1"]]$fit, brent_ret, "Brent - eGARCH-sstd (K=1)")
mean_resid_check(grid_results[["gold_sGARCH_std_1"]]$fit,   gold_ret,  "gold - sGARCH-std (K=1)")
mean_resid_check(grid_results[["hrw_eGARCH_sstd_1"]]$fit,   hrw_ret,   "hrw - eGARCH-sstd (K=1)")
mean_resid_check(grid_results[["srw_sGARCH_sstd_1"]]$fit,   srw_ret,   "srw - sGARCH-sstd (K=1)")
''')
```

On the current window, all five series' K=1 winners pass the mean-equation check cleanly (p = 0.11–0.56) > 0.05 — no leftover autocorrelation in the standardized residuals for any of them. 

```python
ro.r('''
library(FinTS)

residual_check <- function(fit, returns, label = "") {
  vol <- Volatility(fit)
  std_resid <- returns / vol
  sq_resid  <- std_resid^2

  cat(sprintf("--- %s ---\n", label))

  arch <- ArchTest(std_resid, lags = 10)
  cat("ARCH-LM on standardized residuals:  stat =", round(arch$statistic, 2),
      " p =", format.pval(arch$p.value, digits = 4), "\n")

  bt <- Box.test(sq_resid, lag = 10, type = "Ljung-Box")
  cat("Ljung-Box on squared residuals:     stat =", round(bt$statistic, 2),
      " p =", format.pval(bt$p.value, digits = 4), "\n\n")
}
''')
```

```python
# 2) Sign bias test on standardized residuals

ro.r('''
sign_bias_test(na.omit(as.numeric(wti_ret   / Volatility(grid_results[["WTI_eGARCH_sstd_1"]]$fit))),   "WTI residuals")
sign_bias_test(na.omit(as.numeric(brent_ret / Volatility(grid_results[["Brent_eGARCH_sstd_1"]]$fit))), "Brent residuals")
sign_bias_test(na.omit(as.numeric(gold_ret  / Volatility(grid_results[["gold_sGARCH_std_1"]]$fit))),   "Gold residuals")
sign_bias_test(na.omit(as.numeric(hrw_ret   / Volatility(grid_results[["hrw_eGARCH_sstd_1"]]$fit))),   "HRW residuals")
sign_bias_test(na.omit(as.numeric(srw_ret   / Volatility(grid_results[["srw_sGARCH_sstd_1"]]$fit))),   "SRW residuals")
''')
```

WTI and Brent both come out reasonably well here. The overall (joint) test doesn't flag either one as asymmetric — WTI comes in at p=0.097, Brent at p=0.100, both above the 0.05 line.

That said, it's not a fully clean pass. Two of the three individual pieces of the test — the sign-bias term and the positive-size-bias term — are still significant for both WTI and Brent (p in the 0.02–0.04 range). In plain terms: the eGARCH model is catching most of the leverage effect (the tendency for bad news to spike volatility more than good news), but a small amount is still slipping through.

SRW shows a similar pattern. The overall test is clean (p=0.123), but two of its individual pieces are flagged (p=0.020 and p=0.047) — a faint asymmetry, even though it passes the headline number.

Gold and HRW are the two clean cases — nothing flagged anywhere, at any level.

Bottom line: none of this is strong enough to rule out K=1 for any of these five series. But WTI, Brent, and SRW each carry a small, real asymmetry in their residuals that's worth being aware of rather than treating as fully resolved.

```python
# 3) Parameter significance

ro.r('''
param_significance <- function(fit, label = "") {
  s <- summary(fit)
  cat(sprintf("--- %s ---\n", label))
  print(s$par)
  cat("\n")
}

# 3) Parameter significance

param_significance(grid_results[["WTI_eGARCH_sstd_1"]]$fit,   "WTI - eGARCH-sstd (K=1)")
param_significance(grid_results[["Brent_eGARCH_sstd_1"]]$fit, "Brent - eGARCH-sstd (K=1)")
param_significance(grid_results[["gold_sGARCH_std_1"]]$fit,   "Gold - sGARCH-std (K=1)")
param_significance(grid_results[["hrw_eGARCH_sstd_1"]]$fit,   "HRW - eGARCH-sstd (K=1)")
param_significance(grid_results[["srw_sGARCH_sstd_1"]]$fit,   "SRW - sGARCH-sstd (K=1)")
''')
```

For each series' model, we checked whether its asymmetry term (alpha2) is real — meaning, is it doing actual work, or is it just noise?

WTI and Brent both come out clearly significant here, with almost the same value (WTI: -0.052, Brent: -0.047). This backs up something we already suspected: both oil benchmarks have a genuine leverage effect — bad news pushes their volatility up more than good news does. It's also why the earlier sign-bias test came out borderline instead of badly failing: the model is catching most of that effect, just not every last bit of it.

HRW also has a significant alpha2, but it's small and points the opposite way — slightly more sensitive to good news than bad. That's an unusual, weak pattern, not a real leverage effect.

Gold and SRW don't have an alpha2 at all — their winning models are symmetric, so there's nothing to test. That fits Gold fine, since nothing else about it hinted at asymmetry either. SRW's a bit murkier: the sign-bias test did pick up a faint signal earlier, even though the model that won overall doesn't have a way to capture it.

Everything else in these models — the other parameters — came out solid and significant across the board, and none of them are anywhere near the edge of instability.

```python
# Testing whether each model absorbed the volatility clustering we found earlier
# with ARCH-LM/Ljung-Box on raw returns. A significant p-value here means the
# model hasn't fully done its job and needs a closer look.

ro.r('''
library(FinTS)

residual_check <- function(fit, returns, label = "") {
  vol <- Volatility(fit)
  std_resid <- returns / vol
  sq_resid  <- std_resid^2

  cat(sprintf("--- %s ---\n", label))

  arch <- ArchTest(std_resid, lags = 10)
  cat("ARCH-LM on standardized residuals:  stat =", round(arch$statistic, 2),
      " p =", format.pval(arch$p.value, digits = 4), "\n")

  bt <- Box.test(sq_resid, lag = 10, type = "Ljung-Box")
  cat("Ljung-Box on squared residuals:     stat =", round(bt$statistic, 2),
      " p =", format.pval(bt$p.value, digits = 4), "\n\n")
}

residual_check(grid_results[["WTI_eGARCH_sstd_1"]]$fit,   wti_ret,   "WTI - eGARCH-sstd (K=1)")
residual_check(grid_results[["Brent_eGARCH_sstd_1"]]$fit, brent_ret, "Brent - eGARCH-sstd (K=1)")
residual_check(grid_results[["gold_sGARCH_std_1"]]$fit,   gold_ret,  "Gold - sGARCH-std (K=1)")
residual_check(grid_results[["hrw_eGARCH_sstd_1"]]$fit,   hrw_ret,   "HRW - eGARCH-sstd (K=1)")
residual_check(grid_results[["srw_sGARCH_sstd_1"]]$fit,   srw_ret,   "SRW - sGARCH-sstd (K=1)")
''')
```

```python
ro.r('''
residual_check(grid_results[["WTI_eGARCH_sstd_2"]]$fit,   wti_ret,   "WTI - eGARCH-sstd (K=2)")
residual_check(grid_results[["Brent_eGARCH_sstd_2"]]$fit, brent_ret, "Brent - eGARCH-sstd (K=2)")
residual_check(grid_results[["hrw_eGARCH_sstd_2"]]$fit,   hrw_ret,   "HRW - eGARCH-sstd (K=2, ruled out)")
residual_check(grid_results[["hrw_sGARCH_sstd_1"]]$fit,   hrw_ret,   "HRW - sGARCH-sstd (K=1, our pick)")
''')
```

WTI: BIC actually preferred K=1 (eGARCH-sstd), but that model left real
volatility clustering unmodeled (ARCH-LM p=0.003, Ljung-Box p=0.001) —
a turbulent day was still predicting turbulent days ahead in a way the
model wasn't capturing. Moving to K=2 cleared it almost perfectly
(p=0.90 on both), so the extra regime earns its keep here even though
BIC alone wouldn't have picked it. Going with eGARCH-sstd, K=2.

Brent: same story as WTI. K=1 won on BIC but failed the clustering
check (ARCH-LM p=0.046, Ljung-Box p=0.031) — a real gap in what the
model was capturing, not just a close BIC call. K=2 fixes it cleanly
(p=0.87 / p=0.86), so we're using eGARCH-sstd, K=2 here too, for the
same reason: a model that actually accounts for the volatility
clustering beats one that just wins a ranking.

```python
# HRW's eGARCH-sstd failed the clustering check at both K=1 and K=2.
# Testing the next-best BIC candidates for HRW (from cell 54's ranking)
# to see if a different variance/distribution spec absorbs the clustering
# that eGARCH-sstd couldn't. All of these are already fit in grid_results
# from cell 53 — no refitting needed.
ro.r('''
cat("=== HRW candidate sweep (ARCH-LM / Ljung-Box on squared residuals) ===\n\n")

residual_check(grid_results[["hrw_eGARCH_std_1"]]$fit,     hrw_ret, "HRW - eGARCH-std   (K=1, BIC rank 2)")
residual_check(grid_results[["hrw_sGARCH_sstd_1"]]$fit,    hrw_ret, "HRW - sGARCH-sstd  (K=1, BIC rank 3)")
residual_check(grid_results[["hrw_gjrGARCH_sstd_1"]]$fit,  hrw_ret, "HRW - gjrGARCH-sstd(K=1, BIC rank 4)")
residual_check(grid_results[["hrw_sGARCH_std_1"]]$fit,     hrw_ret, "HRW - sGARCH-std   (K=1, BIC rank 5)")
residual_check(grid_results[["hrw_eGARCH_ged_1"]]$fit,     hrw_ret, "HRW - eGARCH-ged   (K=1, BIC rank 6)")
residual_check(grid_results[["hrw_gjrGARCH_std_1"]]$fit,   hrw_ret, "HRW - gjrGARCH-std (K=1, BIC rank 7)")
''')
```

HRW: not going with the BIC winner here.
eGARCH-sstd technically has the best BIC (rank 1), but it fails the
volatility-clustering check (ARCH-LM p=0.040, Ljung-Box p=0.039) — and

(p=0.020). That ruled out "just add a regime" as the fix, so the real
issue looked like the eGARCH variance structure itself doesn't suit HRW,
not the regime count.

Walked down the BIC ranking checking each candidate against the same
clustering test, and sGARCH-sstd (BIC rank 3, only ~15 points behind
eGARCH-sstd) is the first one that actually passes (ARCH-LM p=0.088,
Ljung-Box p=0.090). Every non-eGARCH model passed at roughly the same
p-value, and both eGARCH variants failed — so this isn't really a
close call between similarly-good options, it's a clean split between
"eGARCH doesn't fit HRW" and "everything else does."

Going with sGARCH-sstd, K=1: a small, well-justified step down in raw
BIC in exchange for a model that's actually diagnostically sound,
rather than one that wins a ranking but leaves real structure unmodeled.

Brent: same story as WTI. K=1 won on BIC but failed the clustering
check (ARCH-LM p=0.046, Ljung-Box p=0.031) — borderline on BIC terms,
but a real gap in what the model was capturing. K=2 fixes it cleanly
p=0.87 / p=0.86), so we're using eGARCH-sstd, K=2 here too, for the
same reason: a model that actually accounts for the volatility
clustering beats one that just wins a ranking.

```python
# 4) let's check how would the plot be
#
# Only WTI and Brent are 2-regime models under the current diagnostics
# (both needed K=2 to pass the ARCH-LM/Ljung-Box clustering check).
# Gold, HRW, and SRW all settled on K=1 — a single-regime model has no
# "high-volatility regime" to plot a probability for, so they're excluded
# here rather than plotted meaningfully.

ro.r('''
plot_regime_prob <- function(fit, dates, label) {
  state_info  <- State(fit)
  smooth_prob <- state_info$SmoothProb

  if (length(dim(smooth_prob)) == 3) {
    smooth_prob <- smooth_prob[, 1, ]
  }

  n_regimes <- ncol(smooth_prob)

  uncond_vol <- tryCatch(UncVol(fit), error = function(e) NULL)
  if (!is.null(uncond_vol) && length(uncond_vol) == n_regimes) {
    high_regime <- which.max(uncond_vol)
  } else {
    high_regime <- n_regimes
  }

  prob_high <- smooth_prob[, high_regime]
  dates <- as.Date(dates)

  # Defensive alignment in BOTH directions - SmoothProb can come back
  # longer OR shorter than the date vector depending on burn-in handling
  n <- min(length(prob_high), length(dates))
  prob_high <- tail(prob_high, n)
  dates     <- tail(dates, n)

  plot(dates, prob_high, type = "l", col = "darkred", lwd = 1.5,
       ylim = c(0, 1), xlab = "Date", ylab = "P(High-Vol Regime)",
       main = paste(label, "- Smoothed High-Volatility Regime Probability"))
  abline(h = 0.5, lty = 2, col = "gray50")
}
''')
```

```python
ro.r('''
mean_resid_check(grid_results[["hrw_sGARCH_sstd_1"]]$fit, hrw_ret, "HRW - sGARCH-sstd (K=1)")
sign_bias_test(na.omit(as.numeric(hrw_ret / Volatility(grid_results[["hrw_sGARCH_sstd_1"]]$fit))), "HRW residuals (sGARCH-sstd)")
''')
```

HRW: sGARCH-sstd, K=1 — confirmed clean on every diagnostic, not just the clustering test that prompted the switch away from eGARCH. Mean equation passes (p=0.096), ARCH-LM/Ljung-Box on squared residuals both pass (p≈0.088–0.090), and sign-bias is fully clean — the joint test doesn't reject (p=0.467), and all three individual components are comfortably insignificant (p=0.17–0.75), unlike WTI, Brent, and SRW, which all carry at least one flagged component. HRW is the cleanest of the five series once you look past raw BIC.

So: yes to "WTI has a real turbulent state," not yet to "here's exactly how it behaves."

Gold: Passes every check at K=1 — clean mean equation (p=0.295), no asymmetry expected or found (sign-bias joint p=0.451), and no regime-plausibility concern since it's single-regime. sGARCH-std, K=1 is adequate as-is.

SRW: Also settles at K=1 (sGARCH-sstd), and passes — but more narrowly than Gold. Walking down the BIC ranking, sGARCH-sstd (rank 3, only ~15 BIC points behind) is the first candidate that clears the clustering check (p=0.088–0.090) — so HRW's final model is sGARCH-sstd, K=1, a deliberate step down in raw BIC for a model that's actually diagnostically sound. HRW turns out to be the cleanest of the five series once you look past raw BIC.

Per-regime parameters confirm the extra regime is doing real work: alpha2 is significant in both Regime 1 (-0.0543, p=9.5e-10) and Regime 2 (-0.1337, p=0.0023) — resolving the earlier concern that the asymmetric term wasn't pulling its weight despite winning on BIC. One mild soft spot: Regime 2's alpha1 (plain shock term) isn't significant (p=0.220) — that regime's variance leans on persistence and the leverage term rather than raw shock size, minor and not disqualifying.

WTI: Same story and same fix as Brent — K=1 passed mean equation (p=0.538) and was borderline-clean on sign-bias (joint p=0.097, same two components individually flagged, p≈0.03–0.04), but failed the clustering check (p=0.003/0.001).
```python
#regime_diagnostics
ro.r('''
regime_diagnostics <- function(fit, label = "") {
  P <- TransMat(fit)

  durations <- 1 / (1 - diag(P))

  state_fits <- ExtractStateFit(fit)
  unc_vol <- sapply(state_fits, function(sf) sqrt(252) * UncVol(sf))

  sp <- as.data.frame(State(fit, type = "smoothed")$SmoothProb)
  pct_confident <- mean(apply(sp, 1, max) > 0.9)

  vol_ratio       <- max(unc_vol) / min(unc_vol)
  min_persistence <- min(diag(P))

  cat(sprintf("--- %s ---\n", label))
  cat("Transition matrix:\n"); print(round(P, 4))
  cat("Regime durations (days):", round(durations, 1), "\n")
  cat("Unconditional vol (annualized %):", round(unc_vol, 2), "\n")
  cat("Vol ratio:", round(vol_ratio, 2), " | Min persistence:", round(min_persistence, 3),
      " | % confident days:", round(100 * pct_confident, 1), "%\n\n")

  invisible(list(P = P, durations = durations, unc_vol = unc_vol,
                 vol_ratio = vol_ratio, min_persistence = min_persistence,
                 pct_confident = pct_confident))
}

# Only WTI and Brent have a 2-regime structure under the current model set.
# Gold, HRW, and SRW all settled on K=1 — nothing to diagnose for them.
fit_wti   <- grid_results[["WTI_eGARCH_sstd_2"]]$fit
fit_brent <- grid_results[["Brent_eGARCH_sstd_2"]]$fit

regime_diagnostics(fit_wti,   "WTI - eGARCH-sstd (K=2)")
regime_diagnostics(fit_brent, "Brent - eGARCH-sstd (K=2)")

# Gold, HRW, SRW skipped - their winning models are single-regime (K=1)
''')
```


regime_diagnostics()'s UncVol()-based volatility for the high-vol regime
is unreliable whenever that regime's beta is very close to 1 (near the
stationarity edge) or its parameters are imprecisely estimated — both
apply here. UncVol() effectively divides by (1 - beta), so a beta close
to 1 blows the result up into implausible territory (we saw WTI's
regime-2 vol jump to 355%, Brent's to 1006% — neither is realistic for
daily oil returns). This isn't a bug in the model, just a formula that
breaks down near its own boundary condition.

Fix: measure volatility directly from the actual returns on days the
model confidently classifies as each regime, instead of trusting the
theoretical "if this regime ran forever" formula. This gives a real,
empirical number that can't blow up the way a near-unit-root division
can.

```python
# relaized volatility
ro.r('''
n_wti   <- length(wti_ret)
n_brent <- length(brent_ret)

sp_wti   <- as.data.frame(State(fit_wti,   type = "smoothed")$SmoothProb)
sp_brent <- as.data.frame(State(fit_brent, type = "smoothed")$SmoothProb)

sp_wti   <- tail(sp_wti,   n_wti)
sp_brent <- tail(sp_brent, n_brent)

cat("WTI realized vol, regime-1 days (annualized %):",
    round(sd(wti_ret[sp_wti[,1] > 0.9]) * sqrt(252), 2), "\n")
cat("WTI realized vol, regime-2 days (annualized %):",
    round(sd(wti_ret[sp_wti[,2] > 0.9]) * sqrt(252), 2), "\n")

cat("Brent realized vol, regime-1 days (annualized %):",
    round(sd(brent_ret[sp_brent[,1] > 0.9]) * sqrt(252), 2), "\n")
cat("Brent realized vol, regime-2 days (annualized %):",
    round(sd(brent_ret[sp_brent[,2] > 0.9]) * sqrt(252), 2), "\n")
''')
```

Both WTI and Brent show a genuine two-state structure. Using realized volatility on high-confidence regime days (the model-implied unconditional volatility is unreliable for both series' turbulent regimes here — near-unit-root persistence inflates the formula past anything physically plausible), volatility roughly sextuples for WTI (32.8% → 196.5%, ~6.0x) and quintuples for Brent (33.1% → 157.2%, ~4.75x) between calm and turbulent states. As before, the real difference between the two benchmarks is duration, not intensity: Brent's calm regime runs far longer than WTI's (durations: 1,137.7 vs. 359.9 days), and its turbulent bursts last about twice as long too (22 vs. 11.1 days). Confidence in day-to-day classification remains high for both (WTI 97.3%, Brent 99.2%).

```python
#Checks that each regime's fitted variance dynamics are actually stationary,
# not just well-fitting - i.e. that shocks decay over time rather than
# exploding. Persistence formula differs by model type: eGARCH's stationarity
# depends only on the log-variance AR term (beta), while gjrGARCH also folds
# in the ARCH and asymmetry terms, with 0.5 approximating the probability of
# a negative shock under a roughly symmetric innovation distribution.
#
# WTI and Brent are 2-regime (eGARCH); Gold, HRW, and SRW are all single-
# regime now, so this just runs as a one-state sanity check for them.

ro.r('''
stationarity_check <- function(fit, model_name, label = "") {
  state_fits <- ExtractStateFit(fit)

  cat(sprintf("--- %s ---\n", label))
  for (i in seq_along(state_fits)) {
    p <- state_fits[[i]]$par

    alpha1_sum <- sum(p[grepl("^alpha1", names(p))])
    alpha2_sum <- sum(p[grepl("^alpha2", names(p))])
    beta_sum   <- sum(p[grepl("^beta",   names(p))])

    if (model_name == "eGARCH") {
      persistence <- beta_sum
    } else if (model_name == "gjrGARCH") {
      persistence <- alpha1_sum + beta_sum + 0.5 * alpha2_sum
    } else {
      persistence <- alpha1_sum + beta_sum
    }

    flag <- ifelse(abs(persistence) < 1, "stationary (OK)", "NON-STATIONARY - CHECK")
    cat(sprintf("Regime %d | alpha1=%.4f alpha2=%.4f beta=%.4f | persistence=%.4f | %s\n",
                i, alpha1_sum, alpha2_sum, beta_sum, persistence, flag))
  }
  cat("\n")
}

stationarity_check(grid_results[["WTI_eGARCH_sstd_2"]]$fit,   "eGARCH", "WTI - eGARCH-sstd (K=2)")
stationarity_check(grid_results[["Brent_eGARCH_sstd_2"]]$fit, "eGARCH", "Brent - eGARCH-sstd (K=2)")
stationarity_check(grid_results[["gold_sGARCH_std_1"]]$fit,   "sGARCH", "Gold - sGARCH-std (K=1)")
stationarity_check(grid_results[["hrw_sGARCH_sstd_1"]]$fit,   "sGARCH", "HRW - sGARCH-sstd (K=1)")
stationarity_check(grid_results[["srw_sGARCH_sstd_1"]]$fit,   "sGARCH", "SRW - sGARCH-sstd (K=1)")
''')
```

```python
All fitted regimes are formally stationary (persistence 0.983–0.994 across
WTI/Brent's regimes, 0.986–0.990 for Gold/HRW/SRW's single regime) — no
persistence value is genuinely at the non-stationarity boundary this time.
The unconditional-volatility blowups seen for both WTI and Brent's Regime 2
aren't fully explained by persistence alone, since Brent's Regime 2
(0.9933) isn't more extreme than Gold's or HRW's single regimes, which
show no such instability — the more likely driver is the combination of
high persistence with imprecisely-estimated companion parameters
(especially for WTI, where alpha0_2/alpha1_2/alpha2_2 are all individually
insignificant), which the realized-volatility check was specifically
designed to sidestep.
```

```python
%%R
build_episode_table_vol <- function(dates_vec, vol_vec, percentile = 0.90, min_date = NULL) {
  df <- data.frame(date = as.Date(dates_vec), volatility = vol_vec)
  if (!is.null(min_date)) df <- df[df$date >= min_date, ]

  threshold <- quantile(df$volatility, percentile, na.rm = TRUE)
  df$regime <- as.integer(df$volatility > threshold)
  df$change <- c(NA, diff(df$regime))

  starts <- df$date[which(df$change == 1)]
  ends   <- df$date[which(df$change == -1)]
  if (nrow(df) > 0 && df$regime[1] == 1) starts <- c(df$date[1], starts)

  episodes <- data.frame()
  for (i in seq_along(starts)) {
    start <- starts[i]
    future_ends <- ends[ends > start]
    censored <- length(future_ends) == 0
    end <- if (!censored) future_ends[1] else df$date[nrow(df)]
    duration <- as.numeric(end - start) + 1
    episodes <- rbind(episodes, data.frame(episode_id = i, start_date = start,
                                            end_date = end, duration = duration,
                                            censored = censored))
  }
  list(episodes = episodes, threshold = threshold)
}
```

```python
%%R
```
```python
# Finds which regime index is the high-volatility one
# MSGARCH doesn't guarantee regimes are ordered by volatility level.
get_high_vol_idx <- function(fit) {
  state_fits <- ExtractStateFit(fit)
  unc_vol <- sapply(state_fits, function(sf) sqrt(252) * UncVol(sf))
  which.max(unc_vol)
}

# --- WTI and Brent: real 2-regime models, use SmoothProb-based episodes ---
wti_high_idx   <- get_high_vol_idx(grid_results[["WTI_eGARCH_sstd_2"]]$fit)
brent_high_idx <- get_high_vol_idx(grid_results[["Brent_eGARCH_sstd_2"]]$fit)

cat("WTI high-vol regime index:",   wti_high_idx,   "\n")
cat("Brent high-vol regime index:", brent_high_idx, "\n")

wti_valid_mask   <- !is.na(wti$wti_return)
brent_valid_mask <- !is.na(brent$brent_return)
wti_dates   <- as.character(wti$Date[wti_valid_mask])
brent_dates <- as.character(brent$Date[brent_valid_mask])

wti_vol   <- as.numeric(Volatility(grid_results[["WTI_eGARCH_sstd_2"]]$fit))
brent_vol <- as.numeric(Volatility(grid_results[["Brent_eGARCH_sstd_2"]]$fit))

wti_sp   <- as.data.frame(State(grid_results[["WTI_eGARCH_sstd_2"]]$fit,   type = "smoothed")$SmoothProb)
brent_sp <- as.data.frame(State(grid_results[["Brent_eGARCH_sstd_2"]]$fit, type = "smoothed")$SmoothProb)

# Align lengths defensively before indexing (SmoothProb has come back
# off-by-one before, e.g. Brent's realized-vol check earlier)
n_wti   <- length(wti_dates)
n_brent <- length(brent_dates)
wti_sp   <- tail(wti_sp,   n_wti)
brent_sp <- tail(brent_sp, n_brent)

wti_prob_high   <- as.numeric(wti_sp[,   wti_high_idx])
brent_prob_high <- as.numeric(brent_sp[, brent_high_idx])

cat("WTI lengths   -> dates:", length(wti_dates),   " vol:", length(wti_vol),   " prob:", length(wti_prob_high), "\n")
cat("Brent lengths -> dates:", length(brent_dates), " vol:", length(brent_vol), " prob:", length(brent_prob_high), "\n")

# --- Gold and SRW: single-regime (K=1) models — no regime probability
# exists. Only conditional volatility is meaningful; any "episode" for
# these two is a volatility-percentile crossing, not a regime, and should
# be built with build_episode_table_vol() directly on *_vol, not on a
# SmoothProb that doesn't exist in any meaningful form for K=1.
gold_valid_mask <- !is.na(gold$gold_return)
srw_valid_mask  <- !is.na(srw_wheat$srw_return)
gold_dates <- as.character(gold$Date[gold_valid_mask])
srw_dates  <- as.character(srw_wheat$Date[srw_valid_mask])

gold_vol <- as.numeric(Volatility(grid_results[["gold_sGARCH_std_1"]]$fit))
srw_vol  <- as.numeric(Volatility(grid_results[["srw_sGARCH_sstd_1"]]$fit))

cat("Gold lengths -> dates:", length(gold_dates), " vol:", length(gold_vol), "\n")
cat("SRW lengths  -> dates:", length(srw_dates),  " vol:", length(srw_vol),  "\n")
```

```python
import rpy2.robjects as ro
import pandas as pd
import matplotlib.pyplot as plt

def pull_series(r_name):
    return list(ro.r(r_name))

def plot_regime_volatility(dates_r, vol_r, prob_r, series_label, color="darkred"):
    dates = pd.to_datetime(pull_series(dates_r))
    vol   = pull_series(vol_r)
    prob  = pull_series(prob_r)

    if not (len(dates) == len(vol) == len(prob)):
        print(f"WARNING: lengths differ (dates={len(dates)}, vol={len(vol)}, prob={len(prob)}) "
              f"- check the date-alignment step in the R chunk before trusting this plot")

    fig, axes = plt.subplots(2, 1, figsize=(14, 8), sharex=True)

    axes[0].plot(dates, vol, color=color, linewidth=0.8)
    axes[0].fill_between(dates, 0, max(vol), where=[p > 0.5 for p in prob],
                          color="orange", alpha=0.15, label="High-vol regime")
    axes[0].set_title(f"{series_label} Conditional Volatility (MSGARCH)")
    axes[0].set_ylabel("Volatility (Std. Dev.)")
    axes[0].legend(loc="upper right")
    axes[0].grid(True, alpha=0.3)

    axes[1].plot(dates, prob, color="darkblue", linewidth=0.8)
    axes[1].axhline(y=0.5, color="red", linestyle="--", alpha=0.5, label="0.5 threshold")
    axes[1].set_title(f"Probability of High-Volatility Regime — {series_label}")
    axes[1].set_ylabel("Probability")
    axes[1].set_xlabel("Date")
    axes[1].legend(loc="upper right")
    axes[1].grid(True, alpha=0.3)

    plt.tight_layout()
    plt.show()

    
plot_regime_volatility("wti_dates",   "wti_vol",   "wti_prob_high",   "wti")
plot_regime_volatility("brent_dates", "brent_vol", "brent_prob_high", "Brent", color="steelblue")
```

```python
import pandas as pd
import rpy2.robjects as ro

def build_crisis_table(dates_r, prob_r, label, threshold=0.5):
    dates = pd.to_datetime(list(ro.r(dates_r)))
    prob  = list(ro.r(prob_r))

    df = pd.DataFrame({'prob_high_vol': prob}, index=dates)
    df['regime'] = (df['prob_high_vol'] > threshold).astype(int)
    df['change'] = df['regime'].diff()

    starts = df[df['change'] == 1].index
    ends   = df[df['change'] == -1].index

    # edge case: if the series already starts inside a high-vol episode,
    # diff() won't catch that first transition - add it manually
    if df['regime'].iloc[0] == 1:
        starts = pd.Index([df.index[0]]).append(starts)

    episodes = []
    for i, start in enumerate(starts):
        future_ends = ends[ends > start]
        end = future_ends[0] if len(future_ends) > 0 else df.index[-1]
        duration = len(df.loc[start:end])
        episodes.append({
            'Episode #':       i + 1,
            'Start date':      start.strftime('%Y-%m-%d'),
            'End date':        end.strftime('%Y-%m-%d'),
            'Duration (days)': duration
        })

    table = pd.DataFrame(episodes)

    print(f"=== {label}: high-volatility episodes ===")
    print(table.to_string(index=False))
    print(f"\nTotal episodes:   {len(table)}")
    print(f"Average duration: {table['Duration (days)'].mean():.1f} days")
    print(f"Longest episode:  {table['Duration (days)'].max()} days")
    print(f"Shortest episode: {table['Duration (days)'].min()} days\n")

    return table

# Gold and SRW are single-regime (K=1) now — no regime probability exists
# for them, so there's no "episode" in the Markov-switching sense to
# build. Only WTI and Brent have a genuine high-vol regime to extract.
wti_crisis_table   = build_crisis_table("wti_dates",   "wti_prob_high",   "WTI")
brent_crisis_table = build_crisis_table("brent_dates", "brent_prob_high", "Brent")
```

Under the current model set, WTI shows 8 high-volatility episodes across the sample, averaging 15.0 days, ranging from brief 3-day spikes to a 41-day stretch beginning 2020-03-05 — the COVID/oil-price-war crash. Brent shows only 3 episodes, but they're longer on average (26.7 days) and more concentrated around major, identifiable shocks: the same March 2020 crash (34 days), the February 2022 Russian invasion of Ukraine (16 days), and a February–April 2026 episode (30 days).  Both series flag the same two historically major events (COVID 2020, the 2022 invasion) independently, which is a good sign the regime classification is capturing real market stress rather than noise — even though, as noted earlier, the internal parameters of WTI's turbulent regime are less precisely estimated than Brent's.

Regime duration is a direct function of how persistent the fitted transition probabilities are (Hamilton, 1989) — more persistent regimes produce fewer, longer episodes; less persistent regimes flicker between states more often. The smoothed probabilities used to build these episode tables also draw on the full sample rather than only past information (Ardia et al., 2019), which is part of why well-persisted regimes come out as clean, confident blocks rather than noisy day-to-day flips.

This framing now only applies to WTI and Brent — the two series that still have a regime structure to describe. Gold, HRW, and SRW are single-regime (K=1) under the current diagnostics, so there's no transition probability, no smoothed regime classification, and no "episode" in this sense for any of them.

# Now let's check how would XGboost work

```python
# Sanity-checking the 0.5 cutoff we've been using for WTI/Brent's "high-vol
# episode" definition — is 0.5 doing anything special, or would we get a
# totally different story at 0.3 or 0.7? Sweeping the threshold and
# comparing episode counts/durations across the range tells us whether the
# regime-episode results are a real, stable feature of the data or just an
# artifact of where we happened to draw the line.

import pandas as pd
import rpy2.robjects as ro

def build_crisis_table(dates_r, prob_r, label, threshold=0.5, verbose=True):
    dates = pd.to_datetime(list(ro.r(dates_r)))
    prob  = list(ro.r(prob_r))

    df = pd.DataFrame({'prob_high_vol': prob}, index=dates)
    df['regime'] = (df['prob_high_vol'] > threshold).astype(int)
    df['change'] = df['regime'].diff()

    starts = df[df['change'] == 1].index
    ends   = df[df['change'] == -1].index

    if df['regime'].iloc[0] == 1:
        starts = pd.Index([df.index[0]]).append(starts)

    episodes = []
    for i, start in enumerate(starts):
        future_ends = ends[ends > start]
        end = future_ends[0] if len(future_ends) > 0 else df.index[-1]
        duration = len(df.loc[start:end])
        episodes.append({
            'Episode #':       i + 1,
            'Start date':      start.strftime('%Y-%m-%d'),
            'End date':        end.strftime('%Y-%m-%d'),
            'Duration (days)': duration
        })

    table = pd.DataFrame(episodes)

    if verbose:
        print(f"=== {label} (threshold={threshold}) ===")
        if len(table) > 0:
            print(f"Episodes: {len(table)} | Avg duration: {table['Duration (days)'].mean():.1f} "
                  f"| Longest: {table['Duration (days)'].max()} | Shortest: {table['Duration (days)'].min()}\n")
        else:
            print("No episodes found at this threshold.\n")

    return table


def threshold_sensitivity_check(dates_r, prob_r, label, thresholds=[0.3, 0.4, 0.5, 0.6, 0.7]):
    summary_rows = []
    for t in thresholds:
        table = build_crisis_table(dates_r, prob_r, label, threshold=t, verbose=False)
        if len(table) > 0:
            summary_rows.append({
                'Threshold':        t,
                'Num episodes':     len(table),
                'Avg duration':     round(table['Duration (days)'].mean(), 1),
                'Longest episode':  table['Duration (days)'].max(),
                'Shortest episode': table['Duration (days)'].min()
            })
        else:
            summary_rows.append({
                'Threshold': t, 'Num episodes': 0, 'Avg duration': None,
                'Longest episode': None, 'Shortest episode': None
            })

    summary = pd.DataFrame(summary_rows)
    print(f"\n{'='*60}")
    print(f"THRESHOLD SENSITIVITY: {label}")
    print(f"{'='*60}")
    print(summary.to_string(index=False))
    print()
    return summary


# Only WTI and Brent have a regime probability to test — Gold and SRW are
# single-regime (K=1) under the current model set, so there's no
# "threshold" to sweep for them; a probability-threshold sensitivity check
# doesn't apply to a series with no regime probability in the first place.
thresholds_to_test = [0.3, 0.4, 0.5, 0.6, 0.7]

wti_sensitivity   = threshold_sensitivity_check("wti_dates",   "wti_prob_high",   "WTI",   thresholds_to_test)
brent_sensitivity = threshold_sensitivity_check("brent_dates", "brent_prob_high", "Brent", thresholds_to_test)
```

Both series' episode counts are stable across the full 0.3–0.7 threshold range: Brent holds at exactly 3 episodes regardless of cutoff, and WTI stays within a narrow 6–9 episode band. Durations shift only modestly and in the expected direction — stricter thresholds trim a few days off each episode's edges rather than reshaping the overall pattern. This confirms the regime-episode structure for WTI and Brent is a genuine, well-separated feature of the fitted models, not an artifact of the specific 0.5 cutoff used elsewhere in the analysis.
As the number of episodes in most indices with MSGarch are so small that we can run no ML model on them, we switch to one Regime Garch family for predicting the length of episodes. 

As we can see, the number of episods in case of srw changes from 2 to 64 with threshold 60% in Garch.

```python
def crisis_episode_stats(vol_series, percentile):
    threshold = vol_series.quantile(percentile)
    regime = (vol_series > threshold).astype(int)
    change = regime.diff()
    starts = vol_series.index[change == 1]
    ends   = vol_series.index[change == -1]
    if regime.iloc[0] == 1:
        starts = starts.insert(0, vol_series.index[0])

    durations = []
    for start in starts:
        future_ends = ends[ends > start]
        if len(future_ends) == 0:
            continue  # right-censored — true duration unknown, exclude
        end = future_ends[0]
        durations.append(len(vol_series.loc[start:end]))
    return np.array(durations), threshold

percentiles_to_test = [0.70, 0.75, 0.80, 0.85, 0.90, 0.92, 0.95]
sensitivity_rows = []
for p in percentiles_to_test:
    durations, thresh = crisis_episode_stats(master['GARCH_VOL_FULL_SAMPLE'].dropna(), p)
    sensitivity_rows.append({
        'Percentile cutoff':   p,
        'Volatility threshold': round(thresh, 4),
        'Num episodes':        len(durations),
        'Mean duration':       round(durations.mean(), 1),
        'Median duration':     int(np.median(durations)),
        'Min / Max duration':  f"{durations.min()} / {durations.max()}",
    })

sensitivity_df = pd.DataFrame(sensitivity_rows)
print("=== Crisis-definition threshold sensitivity (with duration stats) ===")
print(sensitivity_df.to_string(index=False))
```

```python
# ═══════════════════════════════════════════════════════════════════════
# Comparing crisis-definition methods: percentile (current) vs.
# MAD-robust z-score vs. log-transform + 3σ vs. EVT / peaks-over-threshold
# ═══════════════════════════════════════════════════════════════════════
from scipy.stats import genpareto, skew

vol = master['GARCH_VOL_FULL_SAMPLE'].dropna()

def episodes_from_threshold(vol_series, threshold):
    """Same contiguous-run extraction logic used elsewhere in the notebook,
    generalized to take a raw threshold value instead of a percentile.
    Right-censored episodes (still ongoing at the end of the sample) are
    excluded -- their true duration is unknown, so keeping a truncated
    value would corrupt the duration stats below."""
    regime = (vol_series > threshold).astype(int)
    change = regime.diff()
    starts = vol_series.index[change == 1]
    ends   = vol_series.index[change == -1]
    if regime.iloc[0] == 1:
        starts = starts.insert(0, vol_series.index[0])
    durations = []
    for start in starts:
        future_ends = ends[ends > start]
        if len(future_ends) == 0:
            continue  # right-censored, exclude
        end = future_ends[0]
        durations.append(len(vol_series.loc[start:end]))
    return np.array(durations)

# ── Method: current baseline (90th percentile) ───────────────────────
def threshold_percentile(vol_series, p=0.90):
    return vol_series.quantile(p)

# ── Method 2: MAD-based robust z-score (Iglewicz & Hoaglin, 1993) ────
def threshold_mad_zscore(vol_series, z_cutoff=3.5):
    median = vol_series.median()
    mad = (vol_series - median).abs().median()
    return median + (z_cutoff / 0.6745) * mad

# ── Method 3: log-transform, then classic 3-sigma rule ────────────────
def threshold_log_sigma(vol_series, n_sigma=3):
    log_vol = np.log(vol_series)
    print(f"    Skewness -- raw: {skew(vol_series):.3f} | log: {skew(log_vol):.3f} "
          f"(log should be closer to 0 if this method is well justified)")
    mu, sigma = log_vol.mean(), log_vol.std()
    return np.exp(mu + n_sigma * sigma)

# ── Method 4: EVT / Peaks-Over-Threshold ──────────────────────────────
# Fits a Generalized Pareto Distribution to exceedances over a range of
# candidate thresholds, and picks the threshold where the fitted shape
# parameter is most stable -- an automated stand-in for eyeballing a
# mean-residual-life plot.
def threshold_evt_pot(vol_series, low=0.80, high=0.96, step=0.01, window=3):
    vals = vol_series.values
    candidates = np.quantile(vals, np.arange(low, high, step))
    shapes = []
    for u in candidates:
        exceed = vals[vals > u] - u
        if len(exceed) < 15:
            shapes.append(np.nan)
            continue
        try:
            c, _, _ = genpareto.fit(exceed, floc=0)
            shapes.append(c)
        except Exception:
            shapes.append(np.nan)
    shapes = np.array(shapes)
    rolling_std = pd.Series(shapes).rolling(window, center=True).std().values
    stable_idx = np.nanargmin(rolling_std)
    selected = candidates[stable_idx]
    diagnostics = pd.DataFrame({
        'candidate_threshold': candidates.round(4),
        'gpd_shape_param': np.round(shapes, 3),
        'rolling_std_of_shape': np.round(rolling_std, 4),
    })
    return selected, diagnostics

# ── Run all four and build the comparison table ───────────────────────
def summarize(name, threshold, durations):
    return {
        'Method': name,
        'Threshold': round(threshold, 4),
        'Num episodes': len(durations),
        'Mean duration': round(durations.mean(), 1) if len(durations) else np.nan,
        'Median duration': int(np.median(durations)) if len(durations) else np.nan,
        'Min / Max': f"{durations.min()}/{durations.max()}" if len(durations) else "-",
    }

rows = []

thr = threshold_percentile(vol, 0.90)
rows.append(summarize('Percentile (90th) -- current', thr, episodes_from_threshold(vol, thr)))

print("Method 3 (log + 3sigma) diagnostics:")
thr = threshold_log_sigma(vol, n_sigma=3)
rows.append(summarize('Log-transform + 3sigma', thr, episodes_from_threshold(vol, thr)))

thr = threshold_mad_zscore(vol, z_cutoff=3.5)
rows.append(summarize('MAD robust z-score (|z|>3.5)', thr, episodes_from_threshold(vol, thr)))

thr, evt_diagnostics = threshold_evt_pot(vol)
rows.append(summarize('EVT / Peaks-over-threshold', thr, episodes_from_threshold(vol, thr)))

comparison_df = pd.DataFrame(rows)
print("\n=== Crisis-definition method comparison ===")
print(comparison_df.to_string(index=False))

print("\n=== EVT threshold-stability diagnostics (mean-residual-life-style check) ===")
print(evt_diagnostics.to_string(index=False))
print("\n(Look for the stretch where 'gpd_shape_param' flattens out -- that's the")
print(" region EVT theory says the threshold should sit in; 'rolling_std_of_shape'")
print(" picks the flattest point automatically as the selected threshold above.)")
```

We tested four different ways of defining a "crisis" — a simple percentile cutoff, a robust MAD-based z-score, a log-transformed 3-sigma rule, and an Extreme Value Theory (EVT) approach — and they didn't agree. Episode counts ranged from just 2 (log+3σ) to 87 (EVT), depending entirely on which statistical rule was used, which is a bigger swing than anything else in this analysis.

We ruled out the two extremes. Log+3σ assumes the data is close to symmetric after a log transform, but its own diagnostic showed it still wasn't (skewness of 1.01, not close to 0) — so its very conservative result (just 2 episodes) isn't well supported by its own assumptions. EVT, despite sounding the most rigorous, didn't actually hold up under its own check either: the shape parameter it's supposed to stabilize around never really flattened out across the candidate range, meaning the "optimal" threshold it picked wasn't a genuine plateau, just the smoothest point in an otherwise noisy sequence.

That left the 90th percentile as the most defensible practical choice. It sits between the two extremes, it's simple and transparent — no distributional assumptions to fail — and, importantly, it gives us a workable sample of 46 crisis episodes to train a duration model on. A stricter definition (like the MAD-based 10 episodes) would arguably be more statistically conservative, but leaves too little data for a machine learning model to learn meaningful patterns from. Given the goal is prediction, not just detection, we chose the threshold that balances a reasonable definition of "crisis" against having enough data to actually build a model.

```python
# ── Step 1: derived Brent price series ──────────────────────────────────
# real (inflation-adjusted) Brent price, so the price level isn't confounded
# by inflation trend when it's used as a feature later
cpi_aligned = cpi_daily.reindex(brent_price.index, method='ffill')
brent_real  = (brent_price.squeeze() / cpi_aligned) * 100
brent_real.name = "BRENT_REAL"
print("BRENT_REAL sample:\n", brent_real.head())

# net oil price shock: how far today's price sits above its own trailing
# 12-month high (shift(1) so it's yesterday's high, not today's — no lookahead)
log_price   = np.log(brent_price.squeeze())
rolling_max = log_price.rolling(window=252).max().shift(1)
net_price   = np.maximum(0, log_price - rolling_max)
net_price.name = "NET_OIL_PRICE"
print("NET_OIL_PRICE sample:\n", net_price.head())

```

```python
# ── Step 2: single-regime eGARCH volatility, fit on the FULL history ────
# This deliberately uses a single-regime spec (eGARCH with skewed-t errors),
# matching Brent's own K=1 BIC winner from the MSGARCH ranking (Section 2) —
# not the 2-regime eGARCH-sstd K=2 model used for the WTI/Brent volatility
# diagnostics. Crisis labeling here needs one continuous volatility series
# to threshold, not a regime probability, so K=1 is the right tool for this
# specific job even though K=2 was the better fit for Brent overall.
# NOTE on data leakage: this fit sees the entire sample, future crises
# included. That's fine here because we're only using it to LABEL which
# historical periods count as "crises" (i.e. it defines duration_days,
# our target y — not a feature). This column would NEVER be fed into the
# model as a predictor. So we named it GARCH_VOL_FULL_SAMPLE below instead
# of just GARCH_VOL. The leakage-free version used as an actual feature
# gets refit separately, per-crisis, in Section 6.

brent_ret = np.log(brent_price.squeeze()).diff().dropna() * 100

garch_res = arch_model(brent_ret, mean='Zero', vol='EGARCH', p=1, o=1, q=1,
                        dist='skewt', rescale=True).fit(disp='off')

garch_vol_full_sample = pd.Series(garch_res.conditional_volatility,
                                   index=brent_ret.index,
                                   name='GARCH_VOL_FULL_SAMPLE')
print(garch_vol_full_sample.head())

# ── Step 3: master feature frame ─────────────────────────────────────────
# Only forward-fill here — no bfill. BRENT_REAL and NET_OIL_PRICE both
# depend on rolling windows and have genuine, structural NaNs in their
# first ~252 rows (not enough history yet to compute a trailing 12-month
# high). Backfilling would fabricate values for a period where none
# legitimately exist — leave those rows as NaN and handle them explicitly
# downstream (e.g. exclude from crisis detection) rather than silently
# filling them.
master = pd.DataFrame({
    'OVX': ovx.squeeze(), 'DXY': dxy.squeeze(), 'FSI': fsi_daily,
    'IGREA': igrea_daily, 'GPR': gpr_daily,
    'BRENT_REAL': brent_real, 'NET_OIL_PRICE': net_price,
}).ffill()

master['GARCH_VOL_FULL_SAMPLE'] = garch_vol_full_sample.reindex(master.index).ffill()
master['CRUDE_INVENTORIES'] = crude_inventories_daily.reindex(master.index).ffill()

print("Master frame shape:", master.shape)
print("\nMissing values:\n", master.isna().sum())
print("\nFirst valid NET_OIL_PRICE date:", master['NET_OIL_PRICE'].first_valid_index())
#
```

```python
# ── Rebuild crisis_table from the 90th-percentile definition, explicitly ──

high_vol_threshold = master['GARCH_VOL_FULL_SAMPLE'].quantile(0.90)
master['HIGH_VOL_REGIME'] = (master['GARCH_VOL_FULL_SAMPLE'] > high_vol_threshold).astype(int)
print("90th-percentile threshold:", round(high_vol_threshold, 4))

regime = master['HIGH_VOL_REGIME']
change = regime.diff()
crisis_starts = master.index[change ==  1]
crisis_ends   = master.index[change == -1]

# Edge case: if the series is already in a high-vol regime on day 1,
# .diff() has nothing to compare against and won't register that as a
# start — check explicitly rather than assume it away.
if regime.iloc[0] == 1:
    print("NOTE: HIGH_VOL_REGIME is already 1 on the first day — adding "
          "the sample start as an implicit (and left-censored) crisis start.")
    crisis_starts = pd.Index([master.index[0]]).append(crisis_starts)

crisis_periods = []
for i, start in enumerate(crisis_starts):
    future_ends = crisis_ends[crisis_ends > start]
    censored = len(future_ends) == 0
    end      = future_ends[0] if not censored else master.index[-1]
    duration = len(master.loc[start:end])
    crisis_periods.append({
        'crisis_id':     i + 1,
        'start_date':    start,
        'end_date':      end,
        'duration_days': duration,
        'log_duration':  np.log(duration),
        'censored':      censored
    })

crisis_table_full = pd.DataFrame(crisis_periods)
n_censored = crisis_table_full['censored'].sum()

# Drop right-censored episodes — their true duration is unknown (the
# crisis may still be ongoing past the end of our data), so keeping a
# truncated duration_days value would corrupt the target variable.
crisis_table = crisis_table_full[~crisis_table_full['censored']].drop(columns='censored').reset_index(drop=True)

print(f"Crisis episodes (raw): {len(crisis_table_full)} | Censored (dropped): {n_censored} | "
      f"Final: {len(crisis_table)}")
```

```python
print(crisis_table.to_string(index=False))
print("\nAvg duration:", crisis_table['duration_days'].mean().round(1), "days")
print("Longest:", crisis_table['duration_days'].max(), "days")
print("Shortest:", crisis_table['duration_days'].min(), "days")
```

```python
# ── Build df_model: one row per crisis, causal features only ────────────
# Market features use a 5-day window BEFORE each crisis's start date.
# Volatility feature uses a single-regime eGARCH(1,1) with skewed-t errors
# (Brent's confirmed K=1 BIC winner), refit causally on returns up to
# (not including) each crisis's start — the leakage-free counterpart to
# GARCH_VOL_FULL_SAMPLE, which is full-sample and only used to LABEL
# crises. No MSGARCH/regime-switching here: K=1 avoids both the
# convergence risk and the sample-size loss of refitting a 2-regime
# model on short, early-history windows.

feature_cols_market = ['OVX', 'DXY', 'FSI', 'IGREA', 'GPR', 'BRENT_REAL', 'NET_OIL_PRICE','CRUDE_INVENTORIES']

rows = []
for _, crisis in crisis_table.iterrows():
    start = pd.Timestamp(crisis['start_date'])

    # ── Market features: 5-day mean immediately before crisis start ──
    window = master.loc[:start - pd.Timedelta(days=1)].tail(5)
    feat = {col: window[col].mean() for col in feature_cols_market if col in window.columns}

    # ── Causal single-regime eGARCH refit: data up to (not incl.) start ──
    brent_sub = brent_ret.loc[:start].iloc[:-1]    
    if len(brent_sub) < 100:  # guard: need enough history for a stable fit
        print(f"Crisis #{crisis['crisis_id']} ({start.date()}): only {len(brent_sub)} "
              f"obs available — skipping eGARCH refit, using NaN for GARCH_VOL")
        feat['GARCH_VOL'] = np.nan
    else:
        try:
            garch_roll = arch_model(brent_sub, mean='Zero', vol='EGARCH', p=1, o=1, q=1,
                                     dist='skewt', rescale=True).fit(disp='off')
            feat['GARCH_VOL'] = float(garch_roll.conditional_volatility[-1])
        except Exception as e:
            print(f"Crisis #{crisis['crisis_id']} ({start.date()}): eGARCH refit failed — {e}")
            feat['GARCH_VOL'] = np.nan

    # ── Initial drop: same-day (signed) return on crisis start date ──
    if start in brent_ret.index:
        feat['INITIAL_DROP'] = brent_ret.loc[start]
    else:
        prior = brent_ret.loc[:start]
        feat['INITIAL_DROP'] = prior.iloc[-1] if len(prior) else np.nan

    feat['crisis_id']     = crisis['crisis_id']
    feat['start_date']    = start
    feat['duration_days'] = crisis['duration_days']
    feat['log_duration']  = crisis['log_duration']
    rows.append(feat)

df_model = pd.DataFrame(rows)

meta_cols = ['crisis_id', 'start_date', 'duration_days', 'log_duration']
feature_cols = [c for c in df_model.columns if c not in meta_cols]

n_before = len(df_model)
df_model = df_model.dropna(subset=feature_cols).reset_index(drop=True)
print(f"Dropped {n_before - len(df_model)} row(s) with missing features. {len(df_model)} remain.")

df_model = df_model.set_index('crisis_id')
X     = df_model[feature_cols].values
y     = df_model['duration_days'].values
y_log = df_model['log_duration'].values

print(f"df_model shape: {df_model.shape}")
print(f"Feature columns: {feature_cols}")
print(f"\nMissing values per column:\n{df_model[feature_cols].isna().sum()}")
```

```python
# Crisis #1 (2010-05-20) predates both the 252-day NET_OIL_PRICE window
# and enough history for a stable causal GARCH refit — genuinely missing,
# not fabricatable. Drop rather than impute.
n_before = len(df_model)
df_model = df_model.dropna(subset=feature_cols).reset_index(drop=True)
print(f"Dropped {n_before - len(df_model)} row(s) with missing features. {len(df_model)} remain.")

X     = df_model[feature_cols].values
y     = df_model['duration_days'].values
y_log = df_model['log_duration'].values

assert not np.isnan(X).any(), "NaNs still in X — check df_model before running LOOCV"
print("X is clean — ready for LOOCV.")
```

```python
print(df_model[feature_cols].describe())
```

```python
from scipy.stats import skew
from sklearn.model_selection import TimeSeriesSplit

XGB_PARAMS = dict(
    n_estimators     = 100,
    max_depth        = 3,
    learning_rate    = 0.05,
    subsample        = 0.8,
    colsample_bytree = 0.8,
    reg_alpha        = 0.5,
    reg_lambda       = 1.0,
    random_state     = 42,
    n_jobs           = -1,
    verbosity        = 0
)

def run_timeseries_cv(df_model, X, y, y_log, n_splits=5, xgb_params=XGB_PARAMS):
    """Time-based (forward-chaining) CV: train only on earlier crises, test
    only on later ones. Primary validation method. Note: the earliest chunk
    of crises can never appear in a test fold (nothing precedes them to
    train on) -- they're used for training only, and excluded from the
    reported metrics below, not silently dropped from the dataset."""
    order = np.argsort(df_model['start_date'].values)
    X_sorted, y_sorted, y_log_sorted = X[order], y[order], y_log[order]
    df_sorted = df_model.iloc[order]

    tscv = TimeSeriesSplit(n_splits=n_splits)
    y_true_idx, pred_log, pred_raw = [], [], []

    for train_idx, test_idx in tscv.split(X_sorted):
        model_log = xgb.XGBRegressor(**xgb_params)
        model_log.fit(X_sorted[train_idx], y_log_sorted[train_idx])
        pred_log.extend(model_log.predict(X_sorted[test_idx]))

        model_raw = xgb.XGBRegressor(**xgb_params)
        model_raw.fit(X_sorted[train_idx], y_sorted[train_idx])
        pred_raw.extend(model_raw.predict(X_sorted[test_idx]))

        y_true_idx.extend(test_idx)

    df_eval = df_sorted.iloc[y_true_idx].copy()
    y_eval = y_sorted[y_true_idx]
    pred_log = np.exp(np.array(pred_log))
    pred_raw = np.array(pred_raw)

    n_excluded = len(y_sorted) - len(y_eval)
    print(f"TimeSeriesSplit ({n_splits}): evaluated on {len(y_eval)}/{len(y_sorted)} crises")
    print(f"Earliest {n_excluded} crises never in a test fold (by construction) -- "
          f"first evaluable crisis: {df_eval['start_date'].iloc[0].date()}")

    return df_eval, y_eval, pred_log, pred_raw


def evaluate(y_true, y_pred, label):
    """Print RMSE / MAE / R² / MAPE / SMAPE for one set of predictions.
     MAPE is unstable here -- several crises last
    only 2 days, so a 1-day error is already a 50%+ "error". SMAPE is
    added as a bounded (0-200%), more robust alternative. Treat MAPE as
    diagnostic only, not as a metric to optimize or headline."""
    rmse  = np.sqrt(mean_squared_error(y_true, y_pred))
    mae   = mean_absolute_error(y_true, y_pred)
    r2    = r2_score(y_true, y_pred)
    mape  = np.mean(np.abs((y_true - y_pred) / y_true)) * 100
    smape = np.mean(2 * np.abs(y_pred - y_true) /
                     (np.abs(y_true) + np.abs(y_pred) + 1e-9)) * 100
    print(f"  [{label}]")
    print(f"    RMSE  : {rmse:.1f} days")
    print(f"    MAE   : {mae:.1f} days")
    print(f"    R²    : {r2:.3f}")
    print(f"    MAPE  : {mape:.1f}%  (unstable for short durations -- diagnostic only)")
    print(f"    SMAPE : {smape:.1f}%  (bounded 0-200%, more robust to small y_true)")
    return rmse, mae, r2

def pick_best(y, y_pred_log, y_pred_raw):
    """we decide to use the log-transformed target or the raw data
    based on the distributional skewness of the target, not on the model performance."""
    raw_skew = skew(y)
    log_skew = skew(np.log(y))
    use_log  = abs(log_skew) < abs(raw_skew)

    print("=== Target transform selection (a priori, not leakage-prone) ===")
    print(f"  Skewness -- raw duration: {raw_skew:.3f} | log duration: {log_skew:.3f}")
    print(f"  -> Using {'log' if use_log else 'raw'}-target "
          f"(closer to symmetric; decided before looking at model performance)")

    print("\n=== LOOCV performance (both reported for transparency; only one was chosen) ===")
    _, _, r2_log = evaluate(y, y_pred_log, "XGB — log-target (exp back)")
    _, _, r2_raw = evaluate(y, y_pred_raw, "XGB — raw target")

    if use_log:
        return y_pred_log, "Log-target", True, r2_log
    else:
        return y_pred_raw, "Raw-target", False, r2_raw

# which features were most important to the model? SHAP values give a clear answer.
def plot_shap(model, X, target, feature_cols, save_path=None):
    """Retrain on the full dataset for stable SHAP values, then plot feature importance."""
    model.fit(X, target)

    explainer   = shap.TreeExplainer(model)
    shap_values = explainer.shap_values(X)

    importance = pd.Series(
        np.abs(shap_values).mean(axis=0),
        index=feature_cols
    ).sort_values(ascending=False)

    print("=== SHAP Feature Importance (mean |SHAP|) ===")
    for feat, val in importance.items():
        print(f"  {feat:<25} {val:.4f}")

    fig, axes = plt.subplots(1, 2, figsize=(16, 6))

    importance.sort_values().plot.barh(ax=axes[0], color='steelblue', edgecolor='navy', alpha=0.8)
    axes[0].set_title('SHAP Feature Importance\n(Mean |SHAP value|)')
    axes[0].set_xlabel('Mean |SHAP value|')
    axes[0].grid(axis='x', alpha=0.3)

    plt.sca(axes[1])
    shap.summary_plot(shap_values, X, feature_names=feature_cols, show=False)
    axes[1].set_title('SHAP Summary (dot plot)')

    plt.tight_layout()
    if save_path:
        plt.savefig(save_path, dpi=150, bbox_inches='tight')
    plt.show()
    return shap_values

def report_results(df_model, y, best_pred):
    """Actual vs. predicted duration per crisis, with error and % error columns."""
    results = df_model[['start_date', 'duration_days']].copy()
    results['predicted_days'] = np.round(best_pred).astype(int)
    results['error_days']     = results['predicted_days'] - results['duration_days']
    results['abs_pct_error']  = (np.abs(results['error_days']) / results['duration_days'] * 100).round(1)

    print("=== Crisis Duration Predictions ===")
    print(results.to_string())
    return results

df_model_eval, y_eval, pred_log_eval, pred_raw_eval = run_timeseries_cv(df_model, X, y, y_log)

best_pred, label, used_log, r2 = pick_best(y_eval, pred_log_eval, pred_raw_eval)

plot_predictions(df_model_eval, y_eval, best_pred, r2)

shap_values = plot_shap(xgb.XGBRegressor(**XGB_PARAMS), X, y_log if used_log else y, feature_cols)

results = report_results(df_model_eval, y_eval, best_pred)

```

We picked the log-target back when we decided on the transform rule — before ever looking at how the models would perform, based purely on which version (log or raw) looked more symmetric (raw skewness 3.816 vs. log's 1.335). That's the log-target, evaluated here on the 35 of 45 crises that TimeSeriesSplit can actually test (the earliest 10 are used for training only, never held out, since there's nothing before them to train on).

Once we ran both, log-target came out ahead across every metric — R² (0.207 vs. 0.069), RMSE (19.9 vs. 21.6 days), and MAE (8.4 vs. 12.1 days).

```python
# ── Sanity check before running any baseline that uses df_model + y together ──
assert len(df_model) == len(y), (
    f"df_model ({len(df_model)} rows) and y ({len(y)} rows) are out of sync — "
    f"rebuild y from the current df_model before running baselines"
)
print(f"df_model and y are aligned: {len(df_model)} rows each.")


# compare XGBoost’s performance against the simplest possible predictors
#  and show whether your model adds real value.
def baseline_mean_median_ts(df_model, y, n_splits=5):
    """TimeSeriesSplit 'predict the mean/median of THIS FOLD'S TRAINING
    crises' baseline -- same forward-chaining discipline as the real
    model, so it's a fair, leakage-free comparison point."""
    order = np.argsort(df_model['start_date'].values)
    y_sorted = y[order]

    tscv = TimeSeriesSplit(n_splits=n_splits)
    y_true_idx, pred_mean, pred_median = [], [], []
    for train_idx, test_idx in tscv.split(y_sorted.reshape(-1, 1)):
        pred_mean.extend([y_sorted[train_idx].mean()] * len(test_idx))
        pred_median.extend([np.median(y_sorted[train_idx])] * len(test_idx))
        y_true_idx.extend(test_idx)

    y_eval = y_sorted[y_true_idx]
    pred_mean, pred_median = np.array(pred_mean), np.array(pred_median)

    print(f"TimeSeriesSplit ({n_splits}): evaluated on {len(y_eval)}/{len(y_sorted)} crises")
    print("=== Baseline: mean duration (TimeSeriesSplit) ===")
    evaluate(y_eval, pred_mean, "Naive -- mean duration")
    print("=== Baseline: median duration (TimeSeriesSplit) ===")
    evaluate(y_eval, pred_median, "Naive -- median duration")
    return y_eval, pred_mean, pred_median

# predict each crisis’s duration as the duration of the previous crisis in time, 
# used to compare it against the XGBoost model’s performance. 
# The first crisis has no predecessor and is excluded from evaluation.
def baseline_persistence(df_model, y):
    """Predict each crisis's duration as the duration of the
    chronologically PREVIOUS crisis (a simple persistence/AR(1)-style
    benchmark). The first crisis has no predecessor and is excluded."""
    order = np.argsort(df_model['start_date'].values)
    y_sorted = y[order]
    pred = np.roll(y_sorted, 1)
    valid = np.arange(len(y_sorted)) != 0   # drop the first (no predecessor)
    print("=== Baseline: previous-crisis persistence ===")
    evaluate(y_sorted[valid], pred[valid], "Naive -- persistence")
    return pred, valid

```

```python
# Multicollinearity check for the volatility/stress-proxy features. Tree
# models handle correlated inputs fine for prediction, but SHAP importance
# is known to split credit somewhat arbitrarily among correlated features.

corr_features = [c for c in feature_cols if c in df_model.columns]

corr_matrix = df_model[corr_features].corr()

print("=== Feature correlation matrix ===")
print(corr_matrix.round(2))

fig, ax = plt.subplots(figsize=(7, 6))
im = ax.imshow(corr_matrix, vmin=-1, vmax=1, cmap='RdBu_r')
ax.set_xticks(range(len(corr_matrix)))
ax.set_yticks(range(len(corr_matrix)))
ax.set_xticklabels(corr_matrix.columns, rotation=45, ha='right')
ax.set_yticklabels(corr_matrix.columns)
for i in range(len(corr_matrix)):
    for j in range(len(corr_matrix)):
        val = corr_matrix.iloc[i, j]
        ax.text(j, i, f"{val:.2f}", ha='center', va='center',
                color='white' if abs(val) > 0.5 else 'black', fontsize=8)
plt.colorbar(im, label='Pearson correlation')
plt.title('Feature correlation -- volatility/stress proxies')
plt.tight_layout()
plt.show()

try:
    from statsmodels.stats.outliers_influence import variance_inflation_factor
    X_vif = df_model[corr_features].dropna()
    vif = pd.Series(
        [variance_inflation_factor(X_vif.values, i) for i in range(X_vif.shape[1])],
        index=corr_features
    ).sort_values(ascending=False)
    print("\n=== Variance Inflation Factor ===")
    print(vif.round(2))
except ImportError:
    print("statsmodels not installed -- run `pip install statsmodels` for the VIF check.")
```

VIF values for several features (DXY: 576.82, CRUDE_INVENTORIES: 201.66, GARCH_VOL: 125.73, OVX: 70.62, BRENT_REAL: 53.65, GPR: 40.30) are far above conventional multicollinearity thresholds. Given the small sample size (45 crises, 10 features), this is more likely an artifact of limited data than genuine feature redundancy — VIF estimates become unstable when the number of observations is close to the number of predictors, and grow rapidly as more features are added at this sample size. We do not attempt to trim features based on this statistic alone, since doing so on such a small sample risks discarding genuinely informative predictors. We do, however, treat SHAP-based feature rankings with appropriate caution: correlated features (e.g., BRENT_REAL and DXY, r=0.68; GPR and NET_OIL_PRICE, r=0.67) may split importance credit in ways that don't cleanly separate their individual contributions. INITIAL_DROP (VIF=1.38) remains the one feature clearly outside this collinear cluster.

```python
# ── Multicollinearity trim (iterative VIF elimination) — exploratory only ──
# We're NOT treating this as our primary feature-selection method, given
# VIF's known instability at this sample size (45 crises, 10 features) —
# see the earlier VIF check, where several features showed VIF > 200,
# almost certainly a small-sample artifact rather than genuine redundancy.
# This is run as a secondary check: does trimming to a smaller, less
# collinear feature set change the model's performance meaningfully?
from statsmodels.stats.outliers_influence import variance_inflation_factor

vif_features = list(feature_cols)
dropped_for_vif = []

while True:
    X_vif = df_model[vif_features].dropna()
    vif = pd.Series(
        [variance_inflation_factor(X_vif.values, i) for i in range(X_vif.shape[1])],
        index=vif_features
    )
    worst_feature, worst_vif = vif.idxmax(), vif.max()
    if worst_vif <= 10 or len(vif_features) <= 2:
        break
    print(f"Dropping '{worst_feature}' (VIF={worst_vif:.1f})")
    vif_features.remove(worst_feature)
    dropped_for_vif.append(worst_feature)

print(f"\nFinal VIF after trim:\n{vif.round(2)}")
print(f"Dropped for multicollinearity: {dropped_for_vif}")

feature_cols_trimmed = [c for c in feature_cols if c not in dropped_for_vif]
print(f"Feature count after trim: {len(feature_cols_trimmed)} (was {len(feature_cols)})")

X_trimmed = df_model[feature_cols_trimmed].values
y_trimmed = df_model['duration_days'].values
y_log_trimmed = df_model['log_duration'].values

df_model_trimmed = df_model.copy()  # same crises, just fewer feature columns
df_eval_trimmed, y_eval_trimmed, pred_log_trimmed, pred_raw_trimmed = run_timeseries_cv(
    df_model_trimmed, X_trimmed, y_trimmed, y_log_trimmed
)
best_trimmed, label_trimmed, use_log_trimmed, r2_trimmed = pick_best(y_eval_trimmed, pred_log_trimmed, pred_raw_trimmed)

print(f"\n=== Comparison: full feature set vs. VIF-trimmed ===")
print(f"Full ({len(feature_cols)} features):    R² = {r2:.3f}")
print(f"Trimmed ({len(feature_cols_trimmed)} features): R² = {r2_trimmed:.3f}")
```

Given the severe VIF values found earlier, we also tested a VIF-trimmed feature set (iteratively dropping the highest-VIF feature until all remaining VIFs were below 10), which reduced the feature count from 10 to 5 — dropping DXY, GARCH_VOL, OVX, GPR, and CRUDE_INVENTORIES, keeping FSI, IGREA, BRENT_REAL, NET_OIL_PRICE, and INITIAL_DROP. Under our primary validation method (TimeSeriesSplit, evaluated on the 35 of 45 crises with sufficient trailing history), the trimmed feature set performed modestly better than the full set: R²=0.244 vs. 0.207. As with the full feature set, this uses the pre-registered log-target, decided before looking at performance. With only 35 evaluable crises, we don't treat this difference as decisive — both results are consistent with a model that has limited, but non-zero, predictive signal, and the gap between them is well within what small-sample noise could produce on its own.

```python
# ═══════════════════════════════════════════════════════════════════════
#  Nested feature selection -> honest out-of-fold performance.
#  Same TimeSeriesSplit discipline as the primary model: for each fold,
#  the top-N features are chosen using ONLY that fold's training data --
#  feature selection never sees the crises it's later evaluated on, and
#  it never sees anything from the future relative to the test fold.
# ═══════════════════════════════════════════════════════════════════════
top_n = 4

def nested_shap_feature_selection_ts(df_model, X, y_log, y, feature_names, n_splits=5, top_n=4, xgb_params=XGB_PARAMS):
    order = np.argsort(df_model['start_date'].values)
    X_sorted, y_log_sorted, y_sorted = X[order], y_log[order], y[order]
    df_sorted = df_model.iloc[order]

    tscv = TimeSeriesSplit(n_splits=n_splits)
    y_true_idx, preds_log, selected_per_fold = [], [], []

    for train_idx, test_idx in tscv.split(X_sorted):
        X_train, t_train = X_sorted[train_idx], y_log_sorted[train_idx]

        selector = xgb.XGBRegressor(**xgb_params)
        selector.fit(X_train, t_train)
        sel_importance = pd.Series(
            np.abs(shap.TreeExplainer(selector).shap_values(X_train)).mean(axis=0),
            index=feature_names
        ).sort_values(ascending=False)
        chosen = sel_importance.head(top_n).index.tolist()
        selected_per_fold.append(chosen)
        chosen_idx = [feature_names.index(c) for c in chosen]

        final_model = xgb.XGBRegressor(**xgb_params)
        final_model.fit(X_train[:, chosen_idx], t_train)
        preds_log.extend(final_model.predict(X_sorted[test_idx][:, chosen_idx]))
        y_true_idx.extend(test_idx)

    df_eval = df_sorted.iloc[y_true_idx].copy()
    y_eval = y_sorted[y_true_idx]
    preds = np.exp(np.array(preds_log))

    n_excluded = len(y_sorted) - len(y_eval)
    print(f"Nested TimeSeriesSplit ({n_splits}): evaluated on {len(y_eval)}/{len(y_sorted)} crises "
          f"({n_excluded} earliest never in a test fold, by construction)")

    return df_eval, y_eval, preds, selected_per_fold

df_eval_nested, y_nested, pred_nested, selected_per_fold = nested_shap_feature_selection_ts(
    df_model, X, y_log, y, feature_cols, top_n=top_n
)

print("\n=== Nested feature selection, TimeSeriesSplit (honest, leakage-free) ===")
_, _, r2_nested = evaluate(y_nested, pred_nested, "XGB -- nested (honest) feature selection")

from collections import Counter
n_folds_effective = len(selected_per_fold)
selection_counts = Counter(f for fold in selected_per_fold for f in fold)
print(f"\nFeature selection frequency across {n_folds_effective} outer folds:")
for feat, count in sorted(selection_counts.items(), key=lambda kv: -kv[1]):
    print(f"  {feat:<25} selected in {count}/{n_folds_effective} folds")

results_nested = report_results(df_eval_nested, y_nested, pred_nested)
plot_predictions(df_eval_nested, y_nested, pred_nested, r2_nested)
```

We tested the model's feature selection the honest way: for each fold, the model was only allowed to pick its 4 most important features using that fold's training crises — never the ones it would go on to predict. Evaluated the same way as our primary model (TimeSeriesSplit, on the 35 of 45 crises with enough trailing history), this nested version scores R² = 0.193 — close to, though slightly below, the full 10-feature model's 0.207.

The features it actually relied on were consistent across folds: FSI and INITIAL_DROP were selected in all 5, OVX in 4, GARCH_VOL in 3, with BRENT_REAL and IGREA picked occasionally. That stability across folds is itself informative — the model isn't leaning on a different, arbitrary set of features every time the training data shifts slightly.

```python
from scipy.stats import bootstrap

# Bootstraps the R² statistic by resampling the out-of-fold (y_true, y_pred)
# pairs already produced by the nested TimeSeriesSplit above
# (pred_nested / y_nested). This does NOT re-run feature selection or
# hyperparameter choice per resample -- that would require ~10,000x the
# runtime of the nested procedure above and isn't feasible. This is the
# standard practical compromise: bootstrap the statistic given a fixed
# set of out-of-fold predictions, not the full modeling pipeline.

res = bootstrap(
    (y_nested, pred_nested),
    statistic=lambda yt, yp: r2_score(yt, yp),
    n_resamples=10000,
    method='BCa',
    paired=True,
    random_state=42
)

print(f"Nested TimeSeriesSplit R² = {r2_nested:.3f}")
print(f"Bootstrap 95% BCa CI:    [{res.confidence_interval.low:.3f}, "
      f"{res.confidence_interval.high:.3f}]  (n_resamples=10000)")
```

The bootstrapped 95% confidence interval for the nested TimeSeriesSplit R² is wide and crosses zero ([-0.561, 0.433]), meaning we cannot rule out the possibility that the model's true explanatory power is close to none — or even negative. This reflects the small evaluation sample (35 crises, since TimeSeriesSplit reserves the earliest 10 for training only) and the outsized influence of a few extreme-duration episodes on the R² statistic.

Against naive baselines — now evaluated the same forward-chaining way as the model itself, rather than leave-one-out — the picture is consistent with our earlier validation scheme. Predicting the historical mean duration for every crisis yields R² = -0.041, and predicting the median does slightly worse at R² = -0.087; both are comfortably below zero. Previous-crisis persistence is worse still (R² = -1.198), confirming crisis durations carry no meaningful autocorrelation from one episode to the next. Against this backdrop, the model's nested R² of 0.193 is a real, if modest, improvement over every naive baseline tested. The honest conclusion is unchanged: some signal beyond naive guessing, but the wide, zero-crossing confidence interval means this should be read as suggestive rather than conclusive at this sample size.

The XGBoost model predicts the duration of high-volatility "crisis" episodes in days, using macro-financial covariates and a causal, single-regime GARCH-based volatility feature as inputs.

```python
from sklearn.model_selection import GridSearchCV, KFold, TimeSeriesSplit
```

```python
from statsmodels.stats.outliers_influence import variance_inflation_factor

def nested_vif_trim_ts(df_model, X_full, feature_cols, y, y_log, n_splits=5,
                        vif_threshold=10, xgb_params=XGB_PARAMS):
    """VIF-based feature trimming, nested inside TimeSeriesSplit: for each
    outer fold, VIF elimination runs using ONLY that fold's training data,
    never the test fold. No feature selection step here ever sees a crisis
    before it's actually available."""
    order = np.argsort(df_model['start_date'].values)
    X_sorted, y_sorted, y_log_sorted = X_full[order], y[order], y_log[order]
    df_sorted = df_model.iloc[order]

    tscv = TimeSeriesSplit(n_splits=n_splits)
    y_true_idx, preds_log, selected_per_fold = [], [], []

    for fold_i, (train_idx, test_idx) in enumerate(tscv.split(X_sorted)):
        X_train_df = pd.DataFrame(X_sorted[train_idx], columns=feature_cols)
        vif_features = list(feature_cols)

        # Guard: VIF is unstable/undefined when rows are too close to (or
        # fewer than) the number of features. Fall back to the full set
        # rather than trust a degenerate VIF estimate.
        if X_train_df.shape[0] <= len(vif_features) + 2:
            print(f"Fold {fold_i+1}: only {X_train_df.shape[0]} training rows -- "
                  f"too few for stable VIF, using all {len(vif_features)} features")
        else:
            while len(vif_features) > 2:
                X_vif = X_train_df[vif_features].dropna()
                vif = pd.Series(
                    [variance_inflation_factor(X_vif.values, i) for i in range(X_vif.shape[1])],
                    index=vif_features
                )
                worst_feature, worst_vif = vif.idxmax(), vif.max()
                if worst_vif <= vif_threshold:
                    break
                vif_features.remove(worst_feature)

        selected_per_fold.append(vif_features)
        chosen_idx = [feature_cols.index(c) for c in vif_features]

        model = xgb.XGBRegressor(**xgb_params)
        model.fit(X_sorted[train_idx][:, chosen_idx], y_log_sorted[train_idx])
        preds_log.extend(model.predict(X_sorted[test_idx][:, chosen_idx]))
        y_true_idx.extend(test_idx)

    df_eval = df_sorted.iloc[y_true_idx].copy()
    y_eval = y_sorted[y_true_idx]
    preds = np.exp(np.array(preds_log))

    n_excluded = len(y_sorted) - len(y_eval)
    print(f"\nNested VIF-trim, TimeSeriesSplit ({n_splits}): evaluated on "
          f"{len(y_eval)}/{len(y_sorted)} crises ({n_excluded} earliest never in a test fold)")

    return df_eval, y_eval, preds, selected_per_fold


# ── Run it ────────────────────────────────────────────────────────────
df_eval_vif, y_vif, pred_vif, selected_per_fold_vif = nested_vif_trim_ts(
    df_model, X, feature_cols, y, y_log
)

print("\n=== Nested VIF-trim (honest, leakage-free) ===")
_, _, r2_vif_nested = evaluate(y_vif, pred_vif, "XGB -- nested VIF-trim, TimeSeriesSplit")

print(f"\nFeature sets selected per fold:")
for i, feats in enumerate(selected_per_fold_vif):
    print(f"  Fold {i+1}: {feats}")

print(f"\nComparison:")
print(f"Full feature set (nested SHAP, {top_n}):    R² = {r2_nested:.3f}")
print(f"Nested VIF-trim:                            R² = {r2_vif_nested:.3f}")

results_vif = report_results(df_eval_vif, y_vif, pred_vif)
```

```python
tscv = TimeSeriesSplit(n_splits=5)

y_true_ts, y_pred_ts = [], []
for train_idx, test_idx in tscv.split(X_sorted):
    model_ts = xgb.XGBRegressor(**XGB_PARAMS)
    model_ts.fit(X_sorted[train_idx], y_log_sorted[train_idx])
    preds_log = model_ts.predict(X_sorted[test_idx])
    y_pred_ts.extend(np.exp(preds_log))
    y_true_ts.extend(y_sorted[test_idx])

r2_ts = r2_score(y_true_ts, y_pred_ts)
n_excluded = len(y_sorted) - len(y_true_ts)

print(f"TimeSeriesSplit (5) R² = {r2_ts:.3f}  "
      f"(evaluated on {len(y_true_ts)}/{len(y_sorted)} crises — "
      f"the earliest {n_excluded} are never in a test fold, by construction)")
```

This is our best honest result yet for Brent — R² = 0.255, better than both the full feature set (0.207) and the SHAP-selected version (0.193).

What's nice here is that two different, unrelated ways of picking features basically agree with each other. SHAP picks features by how much they influenced predictions; VIF-trim picks by removing redundancy. Different logic entirely — and yet they both keep landing on FSI, IGREA, and INITIAL_DROP. When two independent methods point at the same features, that's a much stronger sign than either one saying so alone.

Even more specifically: in folds 3 through 5, the exact same 5 features got picked every single time — FSI, IGREA, BRENT_REAL, NET_OIL_PRICE, and INITIAL_DROP. And INITIAL_DROP, the feature we just added, made the cut in every fold that had enough data to even attempt the trim. That's a good sign it's pulling real weight, not just along for the ride.

One honest caveat, not a flaw: fold 1 only had 10 training crises to work with — too few for VIF to give a stable answer, so it fell back to using all 10 features for that fold instead of trimming. So this isn't quite "5 features, tested identically 5 times" — it's 4 folds on a consistent, trimmed set, plus one early fold that had to use everything. Worth mentioning if this number goes in the write-up, same as every other caveat we've been upfront about

```python
# ── Final hyperparameter search: ONE set of values, for future use ──
# This answers a different question than the nested test above: "what
# values should the model use going forward", not "how well does tuning
# perform out-of-sample". Because it's a single search (not nested), its
# own score is NOT an honest performance estimate — it's just how to
# pick values. We validate those values honestly in the next step.

final_search_cv = KFold(n_splits=5, shuffle=True, random_state=42)

final_search = GridSearchCV(
    xgb.XGBRegressor(random_state=42, n_jobs=-1, verbosity=0),
    param_grid=XGB_PARAM_GRID,
    cv=final_search_cv,
    scoring='neg_mean_squared_error',
    n_jobs=-1
)
final_search.fit(X_trimmed, y_log_trimmed)

print("=== Final hyperparameter search (single search, VIF-trimmed features) ===")
print(f"Best params found: {final_search.best_params_}")

XGB_PARAMS_TUNED = dict(final_search.best_params_, random_state=42, n_jobs=-1, verbosity=0)

# ── Validate these values honestly, same method as everything else ──
df_eval_final, y_final, pred_log_final, pred_raw_final = run_timeseries_cv(
    df_model, X_trimmed, y_trimmed, y_log_trimmed, xgb_params=XGB_PARAMS_TUNED
)
_, _, r2_final_tuned = evaluate(y_final, pred_log_final, "XGB -- final tuned params, honest check")

print(f"\n=== Does the found combination actually hold up? ===")
print(f"Default XGB_PARAMS (nested VIF-trim):  R² = {r2_vif_nested:.3f}")
print(f"Searched params (honest re-check):     R² = {r2_final_tuned:.3f}")

```

```python
# ── Seed sensitivity check: is 0.358 stable, or did we get lucky? ──────
seeds_to_test = [7, 42, 123, 2024, 99]
seed_results = []

for seed in seeds_to_test:
    search_cv = KFold(n_splits=5, shuffle=True, random_state=seed)
    search = GridSearchCV(
        xgb.XGBRegressor(random_state=seed, n_jobs=-1, verbosity=0),
        param_grid=XGB_PARAM_GRID,
        cv=search_cv,
        scoring='neg_mean_squared_error',
        n_jobs=-1
    )
    search.fit(X_trimmed, y_log_trimmed)

    params_seed = dict(search.best_params_, random_state=42, n_jobs=-1, verbosity=0)
    # NOTE: model's own random_state stays fixed at 42 throughout -- only
    # the search's internal CV shuffling changes with `seed`, so this
    # isolates "did the search find different params" from "does the
    # model behave differently."

    df_eval_seed, y_seed, pred_log_seed, _ = run_timeseries_cv(
        df_model, X_trimmed, y_trimmed, y_log_trimmed, xgb_params=params_seed
    )
    _, _, r2_seed = evaluate(y_seed, pred_log_seed, f"seed={seed}")

    seed_results.append({'seed': seed, 'params': search.best_params_, 'r2_honest': r2_seed})

print("\n=== Seed sensitivity summary ===")
seed_df = pd.DataFrame(seed_results)
print(seed_df.to_string(index=False))
print(f"\nMean R² across seeds: {seed_df['r2_honest'].mean():.3f}")
print(f"Std  R² across seeds: {seed_df['r2_honest'].std():.3f}")
print(f"Range: [{seed_df['r2_honest'].min():.3f}, {seed_df['r2_honest'].max():.3f}]")
```

Since we'd deleted the aggressive per-fold tuning approach after finding it collapsed performance to near zero, we tested a more conservative alternative instead: a single hyperparameter search across the full VIF-trimmed dataset, then validated honestly under our primary TimeSeriesSplit method. This found real, meaningful improvement — but because a single search's result depends on how its internal cross-validation happens to shuffle the data, we re-ran it across 5 different random seeds to check whether the improvement was genuine or a lucky draw. Every seed produced a positive R², ranging from 0.210 to 0.442 (mean 0.328), compared to 0.255 with default hyperparameters and the same trimmed feature set. learning_rate (0.1) and colsample_bytree (0.8) were selected consistently across seeds, while max_depth and n_estimators varied more — likely explaining most of the spread in the final R². We report the mean and range rather than any single seed's result, since presenting one number in isolation would understate real uncertainty in a search of this kind at this sample size. Taken together, this is our strongest result: unlike per-fold tuning, a single global search — evaluated honestly and checked across multiple seeds — produces a real, if moderately variable, improvement over default hyperparameters.

```python
from scipy.stats import bootstrap

# Bootstraps R² by resampling the (y_true, y_pred) pairs from our final,
# locked model (VIF-trimmed features + seed=42's searched hyperparameters,
# validated via TimeSeriesSplit). Same practical compromise as our other
# bootstrap checks: resamples the fixed predictions, doesn't re-run
# feature selection or search per resample.

res_final = bootstrap(
    (y_final, pred_log_final),
    statistic=lambda yt, yp: r2_score(yt, yp),
    n_resamples=10000,
    method='BCa',
    paired=True,
    random_state=42
)

print(f"Final model R² = {r2_final_tuned:.3f}")
print(f"Bootstrap 95% BCa CI: [{res_final.confidence_interval.low:.3f}, "
      f"{res_final.confidence_interval.high:.3f}]  (n_resamples=10000)")
```

```python
from scipy.stats import wilcoxon

def naive_mean_baseline_ts(df_model, y, n_splits=5):
    """Time-respecting naive baseline: predict the mean duration of that
    fold's training crises. Same evaluation scheme as our real model, so
    the comparison is fair."""
    order = np.argsort(df_model['start_date'].values)
    y_sorted = y[order]

    tscv = TimeSeriesSplit(n_splits=n_splits)
    y_true_idx, preds = [], []
    for train_idx, test_idx in tscv.split(y_sorted.reshape(-1, 1)):
        preds.extend([y_sorted[train_idx].mean()] * len(test_idx))
        y_true_idx.extend(test_idx)

    return y_sorted[y_true_idx], np.array(preds)

y_naive, pred_naive = naive_mean_baseline_ts(df_model, y)

model_errors = np.abs(y_final - pred_log_final)
naive_errors = np.abs(y_naive - pred_naive)

print(f"Model MAE:  {model_errors.mean():.2f} days")
print(f"Naive MAE:  {naive_errors.mean():.2f} days")

stat, p_value = wilcoxon(model_errors, naive_errors, alternative='less')
print(f"\nWilcoxon signed-rank test (model error < naive error): "
      f"statistic={stat:.1f}, p={p_value:.4f}")
```

The bootstrap CI on R² is wide and crosses zero ([-0.428, 0.629]) — so we can't pin down precisely how much variance the model explains, and that's worth being upfront about rather than leaning on the 0.358 headline number alone.

But the Wilcoxon test tells a clearer story. Comparing the model's error against just guessing the average duration, crisis by crisis, the model wins decisively (p=0.0003) — MAE of 7.8 days vs. 12.5 days for the naive guess. This test isn't thrown off by a few extreme outliers the way R² is, so it's a more reliable signal that the model is genuinely doing something useful on typical crises, even if we can't say exactly how much.

#### Before fitting any models, we decided whether to predict raw crisis duration or its log-transform based only on the shape of the target variable itself — duration data tends to be right-skewed, and the log transform is a standard way to handle that. This decision was made without looking at model performance, so it couldn't have been chosen simply because it happened to score better.

### Threshold definition:
#### Crisis episodes were defined using a fixed 90th-percentile volatility cutoff, chosen before any modeling took place. A sensitivity check across nearby cutoffs (80th–95th percentile) was later run to confirm the resulting number of episodes stays reasonably stable — not to search for a threshold that improved model results.

```python
# ═══════════════════════════════════════════════════════════════════════
# CONFIRMATORY HOLDOUT (guards against model-variant selection leakage)
#
# Every comparison so far (fixed vs. tuned params, full vs. VIF-trimmed
# features, different CV strategies) was evaluated on the same 45 crises —
# even with proper nested CV, repeatedly comparing variants against the
# same data carries a mild risk of favoring whatever looked best by chance.
#
# This cell locks away the most recent crises BEFORE doing anything, and
# re-derives VIF-based feature selection using ONLY the remaining
# (development) crises. Hyperparameters use the fixed XGB_PARAMS defaults,
# not a fresh search . XGB_PARAMS was also decided before any tuning work began
# , so using it here can't leak anything from the holdout crises the way 
# a fresh dev-only search indirectly could. The holdout is predicted exactly once.
# ═══════════════════════════════════════════════════════════════════════
from statsmodels.stats.outliers_influence import variance_inflation_factor

N_HOLDOUT = 6   # most recent crises, reserved -- chronological, not random

df_sorted = df_model.sort_values('start_date').reset_index(drop=True)
holdout_dates = df_sorted['start_date'].iloc[-N_HOLDOUT:].values
holdout_mask  = df_model['start_date'].isin(holdout_dates)
dev_mask      = ~holdout_mask

print(f"Development set: {dev_mask.sum()} crises | Holdout: {holdout_mask.sum()} crises")
print("Holdout crises (untouched until this cell):")
print(df_model.loc[holdout_mask, ['start_date', 'duration_days']].to_string())

X_dev     = df_model.loc[dev_mask, feature_cols].values
y_dev     = df_model.loc[dev_mask, 'duration_days'].values
y_dev_log = df_model.loc[dev_mask, 'log_duration'].values
X_holdout = df_model.loc[holdout_mask, feature_cols].values
y_holdout = df_model.loc[holdout_mask, 'duration_days'].values

# 1) VIF-based feature selection using DEV data only
dev_df = pd.DataFrame(X_dev, columns=feature_cols)
vif_features_holdout = list(feature_cols)
while len(vif_features_holdout) > 2:
    X_vif = dev_df[vif_features_holdout].dropna()
    vif = pd.Series(
        [variance_inflation_factor(X_vif.values, i) for i in range(X_vif.shape[1])],
        index=vif_features_holdout
    )
    worst_feature, worst_vif = vif.idxmax(), vif.max()
    if worst_vif <= 10:
        break
    print(f"Dropping '{worst_feature}' (VIF={worst_vif:.1f})")
    vif_features_holdout.remove(worst_feature)

holdout_features = vif_features_holdout
print("\nFeatures selected using DEV set only (VIF-trim):", holdout_features)

dev_idx    = [feature_cols.index(c) for c in holdout_features]
X_dev_sel  = X_dev[:, dev_idx]
X_hold_sel = X_holdout[:, dev_idx]

# 2) Fit on ALL dev data using fixed XGB_PARAMS, predict the
#    never-touched holdout -- once
final_model = xgb.XGBRegressor(**XGB_PARAMS)
final_model.fit(X_dev_sel, y_dev_log)
holdout_pred = np.exp(final_model.predict(X_hold_sel))

print("\n=== Confirmatory holdout result (evaluated ONCE) ===")
_, _, r2_holdout = evaluate(y_holdout, holdout_pred, "XGB -- confirmatory holdout")
print(f"\nNote: R2 on only {N_HOLDOUT} points is a noisy estimate -- "
      f"treat this as a generalization sanity check, not a precise metric.")

holdout_results = pd.DataFrame({
    'start_date':     df_model.loc[holdout_mask, 'start_date'].values,
    'actual_days':    y_holdout,
    'predicted_days': np.round(holdout_pred).astype(int),
})
print("\n", holdout_results.to_string(index=False))
```

On the 6 fully held-out crises — never touched by feature selection or model fitting — the model predicted with MAE=1.6 days and RMSE=2.2 days, using default hyperparameters rather than a fresh search on the smaller dev set. That's actually slightly tighter than our earlier tuned version of this same check, supporting the decision to drop the dev-only search. VIF-trim also landed on the same 5 features here as it did on the full dataset (FSI, IGREA, BRENT_REAL, NET_OIL_PRICE, INITIAL_DROP), so the feature set isn't dependent on these particular crises being included. R²=0.024 looks low, but these 6 durations are all short and close together, so there's very little variance for R² to explain in the first place — the absolute errors are the more meaningful number here. The one real miss was 2025-06-25 (8 actual vs. 3 predicted), the model's largest error in this batch,which might show that the model still lean toward under-predicting longer episodes even at this small scale.

```python
# Cheap robustness check: does the nested VIF-trim result depend on the 3
# episodes that start within 5 days of another episode? Drops the later
# episode in each close pair from the already-built feature matrix and
# reruns the same nested VIF-trim procedure used for the headline number.
# (A full merge-into-one-episode declustering would require rebuilding
# features from scratch for the new episode boundaries -- this is the
# cheaper "does it still hold without these 3 rows" version.)

gaps = df_model.sort_values('start_date')['start_date'].diff().dt.days
close_start_dates = df_model.sort_values('start_date').loc[gaps <= 5, 'start_date']
keep_mask = ~df_model['start_date'].isin(close_start_dates)

df_model_declustered = df_model.loc[keep_mask].copy()
X_declustered     = df_model.loc[keep_mask, feature_cols].values
y_declustered     = df_model.loc[keep_mask, 'duration_days'].values
y_declustered_log = df_model.loc[keep_mask, 'log_duration'].values

df_eval_decl, y_decl, pred_decl, _ = nested_vif_trim_ts(
    df_model_declustered, X_declustered, feature_cols, y_declustered, y_declustered_log
)

_, _, r2_declustered = evaluate(y_decl, pred_decl, "XGB -- nested VIF-trim, clustered pairs de-duplicated")

print(f"\nWith clustering as-is       (N={len(df_model)}):     R² = {r2_vif_nested:.3f}")
print(f"With clustered pairs dropped (N={len(df_model_declustered)}): R² = {r2_declustered:.3f}")
```

Dropping the 3 crisis pairs that started within 5 days of each other brings R² down slightly, from 0.255 to 0.232. That's a small dip, not a collapse — so the result doesn't depend on those pairs being there, but it was getting a modest boost from them. Some of this drop is likely just from having 3 fewer data points to work with in an already small sample (one early fold now only has 7 training crises), not necessarily the pairs themselves being special. Overall: the model holds up reasonably well without them, but it's fair to call this a mild sensitivity rather than a fully clean result.

```python
# Permutation test: is the model's real R² actually meaningful, or could
# a model like this reach a similar score on pure noise?
#
# We randomly scramble which duration belongs to which crisis, 200
# different ways, destroying any real relationship between the features
# and the target on purpose. Then we run the exact same nested VIF-trim
# procedure on each scrambled version and record its R².
#
# This builds a "no real signal exists" baseline. If our actual,
# unshuffled R² clearly beats almost all 200 of these random attempts,
# that's strong evidence the real result reflects a genuine pattern in
# the data -- not just a flexible model finding structure in noise.

n_permutations = 200
rng = np.random.default_rng(42)
null_r2 = []

for _ in range(n_permutations):
    y_shuffled = rng.permutation(y)           # break the true X-y link
    y_shuffled_log = np.log(y_shuffled)
    df_model_shuffled = df_model.copy()
    df_model_shuffled['duration_days'] = y_shuffled
    df_model_shuffled['log_duration']  = y_shuffled_log

    _, y_eval_shuf, pred_shuf, _ = nested_vif_trim_ts(
        df_model_shuffled, X, feature_cols, y_shuffled, y_shuffled_log
    )
    null_r2.append(r2_score(y_eval_shuf, pred_shuf))

null_r2 = np.array(null_r2)
print(f"Real R² (nested VIF-trim): {r2_vif_nested:.3f}")
print(f"Null distribution (shuffled y): mean={null_r2.mean():.3f}, "
      f"95th percentile={np.percentile(null_r2, 95):.3f}")
print(f"Real R² beats {np.mean(null_r2 < r2_vif_nested)*100:.1f}% of random permutations")
```

```python
p_value = (np.sum(null_r2 >= r2_vif_nested) + 1) / (n_permutations + 1)
print(f"Exact permutation p-value: p = {p_value:.4f} (Phipson & Smyth, 2010 correction)")
```

We ran a permutation test to check whether R²=0.255 reflects a real relationship or could happen by chance: scrambled the crisis durations 200 different ways, reran the exact same model on each scrambled version, and compared. Our real result beat all 200 random attempts — even the best-case random score only reached -0.013, nowhere near our actual 0.255 (p < 0.005). This is strong, formal evidence that the model is picking up on something genuine in the data, not just fitting noise.

Summary of model validation

The final model (XGBoost, log-duration target, VIF-trimmed feature set — FSI, IGREA, BRENT_REAL, NET_OIL_PRICE, INITIAL_DROP — with hyperparameters from a single global search, seed=42) achieves R² = 0.358 under TimeSeriesSplit, evaluated on the 35 of 45 crises with sufficient trailing history.

Bootstrap confidence interval: [-0.428, 0.629] — wide, and crosses zero, reflecting the small evaluation sample and this result's added sensitivity to which hyperparameters the search happens to find.
Naive baselines: mean duration (R²=-0.060), median duration (R²=-0.072), and previous-crisis persistence (R²=-1.142) — all score below zero; the model clearly and significantly outperforms all three (Wilcoxon signed-rank, p=0.0003 vs. the mean baseline).
Hyperparameter search stability: re-running the search across 5 seeds gave R² ranging 0.210–0.442 (mean 0.328) — real, consistently positive, but not precisely pinned down. Separately, an aggressive per-fold search (re-tuning inside every outer fold) collapsed to R²=0.018, confirming that a single global search is more reliable than per-fold tuning at this sample size.
Confirmatory holdout: the 6 most recent crises, held out from all feature and hyperparameter decisions, were predicted with MAE=1.6 days, RMSE=2.2 days using default hyperparameters and dev-only VIF-trim — tight, sensible errors, though this batch happened to contain only short, typical-severity crises and offers no evidence about extreme-event forecasting.
De-clustering check (on the untuned VIF-trim model, R²=0.255): removing 3 crises with overlapping feature windows brought R² down slightly to 0.232 — a mild sensitivity, not a dependency.
Permutation test (also on the untuned VIF-trim model): the real R² beat 100% of 200 label-shuffled permutations (exact p=0.0050, Phipson & Smyth corrected) — strong, formal evidence the result reflects a genuine pattern, not noise.

Taken together, these checks support a consistent conclusion: the model captures a real, statistically distinguishable signal that reliably beats naive guessing and passes a formal significance test. The precise size of that signal is less certain — the tuned model's R² varies with search seed, and its bootstrap interval is wide — so 0.358 should be read as a reasonable point estimate rather than a precise figure. Note that our two most rigorous significance checks (permutation test, de-clustering) were run on the untuned VIF-trim model (R²=0.255) rather than the seed-tuned final model; both models tell the same qualitative story, but this distinction is worth keeping in mind when citing exact figures.

# the AFT and ridge and lasso need to be added afterwards 


```python
# ── XGBoost AFT, same nested discipline as the plain-regression model ───
# AFT uses the low-level xgb.train()/DMatrix API, not XGBRegressor, since
# that's what exposes label_lower_bound / label_upper_bound. Same rule as
# before: feature selection is refit per fold on training data only.

AFT_PARAMS = {
    'objective':                    'survival:aft',
    'eval_metric':                  'aft-nloglik',
    'aft_loss_distribution':        'normal',
    'aft_loss_distribution_scale':  1.0,
    'max_depth':                    3,
    'learning_rate':                0.05,
    'subsample':                    0.8,
    'colsample_bytree':             0.8,
    'reg_alpha':                    0.5,
    'reg_lambda':                   1.0,
}
AFT_ROUNDS = 100

def fit_aft(X_train, lower_train, upper_train, params=AFT_PARAMS, num_boost_round=AFT_ROUNDS):
    dtrain = xgb.DMatrix(X_train)
    dtrain.set_float_info('label_lower_bound', lower_train)
    dtrain.set_float_info('label_upper_bound', upper_train)
    return xgb.train(params, dtrain, num_boost_round=num_boost_round)

def nested_shap_feature_selection_loocv_aft(X, lower, upper, feature_names, top_n=4):
    n = len(X)
    preds = np.full(n, np.nan)
    selected_per_fold = []

    for i in range(n):
        train_idx = np.delete(np.arange(n), i)
        X_train = X[train_idx]
        lower_train, upper_train = lower[train_idx], upper[train_idx]

        selector = fit_aft(X_train, lower_train, upper_train)
        # SHAP values here are on the model's internal log-time scale --
        # fine for ranking feature importance, not day-additive like the
        # plain-regression model's SHAP values.
        importance = pd.Series(
            np.abs(shap.TreeExplainer(selector).shap_values(X_train)).mean(axis=0),
            index=feature_names
        ).sort_values(ascending=False)
        chosen = importance.head(top_n).index.tolist()
        selected_per_fold.append(chosen)
        chosen_idx = [feature_names.index(c) for c in chosen]

        final_model = fit_aft(X_train[:, chosen_idx], lower_train, upper_train)
        dtest = xgb.DMatrix(X[i:i+1, chosen_idx])
        preds[i] = final_model.predict(dtest)[0]   # already back on original time scale

    return preds, selected_per_fold

X_full_aft = df_model[feature_cols].values   # same candidate pool as the plain-regression model
pred_aft, selected_per_fold_aft = nested_shap_feature_selection_loocv_aft(
    X_full_aft, y_lower, y_upper, feature_cols, top_n=top_n
)

print("\n=== XGBoost AFT, nested LOOCV with per-fold feature selection ===")
_, _, r2_aft = evaluate(y, pred_aft, "XGB-AFT -- nested feature selection")

print(f"\nPlain regression (same discipline): R² = {r2_nested:.3f}")
print(f"AFT (same discipline):              R² = {r2_aft:.3f}")
```

We also tested XGBoost's survival:aft objective — modeling crisis duration directly as a time-to-event outcome, using the same nested per-fold feature selection discipline as the plain-regression model. AFT outperformed plain regression on every metric: R² (0.156 vs. 0.112), MAE (8.1 vs. 8.7 days), and RMSE (20.2 vs. 20.7 days). This makes some conceptual sense — AFT's loss function is purpose-built for duration data, rather than treating log-duration as an arbitrary regression target. With zero censored observations in this dataset, AFT's other key advantage (correctly handling episodes with unknown true duration) isn't yet being exercised here; the improvement appears to come from the distributional assumption fitting this data's shape reasonably well. Given the modest sample size, we treat this as a promising but not yet decisive result, worth confirming with the same bootstrap and permutation-test checks applied to the plain-regression model before treating AFT as the final, preferred approach.

```python
# Paired comparison: plain regression vs. AFT, same 45 crises, same
# nested-feature-selection discipline. Two-sided since neither model has
# a directional prior over the other -- we're testing "are these
# meaningfully different," not "is one better."

abs_err_plain = np.abs(y - pred_nested)
abs_err_aft   = np.abs(y - pred_aft)

stat, p_value = wilcoxon(abs_err_plain, abs_err_aft, alternative='two-sided')

print(f"Wilcoxon signed-rank test, plain regression vs. AFT, n={len(y)}:")
print(f"  W = {stat:.1f}, two-sided p = {p_value:.4f}")
print(f"  Mean |error| -- Plain regression: {abs_err_plain.mean():.2f} days | "
      f"AFT: {abs_err_aft.mean():.2f} days")
print(f"  Median |error| -- Plain regression: {np.median(abs_err_plain):.2f} days | "
      f"AFT: {np.median(abs_err_aft):.2f} days")
```

A paired Wilcoxon signed-rank test comparing AFT's and plain regression's per-crisis absolute errors found no statistically significant difference (p=0.153), despite AFT scoring numerically better on every metric (mean error 8.05 vs. 8.69 days, median 2.17 vs. 2.54 days, R² 0.156 vs. 0.112). This is consistent with the wide bootstrap confidence interval already found for the plain-regression model — at 45 crises, differences of this size are within the range expected from sampling variability alone. We report AFT's result as a promising direction, conceptually well-suited to duration data, but do not claim it is definitively superior to plain regression given the current evidence.

```python
# Sanity-checking the 0.5 cutoff we've been using for WTI's "high-vol
# episode" definition — is 0.5 doing anything special, or would we get a
# totally different story at 0.3 or 0.7? Sweeping the threshold and
# comparing episode counts/durations across the range tells us whether the
# regime-episode results are a real, stable feature of the data or just an
# artifact of where we happened to draw the line.

import pandas as pd
import rpy2.robjects as ro

def crisis_table_wti(dates_r, prob_r, label, threshold=0.5, verbose=True):
    dates = pd.to_datetime(list(ro.r(dates_r)))
    prob  = list(ro.r(prob_r))

    df_model_wti = pd.DataFrame({'prob_high_vol': prob}, index=dates)
    df_model_wti['regime'] = (df_model_wti['prob_high_vol'] > threshold).astype(int)
    df_model_wti['change'] = df_model_wti['regime'].diff()

    starts = df_model_wti[df_model_wti['change'] == 1].index
    ends   = df_model_wti[df_model_wti['change'] == -1].index

    if df_model_wti['regime'].iloc[0] == 1:
        starts = pd.Index([df_model_wti.index[0]]).append(starts)

    episodes = []
    for i, start in enumerate(starts):
        future_ends = ends[ends > start]
        end = future_ends[0] if len(future_ends) > 0 else df_model_wti.index[-1]
        duration = len(df_model_wti.loc[start:end])
        episodes.append({
            'Episode #':       i + 1,
            'Start date':      start.strftime('%Y-%m-%d'),
            'End date':        end.strftime('%Y-%m-%d'),
            'Duration (days)': duration
        })

    table = pd.DataFrame(episodes)

    if verbose:
        print(f"=== {label} (threshold={threshold}) ===")
        if len(table) > 0:
            print(f"Episodes: {len(table)} | Avg duration: {table['Duration (days)'].mean():.1f} "
                  f"| Longest: {table['Duration (days)'].max()} | Shortest: {table['Duration (days)'].min()}\n")
        else:
            print("No episodes found at this threshold.\n")

    return table


def threshold_sensitivity_check(dates_r, prob_r, label, thresholds=[0.3, 0.4, 0.5, 0.6, 0.7]):
    summary_rows = []
    for t in thresholds:
        table = crisis_table_wti(dates_r, prob_r, label, threshold=t, verbose=False)
        if len(table) > 0:
            summary_rows.append({
                'Threshold':        t,
                'Num episodes':     len(table),
                'Avg duration':     round(table['Duration (days)'].mean(), 1),
                'Longest episode':  table['Duration (days)'].max(),
                'Shortest episode': table['Duration (days)'].min()
            })
        else:
            summary_rows.append({
                'Threshold': t, 'Num episodes': 0, 'Avg duration': None,
                'Longest episode': None, 'Shortest episode': None
            })

    summary_wti = pd.DataFrame(summary_rows)
    print(f"\n{'='*60}")
    print(f"THRESHOLD SENSITIVITY: {label}")
    print(f"{'='*60}")
    print(summary_wti.to_string(index=False))
    print()
    return summary_wti


# Only WTI and Brent have a regime probability to test — Gold and SRW are
# single-regime (K=1) under the current model set, so there's no
# "threshold" to sweep for them; a probability-threshold sensitivity check
# doesn't apply to a series with no regime probability in the first place.
thresholds_to_test = [0.3, 0.4, 0.5, 0.6, 0.7]

wti_sensitivity = threshold_sensitivity_check("wti_dates", "wti_prob_high", "WTI", thresholds_to_test)
```

```python
def crisis_episode_stats_wti(vol_series, percentile):
    threshold = vol_series.quantile(percentile)
    regime = (vol_series > threshold).astype(int)
    change = regime.diff()
    starts = vol_series.index[change == 1]
    ends   = vol_series.index[change == -1]
    if regime.iloc[0] == 1:
        starts = starts.insert(0, vol_series.index[0])
    durations = []
    for start in starts:
        future_ends = ends[ends > start]
        if len(future_ends) == 0:
            continue
        end = future_ends[0]
        durations.append(len(vol_series.loc[start:end]))
    return np.array(durations), threshold

percentiles_to_test = [0.70, 0.75, 0.80, 0.85, 0.90, 0.92, 0.95]
sensitivity_rows_wti = []
for p in percentiles_to_test:
    durations, thresh = crisis_episode_stats_wti(vol_wti, p)
    sensitivity_rows_wti.append({
        'Percentile cutoff': p, 'Volatility threshold': round(thresh, 4),
        'Num episodes': len(durations), 'Mean duration': round(durations.mean(), 1),
        'Median duration': int(np.median(durations)),
        'Min / Max duration': f"{durations.min()} / {durations.max()}",
    })

sensitivity_df_wti = pd.DataFrame(sensitivity_rows_wti)
print("=== WTI: crisis-definition threshold sensitivity ===")
print(sensitivity_df_wti.to_string(index=False))
```

```python
from scipy.stats import genpareto, skew

vol_wti = master_wti['GARCH_VOL_FULL_SAMPLE_WTI'].dropna()

def episodes_from_threshold_wti(vol_series, threshold):
    regime = (vol_series > threshold).astype(int)
    change = regime.diff()
    starts = vol_series.index[change == 1]
    ends   = vol_series.index[change == -1]
    if regime.iloc[0] == 1:
        starts = starts.insert(0, vol_series.index[0])
    durations = []
    for start in starts:
        future_ends = ends[ends > start]
        if len(future_ends) == 0:
            continue  # right-censored, exclude
        end = future_ends[0]
        durations.append(len(vol_series.loc[start:end]))
    return np.array(durations)

def threshold_percentile(vol_series, p=0.90):
    return vol_series.quantile(p)

def threshold_mad_zscore(vol_series, z_cutoff=3.5):
    median = vol_series.median()
    mad = (vol_series - median).abs().median()
    return median + (z_cutoff / 0.6745) * mad

def threshold_log_sigma(vol_series, n_sigma=3):
    log_vol = np.log(vol_series)
    print(f"    Skewness -- raw: {skew(vol_series):.3f} | log: {skew(log_vol):.3f}")
    mu, sigma = log_vol.mean(), log_vol.std()
    return np.exp(mu + n_sigma * sigma)

def threshold_evt_pot(vol_series, low=0.80, high=0.96, step=0.01, window=3):
    vals = vol_series.values
    candidates = np.quantile(vals, np.arange(low, high, step))
    shapes = []
    for u in candidates:
        exceed = vals[vals > u] - u
        if len(exceed) < 15:
            shapes.append(np.nan)
            continue
        try:
            c, _, _ = genpareto.fit(exceed, floc=0)
            shapes.append(c)
        except Exception:
            shapes.append(np.nan)
    shapes = np.array(shapes)
    rolling_std = pd.Series(shapes).rolling(window, center=True).std().values
    stable_idx = np.nanargmin(rolling_std)
    selected = candidates[stable_idx]
    diagnostics = pd.DataFrame({
        'candidate_threshold': candidates.round(4),
        'gpd_shape_param': np.round(shapes, 3),
        'rolling_std_of_shape': np.round(rolling_std, 4),
    })
    return selected, diagnostics

def summarize(name, threshold, durations):
    return {
        'Method': name, 'Threshold': round(threshold, 4),
        'Num episodes': len(durations),
        'Mean duration': round(durations.mean(), 1) if len(durations) else np.nan,
        'Median duration': int(np.median(durations)) if len(durations) else np.nan,
        'Min / Max': f"{durations.min()}/{durations.max()}" if len(durations) else "-",
    }

rows_wti = []
thr = threshold_percentile(vol_wti, 0.90)
rows_wti.append(summarize('Percentile (90th) -- current', thr, episodes_from_threshold_wti(vol_wti, thr)))

print("WTI Method 3 (log + 3sigma) diagnostics:")
thr = threshold_log_sigma(vol_wti, n_sigma=3)
rows_wti.append(summarize('Log-transform + 3sigma', thr, episodes_from_threshold_wti(vol_wti, thr)))

thr = threshold_mad_zscore(vol_wti, z_cutoff=3.5)
rows_wti.append(summarize('MAD robust z-score (|z|>3.5)', thr, episodes_from_threshold_wti(vol_wti, thr)))

thr, evt_diagnostics_wti = threshold_evt_pot(vol_wti)
rows_wti.append(summarize('EVT / Peaks-over-threshold', thr, episodes_from_threshold_wti(vol_wti, thr)))

comparison_df_wti = pd.DataFrame(rows_wti)
print("\n=== WTI: crisis-definition method comparison ===")
print(comparison_df_wti.to_string(index=False))

print("\n=== EVT threshold-stability diagnostics (WTI) ===")
print(evt_diagnostics_wti.to_string(index=False))
```

```python
def crisis_episode_stats(vol_series, percentile):
    threshold = vol_series.quantile(percentile)
    regime = (vol_series > threshold).astype(int)
    change = regime.diff()
    starts = vol_series.index[change == 1]
    ends   = vol_series.index[change == -1]
    if regime.iloc[0] == 1:
        starts = starts.insert(0, vol_series.index[0])

    durations = []
    for start in starts:
        future_ends = ends[ends > start]
        if len(future_ends) == 0:
            continue  # right-censored — true duration unknown, exclude
        end = future_ends[0]
        durations.append(len(vol_series.loc[start:end]))
    return np.array(durations), threshold

percentiles_to_test = [0.70, 0.75, 0.80, 0.85, 0.90, 0.92, 0.95]
sensitivity_rows = []
for p in percentiles_to_test:
    durations, thresh = crisis_episode_stats(master_wti['GARCH_VOL_FULL_SAMPLE_WTI'].dropna(), p)
    sensitivity_rows.append({
        'Percentile cutoff':   p,
        'Volatility threshold': round(thresh, 4),
        'Num episodes':        len(durations),
        'Mean duration':       round(durations.mean(), 1),
        'Median duration':     int(np.median(durations)),
        'Min / Max duration':  f"{durations.min()} / {durations.max()}",
    })

wti_sensitivity_df = pd.DataFrame(sensitivity_rows)
print("=== WTI crisis-definition threshold sensitivity (with duration stats) ===")
print(wti_sensitivity_df.to_string(index=False))
```

```python
# ═══════════════════════════════════════════════════════════════════════
# Comparing crisis-definition methods (WTI): percentile (current) vs.
# MAD-robust z-score vs. log-transform + 3σ vs. EVT / peaks-over-threshold
# ═══════════════════════════════════════════════════════════════════════
from scipy.stats import genpareto, skew

vol_wti = master_wti['GARCH_VOL_FULL_SAMPLE_WTI'].dropna()

def episodes_from_threshold(vol_series, threshold):
    """Same contiguous-run extraction logic used elsewhere in the notebook,
    generalized to take a raw threshold value instead of a percentile.
    Right-censored episodes (still ongoing at the end of the sample) are
    excluded -- their true duration is unknown, so keeping a truncated
    value would corrupt the duration stats below."""
    regime = (vol_series > threshold).astype(int)
    change = regime.diff()
    starts = vol_series.index[change == 1]
    ends   = vol_series.index[change == -1]
    if regime.iloc[0] == 1:
        starts = starts.insert(0, vol_series.index[0])
    durations = []
    for start in starts:
        future_ends = ends[ends > start]
        if len(future_ends) == 0:
            continue  # right-censored, exclude
        end = future_ends[0]
        durations.append(len(vol_series.loc[start:end]))
    return np.array(durations)

# ── Method: current baseline (90th percentile) ───────────────────────
def threshold_percentile(vol_series, p=0.90):
    return vol_series.quantile(p)

# ── Method 2: MAD-based robust z-score (Iglewicz & Hoaglin, 1993) ────
def threshold_mad_zscore(vol_series, z_cutoff=3.5):
    median = vol_series.median()
    mad = (vol_series - median).abs().median()
    return median + (z_cutoff / 0.6745) * mad

# ── Method 3: log-transform, then classic 3-sigma rule ────────────────
def threshold_log_sigma(vol_series, n_sigma=3):
    log_vol = np.log(vol_series)
    print(f"    Skewness -- raw: {skew(vol_series):.3f} | log: {skew(log_vol):.3f} "
          f"(log should be closer to 0 if this method is well justified)")
    mu, sigma = log_vol.mean(), log_vol.std()
    return np.exp(mu + n_sigma * sigma)

# ── Method 4: EVT / Peaks-Over-Threshold ──────────────────────────────
def threshold_evt_pot(vol_series, low=0.80, high=0.96, step=0.01, window=3):
    vals = vol_series.values
    candidates = np.quantile(vals, np.arange(low, high, step))
    shapes = []
    for u in candidates:
        exceed = vals[vals > u] - u
        if len(exceed) < 15:
            shapes.append(np.nan)
            continue
        try:
            c, _, _ = genpareto.fit(exceed, floc=0)
            shapes.append(c)
        except Exception:
            shapes.append(np.nan)
    shapes = np.array(shapes)
    rolling_std = pd.Series(shapes).rolling(window, center=True).std().values
    stable_idx = np.nanargmin(rolling_std)
    selected = candidates[stable_idx]
    diagnostics = pd.DataFrame({
        'candidate_threshold': candidates.round(4),
        'gpd_shape_param': np.round(shapes, 3),
        'rolling_std_of_shape': np.round(rolling_std, 4),
    })
    return selected, diagnostics

# ── Run all four and build the comparison table ───────────────────────
def summarize(name, threshold, durations):
    return {
        'Method': name,
        'Threshold': round(threshold, 4),
        'Num episodes': len(durations),
        'Mean duration': round(durations.mean(), 1) if len(durations) else np.nan,
        'Median duration': int(np.median(durations)) if len(durations) else np.nan,
        'Min / Max': f"{durations.min()}/{durations.max()}" if len(durations) else "-",
    }

rows = []

thr = threshold_percentile(vol_wti, 0.90)
rows.append(summarize('Percentile (90th) -- current', thr, episodes_from_threshold(vol_wti, thr)))

print("Method 3 (log + 3sigma) diagnostics:")
thr = threshold_log_sigma(vol_wti, n_sigma=3)
rows.append(summarize('Log-transform + 3sigma', thr, episodes_from_threshold(vol_wti, thr)))

thr = threshold_mad_zscore(vol_wti, z_cutoff=3.5)
rows.append(summarize('MAD robust z-score (|z|>3.5)', thr, episodes_from_threshold(vol_wti, thr)))

thr, evt_diagnostics_wti = threshold_evt_pot(vol_wti)
rows.append(summarize('EVT / Peaks-over-threshold', thr, episodes_from_threshold(vol_wti, thr)))

comparison_df_wti = pd.DataFrame(rows)
print("\n=== WTI crisis-definition method comparison ===")
print(comparison_df_wti.to_string(index=False))

print("\n=== WTI EVT threshold-stability diagnostics (mean-residual-life-style check) ===")
print(evt_diagnostics_wti.to_string(index=False))
print("\n(Look for the stretch where 'gpd_shape_param' flattens out -- that's the")
print(" region EVT theory says the threshold should sit in; 'rolling_std_of_shape'")
print(" picks the flattest point automatically as the selected threshold above.)")
```

We tried four ways to define a "crisis" for WTI: a straight percentile cutoff, a robust z-score, a log+3σ rule, and EVT. They didn't all agree — episode counts ranged from 1 to 46 .

Log+3σ was easy to rule out. It assumes the data is roughly symmetric after a log transform, but it isn't (skewness of 1.01), so its answer (just 1 episode covering the whole window) isn't a real crisis definition, just a broken assumption.

EVT looked more rigorous but didn't hold up under its own test. The method only works if the shape parameter flattens out at some threshold — that's supposed to be the sign you've found a genuine cutoff. For WTI, it never flattens; it just climbs steadily from 0.46 to 0.64 across every candidate threshold. So the "best" threshold it picked wasn't a stable point, just the least-steep spot on a line that's still rising. Trusting it here would mean trusting the method at the exact moment it's telling you the data isn't ready for it.

That leaves the 90th percentile as the practical choice — same as we used for Brent. It's simple, doesn't lean on shaky assumptions, and gives us 42 episodes to work with, enough to train a duration model on.

```python
# ── Build crisis_table_wti from the 90th-percentile definition ──
high_vol_threshold_wti = master_wti['GARCH_VOL_FULL_SAMPLE_WTI'].quantile(0.90)
master_wti['HIGH_VOL_REGIME_WTI'] = (master_wti['GARCH_VOL_FULL_SAMPLE_WTI'] > high_vol_threshold_wti).astype(int)
print("WTI 90th-percentile threshold:", round(high_vol_threshold_wti, 4))

regime = master_wti['HIGH_VOL_REGIME_WTI']
change = regime.diff()
crisis_starts = master_wti.index[change ==  1]
crisis_ends   = master_wti.index[change == -1]

# Edge case: if the series is already in a high-vol regime on day 1,
# .diff() has nothing to compare against and won't register that as a
# start — check explicitly rather than assume it away.
if regime.iloc[0] == 1:
    print("NOTE: HIGH_VOL_REGIME_WTI is already 1 on the first day — adding "
          "the sample start as an implicit (and left-censored) crisis start.")
    crisis_starts = pd.Index([master_wti.index[0]]).append(crisis_starts)

crisis_periods = []
for i, start in enumerate(crisis_starts):
    future_ends = crisis_ends[crisis_ends > start]
    censored = len(future_ends) == 0
    end      = future_ends[0] if not censored else master_wti.index[-1]
    duration = len(master_wti.loc[start:end])
    crisis_periods.append({
        'crisis_id':     i + 1,
        'start_date':    start,
        'end_date':      end,
        'duration_days': duration,
        'log_duration':  np.log(duration),
        'censored':      censored
    })

crisis_table_full_wti = pd.DataFrame(crisis_periods)
n_censored_wti = crisis_table_full_wti['censored'].sum()

# Drop right-censored episodes — their true duration is unknown (the
# crisis may still be ongoing past the end of our data), so keeping a
# truncated duration_days value would corrupt the target variable.
crisis_table_wti = crisis_table_full_wti[~crisis_table_full_wti['censored']].drop(columns='censored').reset_index(drop=True)

print(f"WTI crisis episodes (raw): {len(crisis_table_full_wti)} | Censored (dropped): {n_censored_wti} | "
      f"Final: {len(crisis_table_wti)}")
```

```python
feature_cols_market = ['OVX', 'DXY', 'FSI', 'IGREA', 'GPR', 'WTI_REAL', 'NET_OIL_PRICE', 'CRUDE_INVENTORIES_WTI']
rows = []
for _, crisis in crisis_table_wti.iterrows():
    start = pd.Timestamp(crisis['start_date'])

    # ── Market features: 5-day mean immediately before crisis start ──
    window = master_wti.loc[:start - pd.Timedelta(days=1)].tail(5)
    feat = {col: window[col].mean() for col in feature_cols_market if col in window.columns}

    # ── Causal single-regime eGARCH refit: data up to (not incl.) start ──
    wti_sub = wti_ret.loc[:start].iloc[:-1]
    if len(wti_sub) < 100:
        print(f"Crisis #{crisis['crisis_id']} ({start.date()}): only {len(wti_sub)} "
              f"obs available — skipping eGARCH refit, using NaN for GARCH_VOL")
        feat['GARCH_VOL'] = np.nan
    else:
        try:
            garch_roll = arch_model(wti_sub, mean='Zero', vol='EGARCH', p=1, o=1, q=1,
                                     dist='skewt', rescale=True).fit(disp='off')
            feat['GARCH_VOL'] = float(garch_roll.conditional_volatility[-1])
        except Exception as e:
            print(f"Crisis #{crisis['crisis_id']} ({start.date()}): eGARCH refit failed — {e}")
            feat['GARCH_VOL'] = np.nan

    # ── Initial drop: same-day (signed) return on crisis start date ──
    if start in wti_ret.index:
        feat['INITIAL_DROP'] = wti_ret.loc[start]
    else:
        prior = wti_ret.loc[:start]
        feat['INITIAL_DROP'] = prior.iloc[-1] if len(prior) else np.nan

    feat['crisis_id']     = crisis['crisis_id']
    feat['start_date']    = start
    feat['duration_days'] = crisis['duration_days']
    feat['log_duration']  = crisis['log_duration']
    rows.append(feat)

df_model_wti = pd.DataFrame(rows)

meta_cols = ['crisis_id', 'start_date', 'duration_days', 'log_duration']
feature_cols_wti = [c for c in df_model_wti.columns if c not in meta_cols]

n_before = len(df_model_wti)
df_model_wti = df_model_wti.dropna(subset=feature_cols_wti).reset_index(drop=True)
print(f"Dropped {n_before - len(df_model_wti)} row(s) with missing features. {len(df_model_wti)} remain.")

df_model_wti = df_model_wti.set_index('crisis_id')
X_wti     = df_model_wti[feature_cols_wti].values
y_wti     = df_model_wti['duration_days'].values
y_log_wti = df_model_wti['log_duration'].values

print(f"df_model_wti shape: {df_model_wti.shape}")
print(f"Feature columns: {feature_cols_wti}")
```

```python
from scipy.stats import skew
from sklearn.model_selection import TimeSeriesSplit

XGB_PARAMS = dict(
    n_estimators     = 100,
    max_depth        = 3,
    learning_rate    = 0.05,
    subsample        = 0.8,
    colsample_bytree = 0.8,
    reg_alpha        = 0.5,
    reg_lambda       = 1.0,
    random_state     = 42,
    n_jobs           = -1,
    verbosity        = 0
)

def run_timeseries_cv(df_model, X, y, y_log, n_splits=5, xgb_params=XGB_PARAMS):
    """Time-based (forward-chaining) CV: train only on earlier crises, test
    only on later ones. Primary validation method. The earliest chunk of
    crises can never appear in a test fold (nothing precedes them to train
    on) -- used for training only, excluded from reported metrics, not
    silently dropped from the dataset."""
    order = np.argsort(df_model['start_date'].values)
    X_sorted, y_sorted, y_log_sorted = X[order], y[order], y_log[order]
    df_sorted = df_model.iloc[order]

    tscv = TimeSeriesSplit(n_splits=n_splits)
    y_true_idx, pred_log, pred_raw = [], [], []

    for train_idx, test_idx in tscv.split(X_sorted):
        model_log = xgb.XGBRegressor(**xgb_params)
        model_log.fit(X_sorted[train_idx], y_log_sorted[train_idx])
        pred_log.extend(model_log.predict(X_sorted[test_idx]))

        model_raw = xgb.XGBRegressor(**xgb_params)
        model_raw.fit(X_sorted[train_idx], y_sorted[train_idx])
        pred_raw.extend(model_raw.predict(X_sorted[test_idx]))

        y_true_idx.extend(test_idx)

    df_eval = df_sorted.iloc[y_true_idx].copy()
    y_eval = y_sorted[y_true_idx]
    pred_log = np.exp(np.array(pred_log))
    pred_raw = np.array(pred_raw)

    n_excluded = len(y_sorted) - len(y_eval)
    print(f"TimeSeriesSplit ({n_splits}): evaluated on {len(y_eval)}/{len(y_sorted)} crises")
    print(f"Earliest {n_excluded} crises never in a test fold (by construction) -- "
          f"first evaluable crisis: {df_eval['start_date'].iloc[0].date()}")

    return df_eval, y_eval, pred_log, pred_raw


def evaluate(y_true, y_pred, label):
    """Print RMSE / MAE / R² / MAPE / SMAPE for one set of predictions."""
    rmse  = np.sqrt(mean_squared_error(y_true, y_pred))
    mae   = mean_absolute_error(y_true, y_pred)
    r2    = r2_score(y_true, y_pred)
    mape  = np.mean(np.abs((y_true - y_pred) / y_true)) * 100
    smape = np.mean(2 * np.abs(y_pred - y_true) /
                     (np.abs(y_true) + np.abs(y_pred) + 1e-9)) * 100
    print(f"  [{label}]")
    print(f"    RMSE  : {rmse:.1f} days")
    print(f"    MAE   : {mae:.1f} days")
    print(f"    R²    : {r2:.3f}")
    print(f"    MAPE  : {mape:.1f}%  (unstable for short durations -- diagnostic only)")
    print(f"    SMAPE : {smape:.1f}%  (bounded 0-200%, more robust to small y_true)")
    return rmse, mae, r2


df_model_wti_eval, y_wti_eval, pred_log_wti, pred_raw_wti = run_timeseries_cv(
    df_model_wti, X_wti, y_wti, y_log_wti
)

evaluate(y_wti_eval, pred_log_wti, "WTI — XGBoost, log target")
evaluate(y_wti_eval, pred_raw_wti, "WTI — XGBoost, raw target")
```

```python
def pick_best(y, y_pred_log, y_pred_raw):
    """we decide to use the log-transformed target or the raw data
    based on the distributional skewness of the target, not on the model performance."""
    raw_skew = skew(y)
    log_skew = skew(np.log(y))
    use_log  = abs(log_skew) < abs(raw_skew)

    print("=== Target transform selection (a priori, not leakage-prone) ===")
    print(f"  Skewness -- raw duration: {raw_skew:.3f} | log duration: {log_skew:.3f}")
    print(f"  -> Using {'log' if use_log else 'raw'}-target "
          f"(closer to symmetric; decided before looking at model performance)")

    print("\n=== LOOCV performance (both reported for transparency; only one was chosen) ===")
    _, _, r2_log = evaluate(y, y_pred_log, "XGB — log-target (exp back)")
    _, _, r2_raw = evaluate(y, y_pred_raw, "XGB — raw target")

    if use_log:
        return y_pred_log, "Log-target", True, r2_log
    else:
        return y_pred_raw, "Raw-target", False, r2_raw

y_pred_final_wti, target_used_wti, used_log_wti, r2_final_wti = pick_best(y_wti_eval, pred_log_wti, pred_raw_wti)
```

WTI's target-transform choice followed the same a priori rule as Brent: log duration (skewness 0.774) is closer to symmetric than raw duration (2.281), so the log-target model is the one we report — decided before looking at performance, same discipline as before.

That model's R² is -0.160 (raw-target scored worse still, at -0.190). Both are negative, meaning the model underperforms simply predicting the average duration for every crisis. This is a real, honest result, not something to explain away — a 42-crisis sample is small, and it's consistent with the same fragile-but-real signal Brent showed (leakage-free R²=0.112, with a bootstrap CI spanning -0.067 to 0.277). WTI landing in the negative end of that same range is plausible, not alarming on its own.

Before drawing any conclusion about WTI specifically, this needs the same robustness checks Brent went through — naive baseline comparison, nested VIF-trim, bootstrap CI, and a permutation test — to tell whether this is weak-but-real signal or statistically indistinguishable from noise.

```python
# This function creates a clear visual check of your XGBoost model’s performance:
 
def plot_predictions(df_model, y, best_pred, r2, save_path=None):
    """Scatter of actual vs. predicted duration, plus a side-by-side bar chart per crisis."""
    fig, axes = plt.subplots(1, 2, figsize=(15, 6))

    ax = axes[0]
    ax.scatter(y, best_pred, color='steelblue', edgecolors='navy', s=80, zorder=3)
    lims = [min(y.min(), best_pred.min()) - 10, max(y.max(), best_pred.max()) + 10]
    ax.plot(lims, lims, 'r--', label='Perfect prediction')
    for sid, actual, pred in zip(df_model.index, y, best_pred):
        ax.annotate(f'#{sid}', (actual, pred), textcoords='offset points',
                    xytext=(5, 2), fontsize=7, color='gray')
    ax.set_xlabel('Actual Duration (days)')
    ax.set_ylabel('Predicted Duration (days)')
    ax.set_title(f'XGBoost LOOCV: Actual vs Predicted\nR² = {r2:.3f}')
    ax.legend()
    ax.grid(True, alpha=0.3)

    ax2 = axes[1]
    x, w = np.arange(len(df_model)), 0.38
    ax2.bar(x - w/2, y, w, label='Actual', color='steelblue', alpha=0.85)
    ax2.bar(x + w/2, best_pred, w, label='Predicted', color='darkorange', alpha=0.85)
    ax2.set_xticks(x)
    ax2.set_xticklabels(
        [f'#{i}\n{d.strftime("%Y-%m")}' for i, d in zip(df_model.index, df_model['start_date'])],
        fontsize=7, rotation=45, ha='right')
    ax2.set_ylabel('Duration (days)')
    ax2.set_title('Actual vs Predicted Duration per Crisis')
    ax2.legend()
    ax2.grid(axis='y', alpha=0.3)

    plt.tight_layout()
    if save_path:
        plt.savefig(save_path, dpi=150, bbox_inches='tight')
    plt.show()
```

```python
plot_predictions(df_model_wti_eval, y_wti_eval, y_pred_final_wti, r2_final_wti)
```

```python
# which features were most important to the model? SHAP values give a clear answer.
def plot_shap(model, X, target, feature_cols, save_path=None):
    """Retrain on the full dataset for stable SHAP values, then plot feature importance."""
    model.fit(X, target)

    explainer   = shap.TreeExplainer(model)
    shap_values = explainer.shap_values(X)

    importance = pd.Series(
        np.abs(shap_values).mean(axis=0),
        index=feature_cols
    ).sort_values(ascending=False)

    print("=== SHAP Feature Importance (mean |SHAP|) ===")
    for feat, val in importance.items():
        print(f"  {feat:<25} {val:.4f}")

    fig, axes = plt.subplots(1, 2, figsize=(16, 6))

    importance.sort_values().plot.barh(ax=axes[0], color='steelblue', edgecolor='navy', alpha=0.8)
    axes[0].set_title('SHAP Feature Importance\n(Mean |SHAP value|)')
    axes[0].set_xlabel('Mean |SHAP value|')
    axes[0].grid(axis='x', alpha=0.3)

    plt.sca(axes[1])
    shap.summary_plot(shap_values, X, feature_names=feature_cols, show=False)
    axes[1].set_title('SHAP Summary (dot plot)')

    plt.tight_layout()
    if save_path:
        plt.savefig(save_path, dpi=150, bbox_inches='tight')
    plt.show()
    return shap_values

def report_results(df_model, y, best_pred):
    """Actual vs. predicted duration per crisis, with error and % error columns."""
    results = df_model[['start_date', 'duration_days']].copy()
    results['predicted_days'] = np.round(best_pred).astype(int)
    results['error_days']     = results['predicted_days'] - results['duration_days']
    results['abs_pct_error']  = (np.abs(results['error_days']) / results['duration_days'] * 100).round(1)

    print("=== Crisis Duration Predictions ===")
    print(results.to_string())
    return results

sshap_values_wti = plot_shap(xgb.XGBRegressor(**XGB_PARAMS), X_wti,
                             y_log_wti if used_log_wti else y_wti,
                             feature_cols_wti)

results_wti = report_results(df_model_wti_eval, y_wti_eval, y_pred_final_wti)
```

```python
def nested_shap_feature_selection_ts(df_model, X, y_log, y, feature_names, n_splits=5, top_n=4, xgb_params=XGB_PARAMS):
    order = np.argsort(df_model['start_date'].values)
    X_sorted, y_log_sorted, y_sorted = X[order], y_log[order], y[order]
    df_sorted = df_model.iloc[order]

    tscv = TimeSeriesSplit(n_splits=n_splits)
    y_true_idx, preds_log, selected_per_fold = [], [], []

    for train_idx, test_idx in tscv.split(X_sorted):
        X_train, t_train = X_sorted[train_idx], y_log_sorted[train_idx]

        selector = xgb.XGBRegressor(**xgb_params)
        selector.fit(X_train, t_train)
        sel_importance = pd.Series(
            np.abs(shap.TreeExplainer(selector).shap_values(X_train)).mean(axis=0),
            index=feature_names
        ).sort_values(ascending=False)
        chosen = sel_importance.head(top_n).index.tolist()
        selected_per_fold.append(chosen)
        chosen_idx = [feature_names.index(c) for c in chosen]

        final_model = xgb.XGBRegressor(**xgb_params)
        final_model.fit(X_train[:, chosen_idx], t_train)
        preds_log.extend(final_model.predict(X_sorted[test_idx][:, chosen_idx]))
        y_true_idx.extend(test_idx)

    df_eval = df_sorted.iloc[y_true_idx].copy()
    y_eval = y_sorted[y_true_idx]
    preds = np.exp(np.array(preds_log))

    n_excluded = len(y_sorted) - len(y_eval)
    print(f"Nested TimeSeriesSplit ({n_splits}): evaluated on {len(y_eval)}/{len(y_sorted)} crises "
          f"({n_excluded} earliest never in a test fold, by construction)")

    return df_eval, y_eval, preds, selected_per_fold

df_eval_nested_wti, y_nested_wti, pred_nested_wti, selected_per_fold_wti = nested_shap_feature_selection_ts(
    df_model_wti, X_wti, y_log_wti, y_wti, feature_cols_wti, top_n=top_n
)

print("\n=== Nested SHAP feature selection, TimeSeriesSplit (honest, leakage-free) — WTI ===")
_, _, r2_nested_wti = evaluate(y_nested_wti, pred_nested_wti, "WTI XGB -- nested SHAP selection")

from collections import Counter
n_folds_wti = len(selected_per_fold_wti)
selection_counts_wti = Counter(f for fold in selected_per_fold_wti for f in fold)
print(f"\nFeature selection frequency across {n_folds_wti} outer folds:")
for feat, count in sorted(selection_counts_wti.items(), key=lambda kv: -kv[1]):
    print(f"  {feat:<25} selected in {count}/{n_folds_wti} folds")

results_nested_wti = report_results(df_eval_nested_wti, y_nested_wti, pred_nested_wti)
plot_predictions(df_eval_nested_wti, y_nested_wti, pred_nested_wti, r2_nested_wti)
```

WTI's nested SHAP feature selection (TimeSeriesSplit, honest) scores R²=-0.202, worse than the fixed full-feature model (-0.160).WTI_REAL was selected in all 5 folds, CRUDE_INVENTORIES_WTI in 4 of 5, with FSI and OVX selected less consistently. INITIAL_DROP — despite scoring highest on full-sample SHAP importance — was only selected in 1 of 5 folds here; a feature that looks important when evaluated on all the data at once doesn't necessarily stay important once selection is done honestly, per fold, without letting it see the crises it will later be tested on.

```python
print("Do the XGBoost models beat simple, non-ML baselines? (WTI)\n")
y_eval_base_wti, pred_mean_wti, pred_median_wti = baseline_mean_median_ts(df_model_wti, y_wti)
print()
pred_persist_wti, valid_persist_wti = baseline_persistence(df_model_wti, y_wti)
```

Compared against naive baselines — now evaluated the same forward-chaining way as the model itself, rather than leave-one-out — the nested SHAP model (R²=-0.202) is beaten by both the naive mean-duration guess (-0.060) and the naive median-duration guess (-0.200); only the persistence baseline (-1.093, predicting each crisis will last as long as the previous one) scores worse than the model. This means the model isn't just weak, it's currently doing worse than the simplest reasonable guesses: predicting the average or median duration for every crisis. This is consistent with the fixed-feature-set result, and points toward the same conclusion — WTI's crisis durations may not be predictable from this feature set at this sample size, a genuinely different finding from Brent's.

```python
# Multicollinearity check for the volatility/stress-proxy features. Tree
# models handle correlated inputs fine for prediction, but SHAP importance
# is known to split credit somewhat arbitrarily among correlated features.

corr_features_wti = [c for c in feature_cols_wti if c in df_model_wti.columns]

corr_matrix_wti = df_model_wti[corr_features_wti].corr()

print("=== WTI feature correlation matrix ===")
print(corr_matrix_wti.round(2))

fig, ax = plt.subplots(figsize=(7, 6))
im = ax.imshow(corr_matrix_wti, vmin=-1, vmax=1, cmap='RdBu_r')
ax.set_xticks(range(len(corr_matrix_wti)))
ax.set_yticks(range(len(corr_matrix_wti)))
ax.set_xticklabels(corr_matrix_wti.columns, rotation=45, ha='right')
ax.set_yticklabels(corr_matrix_wti.columns)
for i in range(len(corr_matrix_wti)):
    for j in range(len(corr_matrix_wti)):
        val = corr_matrix_wti.iloc[i, j]
        ax.text(j, i, f"{val:.2f}", ha='center', va='center',
                color='white' if abs(val) > 0.5 else 'black', fontsize=8)
plt.colorbar(im, label='Pearson correlation')
plt.title('WTI feature correlation -- volatility/stress proxies')
plt.tight_layout()
plt.show()

try:
    from statsmodels.stats.outliers_influence import variance_inflation_factor
    X_vif_wti = df_model_wti[corr_features_wti].dropna()
    vif_wti = pd.Series(
        [variance_inflation_factor(X_vif_wti.values, i) for i in range(X_vif_wti.shape[1])],
        index=corr_features_wti
    ).sort_values(ascending=False)
    print("\n=== WTI Variance Inflation Factor ===")
    print(vif_wti.round(2))
except ImportError:
    print("statsmodels not installed -- run `pip install statsmodels` for the VIF check.")
```

```python
# ── Nested VIF-trim (honest, leakage-free), TimeSeriesSplit — WTI ──
# We're NOT treating raw VIF as our primary feature-selection method,
# given VIF's known instability at this sample size (42 crises, 9
# features) -- see the check above, where several features showed high
# VIF, almost certainly a small-sample artifact rather than genuine
# redundancy. VIF elimination is recomputed fresh inside each outer
# TimeSeriesSplit fold, using only that fold's training data.

df_eval_vif_wti, y_vif_wti, pred_vif_wti, selected_per_fold_vif_wti = nested_vif_trim_ts(
    df_model_wti, X_wti, feature_cols_wti, y_wti, y_log_wti
)

print("\n=== Nested VIF-trim (honest, leakage-free) — WTI ===")
_, _, r2_vif_nested_wti = evaluate(y_vif_wti, pred_vif_wti, "WTI XGB -- nested VIF-trim, TimeSeriesSplit")

print(f"\nFeature sets selected per fold:")
for i, feats in enumerate(selected_per_fold_vif_wti):
    print(f"  Fold {i+1}: {feats}")

print(f"\n=== Comparison: full feature set (fixed) vs. SHAP-nested vs. VIF-nested (WTI) ===")
print(f"Full, fixed features:  R² = {r2_final_wti:.3f}")
print(f"Nested SHAP selection: R² = {r2_nested_wti:.3f}")
print(f"Nested VIF-trim:       R² = {r2_vif_nested_wti:.3f}")

results_vif_wti = report_results(df_eval_vif_wti, y_vif_wti, pred_vif_wti)    
```

```python
from scipy.stats import bootstrap

# Bootstraps the R² statistic by resampling the out-of-fold (y_true, y_pred)
# pairs already produced by the nested VIF-trim above (pred_vif_wti /
# y_vif_wti). This does NOT re-run feature selection or hyperparameter
# choice per resample -- that would require ~10,000x the runtime of the
# nested procedure above and isn't feasible. This is the standard
# practical compromise: bootstrap the statistic given a fixed set of
# out-of-fold predictions, not the full modeling pipeline.

res_wti = bootstrap(
    (y_vif_wti, pred_vif_wti),
    statistic=lambda yt, yp: r2_score(yt, yp),
    n_resamples=10000,
    method='BCa',
    paired=True,
    random_state=42
)

print(f"Nested VIF-trim R² (WTI) = {r2_vif_nested_wti:.3f}")
print(f"Bootstrap 95% BCa CI:      [{res_wti.confidence_interval.low:.3f}, "
      f"{res_wti.confidence_interval.high:.3f}]  (n_resamples=10000)")
```

We checked how much this -0.132 number could wobble if we'd tested on slightly different data, and the range came out to [-0.343, 0.083]. That's almost entirely on the negative side — only a tiny part into positive territory. 

This lines up with everything else we've found: naive guessing beats the model, and no matter which features we picked, the result stayed in that same negative range. It's a different story from Brent, where the range genuinely stretched well into positive numbers around a positive result. Here, WTI's range sits mostly in negative territory instead.

```python
# Permutation test: is the model's real R² actually meaningful, or could
# a model like this reach a similar (or worse) score on pure noise?
n_permutations = 200
rng = np.random.default_rng(42)
null_r2_wti = []

for _ in range(n_permutations):
    y_shuffled_wti = rng.permutation(y_wti)
    y_shuffled_log_wti = np.log(y_shuffled_wti)
    df_model_shuffled_wti = df_model_wti.copy()
    df_model_shuffled_wti['duration_days'] = y_shuffled_wti
    df_model_shuffled_wti['log_duration']  = y_shuffled_log_wti

    _, y_eval_shuf, pred_shuf, _ = nested_vif_trim_ts(
        df_model_shuffled_wti, X_wti, feature_cols_wti, y_shuffled_wti, y_shuffled_log_wti
    )
    null_r2_wti.append(r2_score(y_eval_shuf, pred_shuf))

null_r2_wti = np.array(null_r2_wti)
print(f"Real R² (nested VIF-trim, WTI): {r2_vif_nested_wti:.3f}")
print(f"Null distribution (shuffled y): mean={null_r2_wti.mean():.3f}, "
      f"95th percentile={np.percentile(null_r2_wti, 95):.3f}")
print(f"Real R² beats {np.mean(null_r2_wti < r2_vif_nested_wti)*100:.1f}% of random permutations")
```

```python
n_at_least_as_extreme_wti = np.sum(null_r2_wti >= r2_vif_nested_wti)
p_value_wti = (n_at_least_as_extreme_wti + 1) / (n_permutations + 1)

print(f"Exact permutation p-value: p = {p_value_wti:.4f} "
      f"(Phipson & Smyth, 2010 correction)")
```

We ran the same permutation check on WTI's best model: scrambled the durations 200 times, reran the model on each, and compared. Our real model beat 74% of those scrambled attempts — better than most, but not enough to call it a real, reliable effect. For reference, Brent's model beat 99% of its scrambled versions, which is a clear, confident result. This one sits in a much murkier middle: there might be a faint real signal here, but with only 42 crises to work with, we genuinely can't tell it apart from luck. Rather than force this into a "works" or "doesn't work" answer, the honest read is: weakly suggestive, not statistically significant.

```python
vif_features_wti_single = list(feature_cols_wti)
dropped_for_vif_wti_single = []

while True:
    X_vif_check = df_model_wti[vif_features_wti_single].dropna()
    vif_check = pd.Series(
        [variance_inflation_factor(X_vif_check.values, i) for i in range(X_vif_check.shape[1])],
        index=vif_features_wti_single
    )
    worst_feature, worst_vif = vif_check.idxmax(), vif_check.max()
    if worst_vif <= 10 or len(vif_features_wti_single) <= 2:
        break
    vif_features_wti_single.remove(worst_feature)
    dropped_for_vif_wti_single.append(worst_feature)

feature_cols_trimmed_wti = vif_features_wti_single
print(f"WTI single-pass VIF trim kept: {feature_cols_trimmed_wti}")

order_wti = np.argsort(df_model_wti['start_date'].values)
X_trimmed_wti     = df_model_wti[feature_cols_trimmed_wti].values[order_wti]
y_trimmed_wti     = df_model_wti['duration_days'].values[order_wti]
y_log_trimmed_wti = df_model_wti['log_duration'].values[order_wti]
```

```python
final_search_cv_wti = KFold(n_splits=5, shuffle=True, random_state=42)
final_search_wti = GridSearchCV(
    xgb.XGBRegressor(random_state=42, n_jobs=-1, verbosity=0),
    param_grid=XGB_PARAM_GRID, cv=final_search_cv_wti,
    scoring='neg_mean_squared_error', n_jobs=-1
)
final_search_wti.fit(X_trimmed_wti, y_log_trimmed_wti)
print("Best params found (WTI):", final_search_wti.best_params_)

XGB_PARAMS_TUNED_WTI = dict(final_search_wti.best_params_, random_state=42, n_jobs=-1, verbosity=0)

df_model_wti_final_eval, y_wti_final, pred_log_wti_final, _ = run_timeseries_cv(
    df_model_wti, X_trimmed_wti, y_trimmed_wti, y_log_trimmed_wti, xgb_params=XGB_PARAMS_TUNED_WTI
)
_, _, r2_final_tuned_wti = evaluate(y_wti_final, pred_log_wti_final, "WTI -- final tuned params, honest check")

seeds_to_test = [7, 42, 123, 2024, 99]
seed_results_wti = []
for seed in seeds_to_test:
    search_cv = KFold(n_splits=5, shuffle=True, random_state=seed)
    search = GridSearchCV(xgb.XGBRegressor(random_state=seed, n_jobs=-1, verbosity=0),
                           param_grid=XGB_PARAM_GRID, cv=search_cv,
                           scoring='neg_mean_squared_error', n_jobs=-1)
    search.fit(X_trimmed_wti, y_log_trimmed_wti)
    params_seed = dict(search.best_params_, random_state=42, n_jobs=-1, verbosity=0)
    _, y_seed, pred_seed, _ = run_timeseries_cv(df_model_wti, X_trimmed_wti, y_trimmed_wti, y_log_trimmed_wti, xgb_params=params_seed)
    _, _, r2_seed = evaluate(y_seed, pred_seed, f"seed={seed}")
    seed_results_wti.append({'seed': seed, 'params': search.best_params_, 'r2_honest': r2_seed})

seed_df_wti = pd.DataFrame(seed_results_wti)
print(seed_df_wti.to_string(index=False))
print(f"Mean R²: {seed_df_wti['r2_honest'].mean():.3f} | Range: [{seed_df_wti['r2_honest'].min():.3f}, {seed_df_wti['r2_honest'].max():.3f}]")
```

Unlike Brent, hyperparameter tuning does not help WTI. The untuned nested VIF-trim model scored R²=-0.132; every tuned version, across all 5 seeds, scored worse (-0.218 to -0.284, mean -0.248). The seeds also cluster tightly here, unlike Brent's wider spread — suggesting this isn't "sometimes the search gets lucky," but a consistent pattern of the search overfitting to noise rather than finding real structure to sharpen. Given this, we treat the untuned nested VIF-trim (R²=-0.132) as WTI's final result going forward, not the tuned version. That tuning reliably makes WTI worse — while it reliably helped Brent — is itself a meaningful difference between the two commodities, worth reporting rather than glossing over.

```python
y_naive_wti, pred_naive_wti = naive_mean_baseline_ts(df_eval_vif_wti, y_vif_wti)

y_naive_wti, pred_naive_wti = naive_mean_baseline_ts(df_model_wti, y_wti)

model_errors_wti = np.abs(y_vif_wti - pred_vif_wti)
naive_errors_wti = np.abs(y_naive_wti - pred_naive_wti)

print(f"Model MAE: {model_errors_wti.mean():.2f} | Naive MAE: {naive_errors_wti.mean():.2f}")
stat, p_value = wilcoxon(model_errors_wti, naive_errors_wti, alternative='less')

print(f"Wilcoxon (model < naive): statistic={stat:.1f}, p={p_value:.4f}")

print(f"Model MAE: {model_errors_wti.mean():.2f} | Naive MAE: {naive_errors_wti.mean():.2f}")
stat, p_value = wilcoxon(model_errors_wti, naive_errors_wti, alternative='less')
print(f"Wilcoxon (model < naive): statistic={stat:.1f}, p={p_value:.4f}")
```

unlike Brent, where multiple independent tests (permutation, bootstrap, Wilcoxon) all agreed WTI shows a real, significant effect, WTI's tests all agree in the other direction — consistently landing in "not statistically significant" territory. That consistency across different types of tests is itself a meaningful finding: it's not that one test happened to be inconclusive, it's that every angle we've checked tells the same "weak, unconfirmed" story.

```python
gaps_wti = df_model_wti.sort_values('start_date')['start_date'].diff().dt.days
close_start_dates_wti = df_model_wti.sort_values('start_date').loc[gaps_wti <= 5, 'start_date']
keep_mask_wti = ~df_model_wti['start_date'].isin(close_start_dates_wti)

df_model_wti_declustered = df_model_wti.loc[keep_mask_wti].copy()
X_declustered_wti     = df_model_wti.loc[keep_mask_wti, feature_cols_wti].values
y_declustered_wti     = df_model_wti.loc[keep_mask_wti, 'duration_days'].values
y_declustered_log_wti = df_model_wti.loc[keep_mask_wti, 'log_duration'].values

_, y_decl_wti, pred_decl_wti, _ = nested_vif_trim_ts(
    df_model_wti_declustered, X_declustered_wti, feature_cols_wti, y_declustered_wti, y_declustered_log_wti
)
_, _, r2_declustered_wti = evaluate(y_decl_wti, pred_decl_wti, "WTI -- clustered pairs de-duplicated")
print(f"With clustering as-is (N={len(df_model_wti)}): R² = {r2_vif_nested_wti:.3f}")
print(f"With clustered pairs dropped (N={len(df_model_wti_declustered)}): R² = {r2_declustered_wti:.3f}")
```

```python

N_HOLDOUT = 6
df_sorted_wti = df_model_wti.sort_values('start_date').reset_index(drop=True)
holdout_dates_wti = df_sorted_wti['start_date'].iloc[-N_HOLDOUT:].values
holdout_mask_wti = df_model_wti['start_date'].isin(holdout_dates_wti)
dev_mask_wti = ~holdout_mask_wti

X_dev_wti = df_model_wti.loc[dev_mask_wti, feature_cols_wti].values
y_dev_wti = df_model_wti.loc[dev_mask_wti, 'duration_days'].values
y_dev_log_wti = df_model_wti.loc[dev_mask_wti, 'log_duration'].values
X_holdout_wti = df_model_wti.loc[holdout_mask_wti, feature_cols_wti].values
y_holdout_wti = df_model_wti.loc[holdout_mask_wti, 'duration_days'].values

dev_df_wti = pd.DataFrame(X_dev_wti, columns=feature_cols_wti)
vif_features_holdout_wti = list(feature_cols_wti)
while len(vif_features_holdout_wti) > 2:
    X_vif_h = dev_df_wti[vif_features_holdout_wti].dropna()
    vif_h = pd.Series([variance_inflation_factor(X_vif_h.values, i) for i in range(X_vif_h.shape[1])], index=vif_features_holdout_wti)
    worst_f, worst_v = vif_h.idxmax(), vif_h.max()
    if worst_v <= 10: break
    vif_features_holdout_wti.remove(worst_f)

dev_idx_wti = [feature_cols_wti.index(c) for c in vif_features_holdout_wti]
final_model_wti = xgb.XGBRegressor(**XGB_PARAMS)
final_model_wti.fit(X_dev_wti[:, dev_idx_wti], y_dev_log_wti)
holdout_pred_wti = np.exp(final_model_wti.predict(X_holdout_wti[:, dev_idx_wti]))

_, _, r2_holdout_wti = evaluate(y_holdout_wti, holdout_pred_wti, "WTI -- confirmatory holdout")
print(pd.DataFrame({'start_date': df_model_wti.loc[holdout_mask_wti, 'start_date'].values,
                     'actual_days': y_holdout_wti, 'predicted_days': np.round(holdout_pred_wti).astype(int)}))
```

WTI's confirmatory holdout — 6 crises never touched by any feature selection or model decision — shows a clear split. Five of six predictions were excellent: three exact matches, two off by only 1-2 days. The sixth, a 71-day crisis starting 2026-03-11, was badly underpredicted at just 3 days, a 68-day miss. That single outlier is what drives the negative R² (-0.230) and the gap between MAE (12.0 days, a reasonable typical error) and RMSE (27.8 days, inflated by the one big miss). This mirrors the same pattern we saw with Brent's holdout: the model handles ordinary-severity crises well but badly underestimates the rare long-duration ones. For WTI, this is a clean illustration of that limit — five accurate predictions in a row, then one genuinely hard case the model had no way to see coming from this feature set.

Summary of model validation — WTI

The final model (XGBoost, log-duration target, nested VIF-trimmed feature set, default hyperparameters) achieves R² = -0.132 under TimeSeriesSplit, evaluated on the 35 of 42 crises with sufficient trailing history.

Bootstrap confidence interval: [-0.343, 0.083] — mostly negative, with only a small sliver crossing into positive territory. Unlike Brent's CI, which spans a wide positive range, WTI's interval sits close to and mostly below zero.
Naive baselines: mean duration (R²=-0.060) and median duration (R²=-0.175) both outperform the model in raw R² terms; only previous-crisis persistence (R²=-1.207) is clearly worse.
Hyperparameter tuning: unlike Brent, tuning consistently hurt WTI's result across all 5 tested seeds (mean R²=-0.248, tightly clustered), so the untuned nested VIF-trim model is used as WTI's final result rather than a tuned variant.
Wilcoxon signed-rank test: model MAE (14.23 days) is slightly lower than the naive baseline's (15.08 days), but the difference is not statistically significant (p=0.2305).
Permutation test: the real R² beat 74.0% of 200 label-shuffled permutations (p≈0.264) — some indication of a weak effect, but well short of conventional statistical significance.
De-clustering check: removing 4 crises with overlapping feature windows brought R² from -0.132 to -0.189 — a small decline, ruling out the concern that the result depends on clustered-episode artifacts.
Confirmatory holdout: the 6 most recent crises, untouched by any feature or model decision, were predicted accurately in 5 of 6 cases (0-2 day error); the sixth, an unusually long 71-day crisis, was substantially underpredicted (68-day error), driving the holdout's negative R².

Taken together, WTI shows a genuinely different picture from Brent: every significance test (bootstrap, permutation, Wilcoxon) lands in ambiguous or non-significant territory, and hyperparameter tuning — which improved Brent's result — consistently degrades WTI's. This isn't evidence of a broken pipeline; it's a coherent, well-tested null-to-weak result. The confirmatory holdout suggests the model handles typical-severity crises reasonably well, but offers no evidence of skill on rare, long-duration events — the case that matters most and is hardest to predict from this feature set.

```python
def nested_vif_trim_ts_aft(df_model, X_full, feature_cols, y_lower, y_upper, y, n_splits=5, vif_threshold=10):
    order = np.argsort(df_model['start_date'].values)
    X_sorted = X_full[order]
    y_lower_sorted, y_upper_sorted, y_sorted = y_lower[order], y_upper[order], y[order]
    df_sorted = df_model.iloc[order]

    tscv = TimeSeriesSplit(n_splits=n_splits)
    y_true_idx, preds = [], []

    for train_idx, test_idx in tscv.split(X_sorted):
        X_train_df = pd.DataFrame(X_sorted[train_idx], columns=feature_cols)
        vif_features = list(feature_cols)

        if X_train_df.shape[0] > len(vif_features) + 2:
            while len(vif_features) > 2:
                X_vif = X_train_df[vif_features].dropna()
                vif = pd.Series(
                    [variance_inflation_factor(X_vif.values, i) for i in range(X_vif.shape[1])],
                    index=vif_features
                )
                worst_feature, worst_vif = vif.idxmax(), vif.max()
                if worst_vif <= vif_threshold:
                    break
                vif_features.remove(worst_feature)

        chosen_idx = [feature_cols.index(c) for c in vif_features]
        model = fit_aft(X_sorted[train_idx][:, chosen_idx], y_lower_sorted[train_idx], y_upper_sorted[train_idx])
        dtest = xgb.DMatrix(X_sorted[test_idx][:, chosen_idx])
        preds.extend(model.predict(dtest))
        y_true_idx.extend(test_idx)

    df_eval = df_sorted.iloc[y_true_idx].copy()
    y_eval = y_sorted[y_true_idx]
    return df_eval, y_eval, np.array(preds)

df_eval_aft_wti, y_aft_wti, pred_aft_wti = nested_vif_trim_ts_aft(
    df_model_wti, X_wti, feature_cols_wti, y_lower_wti, y_upper_wti, y_wti
)
_, _, r2_aft_wti = evaluate(y_aft_wti, pred_aft_wti, "WTI -- nested VIF-trim AFT, TimeSeriesSplit")
```

```python
#percentile sensitivity analysis for gold crises    

def crisis_episode_stats_gold(vol_series, percentile):
    threshold = vol_series.quantile(percentile)
    regime = (vol_series > threshold).astype(int)
    change = regime.diff()
    starts = vol_series.index[change == 1]
    ends   = vol_series.index[change == -1]
    if regime.iloc[0] == 1:
        starts = starts.insert(0, vol_series.index[0])
    durations = []
    for start in starts:
        future_ends = ends[ends > start]
        if len(future_ends) == 0:
            continue
        durations.append(len(vol_series.loc[start:future_ends[0]]))
    return np.array(durations), threshold

percentiles_to_test = [0.70, 0.75, 0.80, 0.85, 0.90, 0.92, 0.95]
sensitivity_rows_gold = []
for p in percentiles_to_test:
    durations, thresh = crisis_episode_stats_gold(master_gold['GARCH_VOL_FULL_SAMPLE_GOLD'].dropna(), p)
    sensitivity_rows_gold.append({
        'Percentile cutoff': p, 'Volatility threshold': round(thresh, 4),
        'Num episodes': len(durations),
        'Mean duration': round(durations.mean(), 1) if len(durations) else np.nan,
        'Median duration': int(np.median(durations)) if len(durations) else np.nan,
        'Min / Max duration': f"{durations.min()} / {durations.max()}" if len(durations) else "n/a",
    })
sensitivity_df_gold = pd.DataFrame(sensitivity_rows_gold)
print("=== Gold: crisis-definition threshold sensitivity ===")
print(sensitivity_df_gold.to_string(index=False))
```

```python
# ═══════════════════════════════════════════════════════════════════════
# Comparing crisis-definition methods (Gold): percentile (current) vs.
# MAD-robust z-score vs. log-transform + 3σ vs. EVT / peaks-over-threshold
# ═══════════════════════════════════════════════════════════════════════
from scipy.stats import genpareto, skew

vol_gold = master_gold['GARCH_VOL_FULL_SAMPLE_GOLD'].dropna()

def episodes_from_threshold_gold(vol_series, threshold):
    """Same contiguous-run extraction logic used elsewhere in the notebook,
    generalized to take a raw threshold value instead of a percentile.
    Right-censored episodes (still ongoing at the end of the sample) are
    excluded -- their true duration is unknown, so keeping a truncated
    value would corrupt the duration stats below."""
    regime = (vol_series > threshold).astype(int)
    change = regime.diff()
    starts = vol_series.index[change == 1]
    ends   = vol_series.index[change == -1]
    if regime.iloc[0] == 1:
        starts = starts.insert(0, vol_series.index[0])
    durations = []
    for start in starts:
        future_ends = ends[ends > start]
        if len(future_ends) == 0:
            continue  # right-censored, exclude
        end = future_ends[0]
        durations.append(len(vol_series.loc[start:end]))
    return np.array(durations)

# ── Method: current baseline (90th percentile) ───────────────────────
def threshold_percentile(vol_series, p=0.90):
    return vol_series.quantile(p)

# ── Method 2: MAD-based robust z-score (Iglewicz & Hoaglin, 1993) ────
def threshold_mad_zscore(vol_series, z_cutoff=3.5):
    median = vol_series.median()
    mad = (vol_series - median).abs().median()
    return median + (z_cutoff / 0.6745) * mad

# ── Method 3: log-transform, then classic 3-sigma rule ────────────────
def threshold_log_sigma(vol_series, n_sigma=3):
    log_vol = np.log(vol_series)
    print(f"    Skewness -- raw: {skew(vol_series):.3f} | log: {skew(log_vol):.3f} "
          f"(log should be closer to 0 if this method is well justified)")
    mu, sigma = log_vol.mean(), log_vol.std()
    return np.exp(mu + n_sigma * sigma)

# ── Method 4: EVT / Peaks-Over-Threshold ──────────────────────────────
def threshold_evt_pot(vol_series, low=0.80, high=0.96, step=0.01, window=3):
    vals = vol_series.values
    candidates = np.quantile(vals, np.arange(low, high, step))
    shapes = []
    for u in candidates:
        exceed = vals[vals > u] - u
        if len(exceed) < 15:
            shapes.append(np.nan)
            continue
        try:
            c, _, _ = genpareto.fit(exceed, floc=0)
            shapes.append(c)
        except Exception:
            shapes.append(np.nan)
    shapes = np.array(shapes)
    rolling_std = pd.Series(shapes).rolling(window, center=True).std().values
    stable_idx = np.nanargmin(rolling_std)
    selected = candidates[stable_idx]
    diagnostics = pd.DataFrame({
        'candidate_threshold': candidates.round(4),
        'gpd_shape_param': np.round(shapes, 3),
        'rolling_std_of_shape': np.round(rolling_std, 4),
    })
    return selected, diagnostics

# ── Run all four and build the comparison table ───────────────────────
def summarize(name, threshold, durations):
    return {
        'Method': name,
        'Threshold': round(threshold, 4),
        'Num episodes': len(durations),
        'Mean duration': round(durations.mean(), 1) if len(durations) else np.nan,
        'Median duration': int(np.median(durations)) if len(durations) else np.nan,
        'Min / Max': f"{durations.min()}/{durations.max()}" if len(durations) else "-",
    }

rows_gold = []

thr = threshold_percentile(vol_gold, 0.90)
rows_gold.append(summarize('Percentile (90th) -- current', thr, episodes_from_threshold_gold(vol_gold, thr)))

print("Gold Method 3 (log + 3sigma) diagnostics:")
thr = threshold_log_sigma(vol_gold, n_sigma=3)
rows_gold.append(summarize('Log-transform + 3sigma', thr, episodes_from_threshold_gold(vol_gold, thr)))

thr = threshold_mad_zscore(vol_gold, z_cutoff=3.5)
rows_gold.append(summarize('MAD robust z-score (|z|>3.5)', thr, episodes_from_threshold_gold(vol_gold, thr)))

thr, evt_diagnostics_gold = threshold_evt_pot(vol_gold)
rows_gold.append(summarize('EVT / Peaks-over-threshold', thr, episodes_from_threshold_gold(vol_gold, thr)))

comparison_df_gold = pd.DataFrame(rows_gold)
print("\n=== Gold crisis-definition method comparison ===")
print(comparison_df_gold.to_string(index=False))

print("\n=== Gold EVT threshold-stability diagnostics (mean-residual-life-style check) ===")
print(evt_diagnostics_gold.to_string(index=False))
print("\n(Look for the stretch where 'gpd_shape_param' flattens out -- that's the")
print(" region EVT theory says the threshold should sit in; 'rolling_std_of_shape'")
print(" picks the flattest point automatically as the selected threshold above.)")
```

We tried four ways to define a "crisis" for Gold: a straight percentile cutoff, a robust MAD z-score, a log+3σ rule, and EVT. Episode counts ranged from 7 (log+3σ) to 24 (EVT) — a much narrower spread than we saw for WTI or Brent, where EVT's own diagnostic failed outright and the four methods disagreed wildly.

Log+3σ is the outlier here. Its own diagnostic shows the log transform genuinely helps for Gold — skewness drops from 1.81 (raw) to 0.89 (log), a real improvement, unlike WTI where the transform barely moved the needle. But 0.89 still isn't close to symmetric, and the method's result — only 7 episodes, averaging just 8.4 days — is too thin a sample to train a duration model on, whatever its statistical merits.

EVT is the interesting case. Unlike WTI and Brent, where the shape parameter never stabilized and the "selected" threshold was just the least-steep point on a line still climbing, Gold's shape parameter genuinely flattens: it settles around -0.24 across thresholds from roughly 1.32 to 1.49, with the rolling standard deviation of the shape parameter dropping by almost an order of magnitude (to 0.002) right in that band. That's a real plateau, not noise — and the threshold EVT lands on (1.32, 24 episodes) sits right next to our 90th-percentile threshold (1.40, 22 episodes).

That convergence is the strongest argument for keeping the 90th percentile as Gold's crisis definition: it isn't just the simplest, most transparent choice (no distributional assumptions to fail), it's independently corroborated by the one method here that's supposed to find the "true" cutoff and, for Gold, actually does. MAD z-score (14 episodes, threshold 1.64) is a reasonable, more conservative middle ground, but with less data to train on. We keep the 90th percentile — 22 episodes — matching the choice used for Brent and WTI, with stronger cross-method support for Gold than either had.

```python
high_vol_threshold_gold = master_gold['GARCH_VOL_FULL_SAMPLE_GOLD'].quantile(0.90)
master_gold['HIGH_VOL_REGIME_GOLD'] = (master_gold['GARCH_VOL_FULL_SAMPLE_GOLD'] > high_vol_threshold_gold).astype(int)
print("Gold 90th-percentile threshold:", round(high_vol_threshold_gold, 4))

regime = master_gold['HIGH_VOL_REGIME_GOLD']
change = regime.diff()
crisis_starts = master_gold.index[change ==  1]
crisis_ends   = master_gold.index[change == -1]
if regime.iloc[0] == 1:
    crisis_starts = pd.Index([master_gold.index[0]]).append(crisis_starts)

crisis_periods = []
for i, start in enumerate(crisis_starts):
    future_ends = crisis_ends[crisis_ends > start]
    censored = len(future_ends) == 0
    end = future_ends[0] if not censored else master_gold.index[-1]
    duration = len(master_gold.loc[start:end])
    crisis_periods.append({'crisis_id': i+1, 'start_date': start, 'end_date': end,
                            'duration_days': duration, 'log_duration': np.log(duration), 'censored': censored})

crisis_table_full_gold = pd.DataFrame(crisis_periods)
n_censored_gold = crisis_table_full_gold['censored'].sum()
crisis_table_gold = crisis_table_full_gold[~crisis_table_full_gold['censored']].drop(columns='censored').reset_index(drop=True)
print(f"Gold crisis episodes (raw): {len(crisis_table_full_gold)} | Censored: {n_censored_gold} | Final: {len(crisis_table_gold)}")
```

```python
feature_cols_market_gold = ['GVZ', 'DXY', 'FSI', 'IGREA', 'GPR', 'EPU', 'GOLD_REAL', 'GOLD_PRICE_SHOCK']

rows = []
for _, crisis in crisis_table_gold.iterrows():
    start = pd.Timestamp(crisis['start_date'])
    window = master_gold.loc[:start - pd.Timedelta(days=1)].tail(5)
    feat = {col: window[col].mean() for col in feature_cols_market_gold if col in window.columns}

    gold_sub = gold_ret.loc[:start].iloc[:-1]
    if len(gold_sub) < 100:
        feat['GARCH_VOL'] = np.nan
        feat['GARCH_PERSISTENCE'] = np.nan
    else:
        try:
            garch_roll = arch_model(gold_sub, mean='Zero', vol='GARCH', p=1, q=1,
                                     dist='t', rescale=True).fit(disp='off')
            feat['GARCH_VOL'] = float(garch_roll.conditional_volatility[-1])
            feat['GARCH_PERSISTENCE'] = float(garch_roll.params['alpha[1]'] + garch_roll.params['beta[1]'])
        except Exception as e:
            print(f"Crisis #{crisis['crisis_id']} ({start.date()}): eGARCH refit failed — {e}")
            feat['GARCH_VOL'] = np.nan
            feat['GARCH_PERSISTENCE'] = np.nan

    if start in gold_ret.index:
        feat['INITIAL_DROP'] = gold_ret.loc[start]
    else:
        prior = gold_ret.loc[:start]
        feat['INITIAL_DROP'] = prior.iloc[-1] if len(prior) else np.nan

    feat['crisis_id'] = crisis['crisis_id']
    feat['start_date'] = start
    feat['duration_days'] = crisis['duration_days']
    feat['log_duration'] = crisis['log_duration']
    rows.append(feat)

df_model_gold = pd.DataFrame(rows)
n_before = len(df_model_gold)
meta_cols = ['crisis_id', 'start_date', 'duration_days', 'log_duration']
feature_cols_gold = [c for c in df_model_gold.columns if c not in meta_cols]
df_model_gold = df_model_gold.dropna(subset=feature_cols_gold).reset_index(drop=True)
print(f"Dropped {n_before - len(df_model_gold)} row(s). {len(df_model_gold)} remain.")

df_model_gold = df_model_gold.set_index('crisis_id')
X_gold     = df_model_gold[feature_cols_gold].values
y_gold     = df_model_gold['duration_days'].values
y_log_gold = df_model_gold['log_duration'].values
print(f"df_model_gold shape: {df_model_gold.shape} | Features: {feature_cols_gold}")
```

```python
df_model_gold_eval, y_gold_eval, pred_log_gold_eval, pred_raw_gold_eval = run_timeseries_cv(
    df_model_gold, X_gold, y_gold, y_log_gold
)
y_pred_final_gold, target_used_gold, used_log_gold, r2_final_gold = pick_best(
    y_gold_eval, pred_log_gold_eval, pred_raw_gold_eval
)
plot_predictions(df_model_gold_eval, y_gold_eval, y_pred_final_gold, r2_final_gold)
shap_values_gold = plot_shap(xgb.XGBRegressor(**XGB_PARAMS), X_gold,
                              y_log_gold if used_log_gold else y_gold, feature_cols_gold)
results_gold = report_results(df_model_gold_eval, y_gold_eval, y_pred_final_gold)
```

This baseline (full 10-feature set, log-target) scores essentially R²=0.002 on TimeSeriesSplit — no real predictive power, evaluated on 15 of 22 crises (earliest 7 excluded by construction, same as Brent/WTI). Note the raw-target actually scored better here (R²=0.169) but log was still chosen, per the a priori skewness rule (2.106 vs 0.524) decided before looking at performance — consistent with the notebook's discipline elsewhere, even when it doesn't pick the "winning" number this time. SHAP puts GVZ (gold's vol index) clearly in front, followed by GARCH_VOL and DXY — a sensible ranking for a safe-haven asset. The per-crisis table shows the same failure mode Brent/WTI's holdout had: the model badly underpredicts the one long episode (121 days, predicted 16), which is what's dragging R² down — short, ordinary crises look reasonably close.

```python
df_eval_nested_gold, y_nested_gold, pred_nested_gold, selected_per_fold_gold = nested_shap_feature_selection_ts(
    df_model_gold, X_gold, y_log_gold, y_gold, feature_cols_gold, top_n=top_n
)

print("\n=== Gold: nested feature selection, TimeSeriesSplit (honest, leakage-free) ===")
_, _, r2_nested_gold = evaluate(y_nested_gold, pred_nested_gold, "Gold XGB -- nested (honest) feature selection")

from collections import Counter
n_folds_effective_gold = len(selected_per_fold_gold)
selection_counts_gold = Counter(f for fold in selected_per_fold_gold for f in fold)
print(f"\nFeature selection frequency across {n_folds_effective_gold} outer folds:")
for feat, count in sorted(selection_counts_gold.items(), key=lambda kv: -kv[1]):
    print(f"  {feat:<25} selected in {count}/{n_folds_effective_gold} folds")

results_nested_gold = report_results(df_eval_nested_gold, y_nested_gold, pred_nested_gold)
plot_predictions(df_eval_nested_gold, y_nested_gold, pred_nested_gold, r2_nested_gold)
```

Nested SHAP selection didn't help — R² actually got worse (-0.027 vs +0.002 for the full feature set), same pattern seen with WTI's nested SHAP result. Notably, GVZ and GARCH_VOL — the top two SHAP features from the full-feature run — aren't both consistently chosen here: GARCH_VOL is picked in 5/5 folds, but GVZ only 3/5, while GPR (4/5) and GOLD_REAL (4/5) show up more often than the full-model's importance ranking suggested. That instability, plus the R² drop, is a first sign that Gold's 22-crisis sample may be too small for per-fold feature selection to reliably outperform just using the full set — worth comparing against nested VIF-trim next before drawing conclusions.

```python
print("Do the XGBoost models beat simple, non-ML baselines? (Gold)\n")
y_eval_base_gold, pred_mean_gold, pred_median_gold = baseline_mean_median_ts(df_model_gold, y_gold)
print()
pred_persist_gold, valid_persist_gold = baseline_persistence(df_model_gold, y_gold)
```

Every naive baseline beats the model here — mean duration (R²=-0.179), median (R²=-0.265), and persistence (R²=-1.397) all score better than the nested-SHAP model's -0.027. This is stronger than WTI's result (where the model at least beat persistence) — for Gold, the model currently adds no value over just guessing the historical average. Combined with the earlier finding that nested SHAP selection made things worse, this points toward the fixed feature set (R²=0.002) still being the least-bad option so far — worth seeing if VIF-trim (Chunk 5/6) changes that picture before drawing a final conclusion.

```python
from statsmodels.stats.outliers_influence import variance_inflation_factor

corr_features_gold = [c for c in feature_cols_gold if c in df_model_gold.columns]
corr_matrix_gold = df_model_gold[corr_features_gold].corr()

print("=== Gold feature correlation matrix ===")
print(corr_matrix_gold.round(2))

fig, ax = plt.subplots(figsize=(7, 6))
im = ax.imshow(corr_matrix_gold, vmin=-1, vmax=1, cmap='RdBu_r')
ax.set_xticks(range(len(corr_matrix_gold))); ax.set_yticks(range(len(corr_matrix_gold)))
ax.set_xticklabels(corr_matrix_gold.columns, rotation=45, ha='right')
ax.set_yticklabels(corr_matrix_gold.columns)
for i in range(len(corr_matrix_gold)):
    for j in range(len(corr_matrix_gold)):
        val = corr_matrix_gold.iloc[i, j]
        ax.text(j, i, f"{val:.2f}", ha='center', va='center',
                color='white' if abs(val) > 0.5 else 'black', fontsize=8)
plt.colorbar(im, label='Pearson correlation')
plt.title('Gold feature correlation -- volatility/stress proxies')
plt.tight_layout(); plt.show()

X_vif_gold = df_model_gold[corr_features_gold].dropna()
vif_gold = pd.Series(
    [variance_inflation_factor(X_vif_gold.values, i) for i in range(X_vif_gold.shape[1])],
    index=corr_features_gold
).sort_values(ascending=False)
print("\n=== Gold Variance Inflation Factor ===")
print(vif_gold.round(2))
```

```python
df_eval_vif_gold, y_vif_gold, pred_vif_gold, selected_per_fold_vif_gold = nested_vif_trim_ts(
    df_model_gold, X_gold, feature_cols_gold, y_gold, y_log_gold
)

print("\n=== Gold: nested VIF-trim (honest, leakage-free) ===")
_, _, r2_vif_nested_gold = evaluate(y_vif_gold, pred_vif_gold, "Gold XGB -- nested VIF-trim, TimeSeriesSplit")

print(f"\nFeature sets selected per fold:")
for i, feats in enumerate(selected_per_fold_vif_gold):
    print(f"  Fold {i+1}: {feats}")

print(f"\n=== Comparison: full vs. SHAP-nested vs. VIF-nested (Gold) ===")
print(f"Full, fixed features:  R² = {r2_final_gold:.3f}")
print(f"Nested SHAP selection: R² = {r2_nested_gold:.3f}")
print(f"Nested VIF-trim:       R² = {r2_vif_nested_gold:.3f}")

results_vif_gold = report_results(df_eval_vif_gold, y_vif_gold, pred_vif_gold)
```

VIF-trim is the best of the three so far — R²=0.085, beating both the full feature set (0.002) and nested SHAP (-0.027), though still modest. The trim only actually kicks in for the later folds: folds 1-3 have too few training rows (7, 10, 13) to compute stable VIF and fall back to all 11 features, but folds 4-5 genuinely trim down to 5 features (FSI, IGREA, EPU, GOLD_PRICE_SHOCK, INITIAL_DROP) — dropping DXY, GVZ, GOLD_REAL, and both GARCH terms, consistent with DXY being the worst multicollinearity offender in the VIF check. Still nowhere near Brent's 0.255, but this is now the strongest, most defensible candidate for Gold's headline number — worth running bootstrap CI next to see how much this could wobble.

```python
#bootstrap CI on the nested VIF-trim result
from scipy.stats import bootstrap

res_gold = bootstrap(
    (y_vif_gold, pred_vif_gold),
    statistic=lambda yt, yp: r2_score(yt, yp),
    n_resamples=10000, method='BCa', paired=True, random_state=42
)

print(f"Nested VIF-trim R² (Gold) = {r2_vif_nested_gold:.3f}")
print(f"Bootstrap 95% BCa CI: [{res_gold.confidence_interval.low:.3f}, {res_gold.confidence_interval.high:.3f}]  (n_resamples=10000)")
```

CI: [-0.228, 0.586] wide, crosses zero, so we can't rule out the true explanatory power being close to none or negative. This is the same pattern as Brent and WTI's bootstrap results — expected given only 15 evaluated crises. On the positive side, the interval leans more toward positive territory than negative (upper bound 0.586 vs lower -0.228), a slightly better shape than WTI's CI, though not as convincingly positive as Brent's. Next step: permutation test  to check if 0.085 beats random chance specifically, which is a sharper test than the CI alone.

```python
n_permutations = 200
rng = np.random.default_rng(42)
null_r2_gold = []

for _ in range(n_permutations):
    y_shuffled_gold = rng.permutation(y_gold)
    y_shuffled_log_gold = np.log(y_shuffled_gold)
    df_model_shuffled_gold = df_model_gold.copy()
    df_model_shuffled_gold['duration_days'] = y_shuffled_gold
    df_model_shuffled_gold['log_duration']  = y_shuffled_log_gold

    _, y_eval_shuf, pred_shuf, _ = nested_vif_trim_ts(
        df_model_shuffled_gold, X_gold, feature_cols_gold, y_shuffled_gold, y_shuffled_log_gold
    )
    null_r2_gold.append(r2_score(y_eval_shuf, pred_shuf))

null_r2_gold = np.array(null_r2_gold)
print(f"Real R² (nested VIF-trim, Gold): {r2_vif_nested_gold:.3f}")
print(f"Null distribution: mean={null_r2_gold.mean():.3f}, 95th pct={np.percentile(null_r2_gold, 95):.3f}")
print(f"Real R² beats {np.mean(null_r2_gold < r2_vif_nested_gold)*100:.1f}% of random permutations")

n_at_least_as_extreme_gold = np.sum(null_r2_gold >= r2_vif_nested_gold)
p_value_gold = (n_at_least_as_extreme_gold + 1) / (n_permutations + 1)
print(f"Exact permutation p-value: p = {p_value_gold:.4f} (Phipson & Smyth, 2010 correction)")
```

real R²=0.085 beats 98.5% of 200 shuffled permutations, p=0.0199, well under the conventional 0.05 threshold. That's better than what the wide, zero-crossing bootstrap CI alone suggested: even though we can't pin down the size of the effect precisely (small-sample bootstrap noise), the permutation test says it's very unlikely this R² arose by chance. This mirrors Brent's pattern (real signal, imprecise magnitude) more than WTI's (no real signal at all).

```python
#seed sensitivity analysis for gold crises
from sklearn.model_selection import GridSearchCV, KFold
from statsmodels.stats.outliers_influence import variance_inflation_factor

if 'XGB_PARAM_GRID' not in dir():
    XGB_PARAM_GRID = {
        'max_depth': [2, 3, 4], 'learning_rate': [0.03, 0.05, 0.1],
        'n_estimators': [50, 100, 150], 'subsample': [0.7, 0.8, 1.0],
        'colsample_bytree': [0.7, 0.8, 1.0], 'reg_alpha': [0, 0.5, 1.0], 'reg_lambda': [1.0, 2.0],
    }

# single-pass VIF trim, for the search only (mirrors Brent/WTI cells 117/183)
vif_features_gold_single = list(feature_cols_gold)
while True:
    X_vif_check_gold = df_model_gold[vif_features_gold_single].dropna()
    vif_check_gold = pd.Series(
        [variance_inflation_factor(X_vif_check_gold.values, i) for i in range(X_vif_check_gold.shape[1])],
        index=vif_features_gold_single)
    worst_feature, worst_vif = vif_check_gold.idxmax(), vif_check_gold.max()
    if worst_vif <= 10 or len(vif_features_gold_single) <= 2:
        break
    vif_features_gold_single.remove(worst_feature)

feature_cols_trimmed_gold = vif_features_gold_single
print(f"Gold single-pass VIF trim kept: {feature_cols_trimmed_gold}")

order_gold = np.argsort(df_model_gold['start_date'].values)
X_trimmed_gold     = df_model_gold[feature_cols_trimmed_gold].values[order_gold]
y_trimmed_gold     = df_model_gold['duration_days'].values[order_gold]
y_log_trimmed_gold = df_model_gold['log_duration'].values[order_gold]

final_search_gold = GridSearchCV(xgb.XGBRegressor(random_state=42, n_jobs=-1, verbosity=0),
                                  param_grid=XGB_PARAM_GRID, cv=KFold(n_splits=5, shuffle=True, random_state=42),
                                  scoring='neg_mean_squared_error', n_jobs=-1)
final_search_gold.fit(X_trimmed_gold, y_log_trimmed_gold)
print("Best params found (Gold):", final_search_gold.best_params_)

XGB_PARAMS_TUNED_GOLD = dict(final_search_gold.best_params_, random_state=42, n_jobs=-1, verbosity=0)

_, y_gold_final, pred_log_gold_final, _ = run_timeseries_cv(
    df_model_gold, X_trimmed_gold, y_trimmed_gold, y_log_trimmed_gold, xgb_params=XGB_PARAMS_TUNED_GOLD)
_, _, r2_final_tuned_gold = evaluate(y_gold_final, pred_log_gold_final, "Gold -- tuned, honest check")
print(f"Untuned nested VIF-trim: R² = {r2_vif_nested_gold:.3f}")
print(f"Tuned (honest re-check): R² = {r2_final_tuned_gold:.3f}")

seed_results_gold = []
for seed in [7, 42, 123, 2024, 99]:
    search = GridSearchCV(xgb.XGBRegressor(random_state=seed, n_jobs=-1, verbosity=0),
                           param_grid=XGB_PARAM_GRID, cv=KFold(n_splits=5, shuffle=True, random_state=seed),
                           scoring='neg_mean_squared_error', n_jobs=-1)
    search.fit(X_trimmed_gold, y_log_trimmed_gold)
    params_seed = dict(search.best_params_, random_state=42, n_jobs=-1, verbosity=0)
    _, y_seed, pred_seed, _ = run_timeseries_cv(df_model_gold, X_trimmed_gold, y_trimmed_gold, y_log_trimmed_gold, xgb_params=params_seed)
    _, _, r2_seed = evaluate(y_seed, pred_seed, f"seed={seed}")
    seed_results_gold.append({'seed': seed, 'r2_honest': r2_seed})

seed_df_gold = pd.DataFrame(seed_results_gold)
print(seed_df_gold.to_string(index=False))
print(f"Mean R²: {seed_df_gold['r2_honest'].mean():.3f} | Range: [{seed_df_gold['r2_honest'].min():.3f}, {seed_df_gold['r2_honest'].max():.3f}]")
```

Tuning is unreliable for Gold — the seed sweep ranges from -0.104 to 0.162 (mean 0.054), actually below the untuned nested VIF-trim's 0.085, with one seed even going negative. This is a third distinct outcome across the three commodities: Brent's tuning reliably helped, WTI's reliably hurt, Gold's is just noisy. We keep the untuned nested VIF-trim (R²=0.085) as Gold's headline number, not because tuning definitively fails like WTI, but because it doesn't show a dependable improvement worth trusting over the simpler default.

```python
from scipy.stats import wilcoxon

y_naive_gold, pred_naive_gold = naive_mean_baseline_ts(df_model_gold, y_gold)

model_errors_gold = np.abs(y_vif_gold - pred_vif_gold)
naive_errors_gold = np.abs(y_naive_gold - pred_naive_gold)

print(f"Model MAE: {model_errors_gold.mean():.2f} | Naive MAE: {naive_errors_gold.mean():.2f}")
stat, p_value = wilcoxon(model_errors_gold, naive_errors_gold, alternative='less')
print(f"Wilcoxon (model < naive): statistic={stat:.1f}, p={p_value:.4f}")
```

Model beats the naive baseline decisively: MAE 17.95 vs. 27.93 days, Wilcoxon p=0.0045 (well under 0.05). This confirms and sharpens what the permutation test already suggested (p=0.0199) — Gold's nested VIF-trim model isn't just marginally better than chance, it's a statistically significant, practically meaningful improvement over predicting the average. Strongest result in the pipeline so far, closer to Brent's outcome (real signal) than WTI's (no signal).

```python
N_HOLDOUT = 6
df_sorted_gold = df_model_gold.sort_values('start_date').reset_index(drop=True)
holdout_dates_gold = df_sorted_gold['start_date'].iloc[-N_HOLDOUT:].values
holdout_mask_gold = df_model_gold['start_date'].isin(holdout_dates_gold)
dev_mask_gold = ~holdout_mask_gold

print(f"Development set: {dev_mask_gold.sum()} crises | Holdout: {holdout_mask_gold.sum()} crises")
print("Holdout crises (untouched until this cell):")
print(df_model_gold.loc[holdout_mask_gold, ['start_date', 'duration_days']].to_string())

X_dev_gold     = df_model_gold.loc[dev_mask_gold, feature_cols_gold].values
y_dev_gold     = df_model_gold.loc[dev_mask_gold, 'duration_days'].values
y_dev_log_gold = df_model_gold.loc[dev_mask_gold, 'log_duration'].values
X_holdout_gold = df_model_gold.loc[holdout_mask_gold, feature_cols_gold].values
y_holdout_gold = df_model_gold.loc[holdout_mask_gold, 'duration_days'].values

# VIF-based feature selection using DEV data only
dev_df_gold = pd.DataFrame(X_dev_gold, columns=feature_cols_gold)
vif_features_holdout_gold = list(feature_cols_gold)
while len(vif_features_holdout_gold) > 2:
    X_vif_h = dev_df_gold[vif_features_holdout_gold].dropna()
    vif_h = pd.Series([variance_inflation_factor(X_vif_h.values, i) for i in range(X_vif_h.shape[1])],
                       index=vif_features_holdout_gold)
    worst_f, worst_v = vif_h.idxmax(), vif_h.max()
    if worst_v <= 10:
        break
    vif_features_holdout_gold.remove(worst_f)

dev_idx_gold = [feature_cols_gold.index(c) for c in vif_features_holdout_gold]
final_model_gold = xgb.XGBRegressor(**XGB_PARAMS)
final_model_gold.fit(X_dev_gold[:, dev_idx_gold], y_dev_log_gold)
holdout_pred_gold = np.exp(final_model_gold.predict(X_holdout_gold[:, dev_idx_gold]))

_, _, r2_holdout_gold = evaluate(y_holdout_gold, holdout_pred_gold, "Gold -- confirmatory holdout")
print(pd.DataFrame({'start_date': df_model_gold.loc[holdout_mask_gold, 'start_date'].values,
                     'actual_days': y_holdout_gold, 'predicted_days': np.round(holdout_pred_gold).astype(int)}))
```

A positive R²=0.123 on the holdout — encouraging on the surface, but this is essentially the model collapsing toward a single typical prediction (~19-34 days) regardless of the actual duration, rather than genuinely discriminating between crises. It gets the two short ones (3, 2 days) badly wrong in absolute terms (predicting 19 both times) but they're small in scale so they don't hurt R² much; it badly underpredicts the one long crisis (121→29, a 92-day miss, same failure mode Brent/WTI's holdouts showed); and it overpredicts the 16-day one (→29) while doing reasonably on the 49-day one (→34). 

```python
gaps_gold = df_model_gold.sort_values('start_date')['start_date'].diff().dt.days
close_dates_gold = df_model_gold.sort_values('start_date').loc[gaps_gold <= 5, 'start_date']
keep_mask_gold = ~df_model_gold['start_date'].isin(close_dates_gold)

df_model_gold_declustered = df_model_gold.loc[keep_mask_gold].copy()
X_declustered_gold     = df_model_gold.loc[keep_mask_gold, feature_cols_gold].values
y_declustered_gold     = df_model_gold.loc[keep_mask_gold, 'duration_days'].values
y_declustered_log_gold = df_model_gold.loc[keep_mask_gold, 'log_duration'].values

_, y_decl_gold, pred_decl_gold, _ = nested_vif_trim_ts(
    df_model_gold_declustered, X_declustered_gold, feature_cols_gold, y_declustered_gold, y_declustered_log_gold
)
_, _, r2_declustered_gold = evaluate(y_decl_gold, pred_decl_gold, "Gold -- clustered pairs de-duplicated")
print(f"With clustering as-is (N={len(df_model_gold)}): R² = {r2_vif_nested_gold:.3f}")
print(f"With clustered pairs dropped (N={len(df_model_gold_declustered)}): R² = {r2_declustered_gold:.3f}")
```

Dropping the 3 close-pair crises crashes R² from 0.085 to 0.015 — a much bigger sensitivity than Brent's mild 0.255→0.232 dip, showing Gold's result leans heavily on a handful of overlapping-window episodes rather than being a robust, broad-based signal.

```python
censor_lookup_gold = crisis_table_gold.set_index('crisis_id')['end_date'] == master_gold.index[-1]
is_censored_gold = df_model_gold.index.to_series().map(censor_lookup_gold).fillna(False).values

y_lower_gold = df_model_gold['duration_days'].values.astype(float)
y_upper_gold = np.where(is_censored_gold, np.inf, y_lower_gold)
print(f"Right-censored episodes: {is_censored_gold.sum()} of {len(is_censored_gold)}")

df_eval_aft_gold, y_aft_gold, pred_aft_gold = nested_vif_trim_ts_aft(
    df_model_gold, X_gold, feature_cols_gold, y_lower_gold, y_upper_gold, y_gold
)
_, _, r2_aft_gold = evaluate(y_aft_gold, pred_aft_gold, "Gold -- nested VIF-trim AFT, TimeSeriesSplit")

abs_err_plain_gold = np.abs(y_vif_gold - pred_vif_gold)
abs_err_aft_gold   = np.abs(y_aft_gold - pred_aft_gold)

stat, p_value = wilcoxon(abs_err_plain_gold, abs_err_aft_gold, alternative='two-sided')
print(f"Wilcoxon signed-rank, plain vs AFT, n={len(y_aft_gold)}: W={stat:.1f}, p={p_value:.4f}")
print(f"Mean |error| -- Plain: {abs_err_plain_gold.mean():.2f} days | AFT: {abs_err_aft_gold.mean():.2f} days")
```

Gold's case has AFT scoring worse on R² but virtually identical on MAE — likely reflects zero censored episodes here too, same as the others, so AFT's main structural advantage isn't being exercised.

```python
master_hrw = pd.DataFrame({
    'OVX': ovx.squeeze(), 'DXY': dxy.squeeze(), 'FSI': fsi_daily,
    'IGREA': igrea_daily, 'GPR': gpr_daily,
    'HRW_REAL': hrw_real, 'HRW_PRICE_SHOCK': hrw_price_shock,
    'WTI_REAL': wti_real,          # oil price pass-through
    'WHEAT_SUR': wheat_sur_daily,  # global wheat stocks-to-use ratio
}).ffill()

master_hrw['GARCH_VOL_FULL_SAMPLE_HRW'] = garch_vol_full_sample_hrw.reindex(master_hrw.index).ffill()
print("HRW master frame shape:", master_hrw.shape)
print("\nMissing values:\n", master_hrw.isna().sum())
```

```python
def crisis_episode_stats_hrw(vol_series, percentile):
    threshold = vol_series.quantile(percentile)
    regime = (vol_series > threshold).astype(int)
    change = regime.diff()
    starts = vol_series.index[change == 1]
    ends   = vol_series.index[change == -1]
    if regime.iloc[0] == 1:
        starts = starts.insert(0, vol_series.index[0])
    durations = []
    for start in starts:
        future_ends = ends[ends > start]
        if len(future_ends) == 0:
            continue
        durations.append(len(vol_series.loc[start:future_ends[0]]))
    return np.array(durations), threshold

percentiles_to_test = [0.70, 0.75, 0.80, 0.85, 0.90, 0.92, 0.95]
sensitivity_rows_hrw = []
for p in percentiles_to_test:
    durations, thresh = crisis_episode_stats_hrw(master_hrw['GARCH_VOL_FULL_SAMPLE_HRW'].dropna(), p)
    sensitivity_rows_hrw.append({
        'Percentile cutoff': p, 'Volatility threshold': round(thresh, 4),
        'Num episodes': len(durations),
        'Mean duration': round(durations.mean(), 1) if len(durations) else np.nan,
        'Median duration': int(np.median(durations)) if len(durations) else np.nan,
        'Min / Max duration': f"{durations.min()} / {durations.max()}" if len(durations) else "n/a",
    })
sensitivity_df_hrw = pd.DataFrame(sensitivity_rows_hrw)
print("=== HRW: crisis-definition threshold sensitivity ===")
print(sensitivity_df_hrw.to_string(index=False))
```

```python
from scipy.stats import genpareto, skew

vol_hrw = master_hrw['GARCH_VOL_FULL_SAMPLE_HRW'].dropna()

def episodes_from_threshold_hrw(vol_series, threshold):
    regime = (vol_series > threshold).astype(int)
    change = regime.diff()
    starts = vol_series.index[change == 1]
    ends   = vol_series.index[change == -1]
    if regime.iloc[0] == 1:
        starts = starts.insert(0, vol_series.index[0])
    durations = []
    for start in starts:
        future_ends = ends[ends > start]
        if len(future_ends) == 0:
            continue  # right-censored, exclude
        end = future_ends[0]
        durations.append(len(vol_series.loc[start:end]))
    return np.array(durations)

def threshold_percentile(vol_series, p=0.90):
    return vol_series.quantile(p)

def threshold_mad_zscore(vol_series, z_cutoff=3.5):
    median = vol_series.median()
    mad = (vol_series - median).abs().median()
    return median + (z_cutoff / 0.6745) * mad

def threshold_log_sigma(vol_series, n_sigma=3):
    log_vol = np.log(vol_series)
    print(f"    Skewness -- raw: {skew(vol_series):.3f} | log: {skew(log_vol):.3f}")
    mu, sigma = log_vol.mean(), log_vol.std()
    return np.exp(mu + n_sigma * sigma)

def threshold_evt_pot(vol_series, low=0.80, high=0.96, step=0.01, window=3):
    vals = vol_series.values
    candidates = np.quantile(vals, np.arange(low, high, step))
    shapes = []
    for u in candidates:
        exceed = vals[vals > u] - u
        if len(exceed) < 15:
            shapes.append(np.nan)
            continue
        try:
            c, _, _ = genpareto.fit(exceed, floc=0)
            shapes.append(c)
        except Exception:
            shapes.append(np.nan)
    shapes = np.array(shapes)
    rolling_std = pd.Series(shapes).rolling(window, center=True).std().values
    stable_idx = np.nanargmin(rolling_std)
    selected = candidates[stable_idx]
    diagnostics = pd.DataFrame({
        'candidate_threshold': candidates.round(4),
        'gpd_shape_param': np.round(shapes, 3),
        'rolling_std_of_shape': np.round(rolling_std, 4),
    })
    return selected, diagnostics

def summarize(name, threshold, durations):
    return {
        'Method': name, 'Threshold': round(threshold, 4),
        'Num episodes': len(durations),
        'Mean duration': round(durations.mean(), 1) if len(durations) else np.nan,
        'Median duration': int(np.median(durations)) if len(durations) else np.nan,
        'Min / Max': f"{durations.min()}/{durations.max()}" if len(durations) else "-",
    }

rows_hrw = []

thr = threshold_percentile(vol_hrw, 0.90)
rows_hrw.append(summarize('Percentile (90th) -- current', thr, episodes_from_threshold_hrw(vol_hrw, thr)))

print("HRW Method 3 (log + 3sigma) diagnostics:")
thr = threshold_log_sigma(vol_hrw, n_sigma=3)
rows_hrw.append(summarize('Log-transform + 3sigma', thr, episodes_from_threshold_hrw(vol_hrw, thr)))

thr = threshold_mad_zscore(vol_hrw, z_cutoff=3.5)
rows_hrw.append(summarize('MAD robust z-score (|z|>3.5)', thr, episodes_from_threshold_hrw(vol_hrw, thr)))

thr, evt_diagnostics_hrw = threshold_evt_pot(vol_hrw)
rows_hrw.append(summarize('EVT / Peaks-over-threshold', thr, episodes_from_threshold_hrw(vol_hrw, thr)))

comparison_df_hrw = pd.DataFrame(rows_hrw)
print("\n=== HRW crisis-definition method comparison ===")
print(comparison_df_hrw.to_string(index=False))

print("\n=== HRW EVT threshold-stability diagnostics ===")
print(evt_diagnostics_hrw.to_string(index=False))
```

We tried four ways to define a "crisis" for HRW: a straight percentile cutoff, a robust MAD z-score, a log+3σ rule, and EVT. Episode counts ranged from 3 (log+3σ) to 56 (EVT) — a wide spread, closer to what we saw for WTI/Brent than to Gold's unusually tight agreement across methods.

Log+3σ can be set aside despite a real improvement in its underlying assumption — skewness drops from 1.311 (raw) to 0.510 (log), a meaningful reduction — but the method's result (just 3 episodes, averaging 16 days) is far too thin a sample to train a duration model on, whatever its statistical merits. MAD z-score (5 episodes) has the same problem.

EVT is the ambiguous case. It selects a threshold (2.2294, 56 episodes) that sits close to our 90th-percentile threshold (2.2930, 33 episodes) — but unlike Gold, where the shape parameter genuinely plateaued across a wide band of candidate thresholds, HRW's shape sequence is noisy throughout (0.092 → 0.096 → 0.079 → ... → 0.059 → 0.120 → 0.134), and the "selected" point is a local dip in the rolling standard deviation rather than a sustained flat region. So this shouldn't be read as EVT independently corroborating the percentile choice the way it did for Gold — it's closer to WTI/Brent's pattern, where EVT's own diagnostic doesn't actually stabilize.

That leaves the 90th percentile as the practical choice on its own merits: simple, with no fragile distributional assumption to fail, and giving us 33 crisis episodes — a workable sample for the duration model, and comfortably more than the 3-5 episodes the stricter methods would leave us with. We keep this threshold, consistent with Brent, WTI, and Gold.

```python
high_vol_threshold_hrw = master_hrw['GARCH_VOL_FULL_SAMPLE_HRW'].quantile(0.90)
master_hrw['HIGH_VOL_REGIME_HRW'] = (master_hrw['GARCH_VOL_FULL_SAMPLE_HRW'] > high_vol_threshold_hrw).astype(int)
print("HRW 90th-percentile threshold:", round(high_vol_threshold_hrw, 4))

regime = master_hrw['HIGH_VOL_REGIME_HRW']
change = regime.diff()
crisis_starts = master_hrw.index[change ==  1]
crisis_ends   = master_hrw.index[change == -1]
if regime.iloc[0] == 1:
    crisis_starts = pd.Index([master_hrw.index[0]]).append(crisis_starts)

crisis_periods = []
for i, start in enumerate(crisis_starts):
    future_ends = crisis_ends[crisis_ends > start]
    censored = len(future_ends) == 0
    end = future_ends[0] if not censored else master_hrw.index[-1]
    duration = len(master_hrw.loc[start:end])
    crisis_periods.append({'crisis_id': i+1, 'start_date': start, 'end_date': end,
                            'duration_days': duration, 'log_duration': np.log(duration), 'censored': censored})

crisis_table_full_hrw = pd.DataFrame(crisis_periods)
n_censored_hrw = crisis_table_full_hrw['censored'].sum()
crisis_table_hrw = crisis_table_full_hrw[~crisis_table_full_hrw['censored']].drop(columns='censored').reset_index(drop=True)
print(f"HRW crisis episodes (raw): {len(crisis_table_full_hrw)} | Censored: {n_censored_hrw} | Final: {len(crisis_table_hrw)}")
```

```python
feature_cols_market_hrw = ['OVX', 'DXY', 'FSI', 'IGREA', 'GPR',
                            'HRW_REAL', 'HRW_PRICE_SHOCK', 'WTI_REAL', 'WHEAT_SUR']

rows = []
for _, crisis in crisis_table_hrw.iterrows():
    start = pd.Timestamp(crisis['start_date'])
    window = master_hrw.loc[:start - pd.Timedelta(days=1)].tail(5)
    feat = {col: window[col].mean() for col in feature_cols_market_hrw if col in window.columns}

    hrw_sub = hrw_ret.loc[:start].iloc[:-1]
    if len(hrw_sub) < 100:
        feat['GARCH_VOL'] = np.nan
        feat['GARCH_PERSISTENCE'] = np.nan
    else:
        try:
            garch_roll = arch_model(hrw_sub, mean='Zero', vol='GARCH', p=1, q=1,
                                     dist='skewt', rescale=True).fit(disp='off')
            feat['GARCH_VOL'] = float(garch_roll.conditional_volatility[-1])
            feat['GARCH_PERSISTENCE'] = float(garch_roll.params['alpha[1]'] + garch_roll.params['beta[1]'])
        except Exception as e:
            print(f"Crisis #{crisis['crisis_id']} ({start.date()}): GARCH refit failed — {e}")
            feat['GARCH_VOL'] = np.nan
            feat['GARCH_PERSISTENCE'] = np.nan

    # ── Initial drop: same-day (signed) return on crisis start date ──
    if start in hrw_ret.index:
        feat['INITIAL_DROP'] = hrw_ret.loc[start]
    else:
        prior = hrw_ret.loc[:start]
        feat['INITIAL_DROP'] = prior.iloc[-1] if len(prior) else np.nan

    feat['crisis_id'] = crisis['crisis_id']
    feat['start_date'] = start
    feat['duration_days'] = crisis['duration_days']
    feat['log_duration'] = crisis['log_duration']
    rows.append(feat)

df_model_hrw = pd.DataFrame(rows)
n_before = len(df_model_hrw)
meta_cols = ['crisis_id', 'start_date', 'duration_days', 'log_duration']
feature_cols_hrw = [c for c in df_model_hrw.columns if c not in meta_cols]
df_model_hrw = df_model_hrw.dropna(subset=feature_cols_hrw).reset_index(drop=True)
print(f"Dropped {n_before - len(df_model_hrw)} row(s). {len(df_model_hrw)} remain.")

df_model_hrw = df_model_hrw.set_index('crisis_id')
X_hrw     = df_model_hrw[feature_cols_hrw].values
y_hrw     = df_model_hrw['duration_days'].values
y_log_hrw = df_model_hrw['log_duration'].values
print(f"df_model_hrw shape: {df_model_hrw.shape} | Features: {feature_cols_hrw}")
```

```python
df_model_hrw_eval, y_hrw_eval, pred_log_hrw_eval, pred_raw_hrw_eval = run_timeseries_cv(
    df_model_hrw, X_hrw, y_hrw, y_log_hrw
)
y_pred_final_hrw, target_used_hrw, used_log_hrw, r2_final_hrw = pick_best(
    y_hrw_eval, pred_log_hrw_eval, pred_raw_hrw_eval
)
plot_predictions(df_model_hrw_eval, y_hrw_eval, y_pred_final_hrw, r2_final_hrw)
shap_values_hrw = plot_shap(xgb.XGBRegressor(**XGB_PARAMS), X_hrw,
                             y_log_hrw if used_log_hrw else y_hrw, feature_cols_hrw)
results_hrw = report_results(df_model_hrw_eval, y_hrw_eval, y_pred_final_hrw)
```

Full-feature baseline is weak — R²=-0.078 (log-target chosen, skewness 0.819 vs. raw's 3.945, consistent with the a priori rule). Both the scatter and bar chart show the same failure mode we've seen everywhere: the model badly underpredicts the one extreme episode (#22, ~230 days, predicted near 0), while clustering everything else into a narrow low-duration band regardless of actual length — it's not really discriminating, just guessing "short" most of the time. Only 20 of 28 crises get evaluated (8 held out for the earliest training fold, matching the 5-split TS discipline). This is a similar starting point to Gold's messy early stages — the real read comes from nested SHAP and VIF-trim next, not this fixed-feature number.

Nested SHAP barely moves the needle — R²=-0.077, essentially unchanged from the full-feature -0.078. Notably, WHEAT_SUR never gets selected at all (not in the top-8 list), confirming the suspicion: its once-a-year step function gives it little power to distinguish between crises, at least in this fold structure. OVX (5/5) and INITIAL_DROP (4/5) dominate selection, consistent with what's been useful for the other commodities too. Same failure mode as before in the predictions table — crisis #14 (36 days) badly underpredicted at 3, crisis #33 (2 days) overpredicted at 7 — small-sample noise dominating.
```python
print("Do the XGBoost models beat simple, non-ML baselines? (HRW)\n")
y_eval_base_hrw, pred_mean_hrw, pred_median_hrw = baseline_mean_median_ts(df_model_hrw, y_hrw)
print()
pred_persist_hrw, valid_persist_hrw = baseline_persistence(df_model_hrw, y_hrw)
```

The model doesn't beat simple guessing here.
```python
from statsmodels.stats.outliers_influence import variance_inflation_factor

corr_features_hrw = [c for c in feature_cols_hrw if c in df_model_hrw.columns]
corr_matrix_hrw = df_model_hrw[corr_features_hrw].corr()

print("=== HRW feature correlation matrix ===")
print(corr_matrix_hrw.round(2))

fig, ax = plt.subplots(figsize=(7, 6))
im = ax.imshow(corr_matrix_hrw, vmin=-1, vmax=1, cmap='RdBu_r')
ax.set_xticks(range(len(corr_matrix_hrw))); ax.set_yticks(range(len(corr_matrix_hrw)))
ax.set_xticklabels(corr_matrix_hrw.columns, rotation=45, ha='right')
ax.set_yticklabels(corr_matrix_hrw.columns)
for i in range(len(corr_matrix_hrw)):
    for j in range(len(corr_matrix_hrw)):
        val = corr_matrix_hrw.iloc[i, j]
        ax.text(j, i, f"{val:.2f}", ha='center', va='center',
                color='white' if abs(val) > 0.5 else 'black', fontsize=8)
plt.colorbar(im, label='Pearson correlation')
plt.title('HRW feature correlation -- volatility/stress proxies')
plt.tight_layout(); plt.show()

X_vif_hrw = df_model_hrw[corr_features_hrw].dropna()
vif_hrw = pd.Series(
    [variance_inflation_factor(X_vif_hrw.values, i) for i in range(X_vif_hrw.shape[1])],
    index=corr_features_hrw
).sort_values(ascending=False)
print("\n=== HRW Variance Inflation Factor ===")
print(vif_hrw.round(2))
```

Severe multicollinearity across most features — 6 of 12 clear VIF > 100, led by GARCH_PERSISTENCE (2158) and DXY (763), worse than Gold's worst offender. WHEAT_SUR is part of the problem too (0.81 corr with GARCH_PERSISTENCE, -0.74 with WTI_REAL) — likely because it only updates once a year, so it's really just tracking slow year-level drift rather than a genuine relationship. Only FSI, IGREA, HRW_PRICE_SHOCK, and INITIAL_DROP sit under VIF=10. With just 28 crises and 12 features, this is expected small-sample instability — the nested VIF-trim step next will likely strip most of this down to that same small reliable core.

```python
df_eval_vif_hrw, y_vif_hrw, pred_vif_hrw, selected_per_fold_vif_hrw = nested_vif_trim_ts(
    df_model_hrw, X_hrw, feature_cols_hrw, y_hrw, y_log_hrw
)

print("\n=== HRW: nested VIF-trim (honest, leakage-free) ===")
_, _, r2_vif_nested_hrw = evaluate(y_vif_hrw, pred_vif_hrw, "HRW XGB -- nested VIF-trim, TimeSeriesSplit")

print(f"\nFeature sets selected per fold:")
for i, feats in enumerate(selected_per_fold_vif_hrw):
    print(f"  Fold {i+1}: {feats}")

print(f"\n=== Comparison: full vs. SHAP-nested vs. VIF-nested (HRW) ===")
print(f"Full, fixed features:  R² = {r2_final_hrw:.3f}")
print(f"Nested SHAP selection: R² = {r2_nested_hrw:.3f}")
print(f"Nested VIF-trim:       R² = {r2_vif_nested_hrw:.3f}")

results_vif_hrw = report_results(df_eval_vif_hrw, y_vif_hrw, pred_vif_hrw)
```

VIF-trim barely moves the needle — R²=-0.073, essentially flat against SHAP's -0.077 and full-feature's -0.078. This is the WTI pattern, not Gold's: multicollinearity wasn't the real problem, so trimming it away doesn't rescue anything. The trim does correctly land on the same reliable core we've seen survive everywhere else (FSI, IGREA, HRW_PRICE_SHOCK, INITIAL_DROP show up in folds 3-5), confirming the earlier VIF diagnosis was right about which features are redundant — it just turns out removing the redundant ones doesn't translate into real predictive power here. Folds 1-2 still fall back to all 12 features (too few training rows, same as Gold's early folds).

Three different feature-selection approaches landing within 0.005 of each other, all solidly negative, initially looked like a clean "no signal" verdict — but the follow-up checks complicate that. 

```python
from scipy.stats import bootstrap

res_hrw = bootstrap(
    (y_vif_hrw, pred_vif_hrw),
    statistic=lambda yt, yp: r2_score(yt, yp),
    n_resamples=10000, method='BCa', paired=True, random_state=42
)

print(f"Nested VIF-trim R² (HRW) = {r2_vif_nested_hrw:.3f}")
print(f"Bootstrap 95% BCa CI: [{res_hrw.confidence_interval.low:.3f}, {res_hrw.confidence_interval.high:.3f}]  (n_resamples=10000)")
```

Bootstrap 95% CI: [-0.161, 0.287] — wide and crosses zero, but leans more toward positive territory than the negative point estimate (-0.073) alone would suggest. This puts HRW in a genuinely ambiguous spot: not confidently positive like Gold, not as flatly negative as WTI. The permutation test next should help settle whether this R² is actually distinguishable from chance

```python
n_permutations = 200
rng = np.random.default_rng(42)
null_r2_hrw = []

for _ in range(n_permutations):
    y_shuffled_hrw = rng.permutation(y_hrw)
    y_shuffled_log_hrw = np.log(y_shuffled_hrw)
    df_model_shuffled_hrw = df_model_hrw.copy()
    df_model_shuffled_hrw['duration_days'] = y_shuffled_hrw
    df_model_shuffled_hrw['log_duration']  = y_shuffled_log_hrw

    _, y_eval_shuf, pred_shuf, _ = nested_vif_trim_ts(
        df_model_shuffled_hrw, X_hrw, feature_cols_hrw, y_shuffled_hrw, y_shuffled_log_hrw
    )
    null_r2_hrw.append(r2_score(y_eval_shuf, pred_shuf))

null_r2_hrw = np.array(null_r2_hrw)
print(f"Real R² (nested VIF-trim, HRW): {r2_vif_nested_hrw:.3f}")
print(f"Null distribution: mean={null_r2_hrw.mean():.3f}, 95th pct={np.percentile(null_r2_hrw, 95):.3f}")
print(f"Real R² beats {np.mean(null_r2_hrw < r2_vif_nested_hrw)*100:.1f}% of random permutations")

n_at_least_as_extreme_hrw = np.sum(null_r2_hrw >= r2_vif_nested_hrw)
p_value_hrw = (n_at_least_as_extreme_hrw + 1) / (n_permutations + 1)
print(f"Exact permutation p-value: p = {p_value_hrw:.4f} (Phipson & Smyth, 2010 correction)")
```

Not significant — real R² beats 77.5% of permutations, p=0.2289 (well above 0.05). But worth noting the null distribution's mean is -0.394, clearly worse than HRW's -0.073, so it's not performing like pure noise either. This lands HRW in a genuine middle ground: weaker than Gold's real signal (p=0.02), but not as flatly null as WTI's result — an honest "inconclusive" rather than a clean yes or no.

```python
from sklearn.model_selection import GridSearchCV, KFold
from statsmodels.stats.outliers_influence import variance_inflation_factor

# single-pass VIF trim, for the search only
vif_features_hrw_single = list(feature_cols_hrw)
while True:
    X_vif_check_hrw = df_model_hrw[vif_features_hrw_single].dropna()
    vif_check_hrw = pd.Series(
        [variance_inflation_factor(X_vif_check_hrw.values, i) for i in range(X_vif_check_hrw.shape[1])],
        index=vif_features_hrw_single)
    worst_feature, worst_vif = vif_check_hrw.idxmax(), vif_check_hrw.max()
    if worst_vif <= 10 or len(vif_features_hrw_single) <= 2:
        break
    vif_features_hrw_single.remove(worst_feature)

feature_cols_trimmed_hrw = vif_features_hrw_single
print(f"HRW single-pass VIF trim kept: {feature_cols_trimmed_hrw}")

order_hrw = np.argsort(df_model_hrw['start_date'].values)
X_trimmed_hrw     = df_model_hrw[feature_cols_trimmed_hrw].values[order_hrw]
y_trimmed_hrw     = df_model_hrw['duration_days'].values[order_hrw]
y_log_trimmed_hrw = df_model_hrw['log_duration'].values[order_hrw]

final_search_hrw = GridSearchCV(
    xgb.XGBRegressor(random_state=42, n_jobs=-1, verbosity=0),
    param_grid=XGB_PARAM_GRID, cv=KFold(n_splits=5, shuffle=True, random_state=42),
    scoring='neg_mean_squared_error', n_jobs=-1
)
final_search_hrw.fit(X_trimmed_hrw, y_log_trimmed_hrw)
print("Best params found (HRW):", final_search_hrw.best_params_)

XGB_PARAMS_TUNED_HRW = dict(final_search_hrw.best_params_, random_state=42, n_jobs=-1, verbosity=0)

_, y_hrw_final, pred_log_hrw_final, _ = run_timeseries_cv(
    df_model_hrw, X_trimmed_hrw, y_trimmed_hrw, y_log_trimmed_hrw, xgb_params=XGB_PARAMS_TUNED_HRW)
_, _, r2_final_tuned_hrw = evaluate(y_hrw_final, pred_log_hrw_final, "HRW -- tuned, honest check")
print(f"\nUntuned nested VIF-trim: R² = {r2_vif_nested_hrw:.3f}")
print(f"Tuned (honest re-check): R² = {r2_final_tuned_hrw:.3f}")

seed_results_hrw = []
for seed in [7, 42, 123, 2024, 99]:
    search = GridSearchCV(xgb.XGBRegressor(random_state=seed, n_jobs=-1, verbosity=0),
                           param_grid=XGB_PARAM_GRID, cv=KFold(n_splits=5, shuffle=True, random_state=seed),
                           scoring='neg_mean_squared_error', n_jobs=-1)
    search.fit(X_trimmed_hrw, y_log_trimmed_hrw)
    params_seed = dict(search.best_params_, random_state=42, n_jobs=-1, verbosity=0)
    _, y_seed, pred_seed, _ = run_timeseries_cv(df_model_hrw, X_trimmed_hrw, y_trimmed_hrw, y_log_trimmed_hrw, xgb_params=params_seed)
    _, _, r2_seed = evaluate(y_seed, pred_seed, f"seed={seed}")
    seed_results_hrw.append({'seed': seed, 'r2_honest': r2_seed})

seed_df_hrw = pd.DataFrame(seed_results_hrw)
print(seed_df_hrw.to_string(index=False))
print(f"Mean R²: {seed_df_hrw['r2_honest'].mean():.3f} | Range: [{seed_df_hrw['r2_honest'].min():.3f}, {seed_df_hrw['r2_honest'].max():.3f}]")
```

Tuning doesn't help — seeds cluster tightly at [-0.080, -0.057], mean -0.065, essentially matching the untuned nested VIF-trim's -0.073. Every seed stays solidly negative, so unlike Brent this isn't seed-dependent luck; it's a consistent non-improvement. we keep the untuned R²=-0.073 as HRW's headline.

```python
N_HOLDOUT = 6
df_sorted_hrw = df_model_hrw.sort_values('start_date').reset_index(drop=True)
holdout_dates_hrw = df_sorted_hrw['start_date'].iloc[-N_HOLDOUT:].values
holdout_mask_hrw = df_model_hrw['start_date'].isin(holdout_dates_hrw)
dev_mask_hrw = ~holdout_mask_hrw

print(f"Development set: {dev_mask_hrw.sum()} crises | Holdout: {holdout_mask_hrw.sum()} crises")
print("Holdout crises (untouched until this cell):")
print(df_model_hrw.loc[holdout_mask_hrw, ['start_date', 'duration_days']].to_string())

X_dev_hrw     = df_model_hrw.loc[dev_mask_hrw, feature_cols_hrw].values
y_dev_hrw     = df_model_hrw.loc[dev_mask_hrw, 'duration_days'].values
y_dev_log_hrw = df_model_hrw.loc[dev_mask_hrw, 'log_duration'].values
X_holdout_hrw = df_model_hrw.loc[holdout_mask_hrw, feature_cols_hrw].values
y_holdout_hrw = df_model_hrw.loc[holdout_mask_hrw, 'duration_days'].values

# VIF-based feature selection using DEV data only
dev_df_hrw = pd.DataFrame(X_dev_hrw, columns=feature_cols_hrw)
vif_features_holdout_hrw = list(feature_cols_hrw)
while len(vif_features_holdout_hrw) > 2:
    X_vif_h = dev_df_hrw[vif_features_holdout_hrw].dropna()
    vif_h = pd.Series([variance_inflation_factor(X_vif_h.values, i) for i in range(X_vif_h.shape[1])],
                       index=vif_features_holdout_hrw)
    worst_f, worst_v = vif_h.idxmax(), vif_h.max()
    if worst_v <= 10:
        break
    vif_features_holdout_hrw.remove(worst_f)

dev_idx_hrw = [feature_cols_hrw.index(c) for c in vif_features_holdout_hrw]
final_model_hrw = xgb.XGBRegressor(**XGB_PARAMS)
final_model_hrw.fit(X_dev_hrw[:, dev_idx_hrw], y_dev_log_hrw)
holdout_pred_hrw = np.exp(final_model_hrw.predict(X_holdout_hrw[:, dev_idx_hrw]))

_, _, r2_holdout_hrw = evaluate(y_holdout_hrw, holdout_pred_hrw, "HRW -- confirmatory holdout")
print(pd.DataFrame({'start_date': df_model_hrw.loc[holdout_mask_hrw, 'start_date'].values,
                     'actual_days': y_holdout_hrw, 'predicted_days': np.round(holdout_pred_hrw).astype(int)}))
```

The model does badly on this final test too. When we hold back the 6 most recent crises and see how well it predicts them without having seen them at all, it basically just guesses a similar number every time — somewhere around 7 to 10 days — no matter how long the crisis actually lasted. Two of those crises really did last only 2 days, so those guesses happen to land somewhat close by luck. But one crisis lasted 24 days, and the model still predicted just 8 — way off. So rather than actually telling short crises apart from long ones, it seems to be defaulting to "a typical crisis lasts about a week or so" and sticking with that regardless of the real signals. With only 22 crises to learn from, that's not shocking, but it fits the overall picture: nothing here suggests the model has found a genuine, reliable pattern for HRW.

```python
gaps_hrw = df_model_hrw.sort_values('start_date')['start_date'].diff().dt.days
close_dates_hrw = df_model_hrw.sort_values('start_date').loc[gaps_hrw <= 5, 'start_date']
keep_mask_hrw = ~df_model_hrw['start_date'].isin(close_dates_hrw)

df_model_hrw_declustered = df_model_hrw.loc[keep_mask_hrw].copy()
_, y_decl_hrw, pred_decl_hrw, _ = nested_vif_trim_ts(
    df_model_hrw_declustered, df_model_hrw.loc[keep_mask_hrw, feature_cols_hrw].values, feature_cols_hrw,
    df_model_hrw.loc[keep_mask_hrw, 'duration_days'].values, df_model_hrw.loc[keep_mask_hrw, 'log_duration'].values
)
_, _, r2_declustered_hrw = evaluate(y_decl_hrw, pred_decl_hrw, "HRW -- declustered")
print(f"As-is (N={len(df_model_hrw)}): R² = {r2_vif_nested_hrw:.3f}")
print(f"Declustered (N={len(df_model_hrw_declustered)}): R² = {r2_declustered_hrw:.3f}")
```

Dropping the 3 crises that started close together barely changes anything — R² goes from -0.073 to -0.065, basically the same. This is different from Gold, where removing those overlapping crises caused a big drop, meaning Gold's result depended heavily on them. For HRW, the model performs about equally poorly whether those crises are included or not — so at least this weak result isn't being artificially propped up by a few coincidentally-timed episodes. It's just consistently unremarkable either way.

```python
censor_lookup_hrw = crisis_table_hrw.set_index('crisis_id')['end_date'] == master_hrw.index[-1]
is_censored_hrw = df_model_hrw.index.to_series().map(censor_lookup_hrw).fillna(False).values

y_lower_hrw = df_model_hrw['duration_days'].values.astype(float)
y_upper_hrw = np.where(is_censored_hrw, np.inf, y_lower_hrw)
print(f"Right-censored episodes: {is_censored_hrw.sum()} of {len(is_censored_hrw)}")

df_eval_aft_hrw, y_aft_hrw, pred_aft_hrw = nested_vif_trim_ts_aft(
    df_model_hrw, X_hrw, feature_cols_hrw, y_lower_hrw, y_upper_hrw, y_hrw
)
_, _, r2_aft_hrw = evaluate(y_aft_hrw, pred_aft_hrw, "HRW -- nested VIF-trim AFT, TimeSeriesSplit")

abs_err_plain_hrw = np.abs(y_vif_hrw - pred_vif_hrw)
abs_err_aft_hrw   = np.abs(y_aft_hrw - pred_aft_hrw)

stat, p_value = wilcoxon(abs_err_plain_hrw, abs_err_aft_hrw, alternative='two-sided')
print(f"Wilcoxon signed-rank, plain vs AFT, n={len(y_aft_hrw)}: W={stat:.1f}, p={p_value:.4f}")
print(f"Mean |error| -- Plain: {abs_err_plain_hrw.mean():.2f} days | AFT: {abs_err_aft_hrw.mean():.2f} days")
```

```python
cpi_aligned_srw = cpi_daily.reindex(srw_price.index, method='ffill')
srw_real = (srw_price.squeeze() / cpi_aligned_srw) * 100
srw_real.name = "SRW_REAL"

log_price_srw   = np.log(srw_price.squeeze())
rolling_max_srw = log_price_srw.rolling(window=252).max().shift(1)
srw_price_shock = np.maximum(0, log_price_srw - rolling_max_srw)
srw_price_shock.name = "SRW_PRICE_SHOCK"

srw_ret = np.log(srw_price.squeeze()).diff().dropna() * 100

garch_res_srw = arch_model(srw_ret, mean='Zero', vol='GARCH', p=1, q=1,
                            dist='skewt', rescale=True).fit(disp='off')

garch_vol_full_sample_srw = pd.Series(garch_res_srw.conditional_volatility,
                                       index=srw_ret.index,
                                       name='GARCH_VOL_FULL_SAMPLE_SRW')
print(f"Fitted persistence (alpha+beta): {garch_res_srw.params['alpha[1]'] + garch_res_srw.params['beta[1]']:.4f}")
```

```python
master_srw = pd.DataFrame({
    'OVX': ovx.squeeze(), 'DXY': dxy.squeeze(), 'FSI': fsi_daily,
    'IGREA': igrea_daily, 'GPR': gpr_daily,
    'SRW_REAL': srw_real, 'SRW_PRICE_SHOCK': srw_price_shock,
    'WTI_REAL': wti_real,          # oil price pass-through
    'WHEAT_SUR': wheat_sur_daily,  # global wheat stocks-to-use ratio
}).ffill()

master_srw['GARCH_VOL_FULL_SAMPLE_SRW'] = garch_vol_full_sample_srw.reindex(master_srw.index).ffill()
print("SRW master frame shape:", master_srw.shape)
print("\nMissing values:\n", master_srw.isna().sum())
```

```python
def crisis_episode_stats_srw(vol_series, percentile):
    threshold = vol_series.quantile(percentile)
    regime = (vol_series > threshold).astype(int)
    change = regime.diff()
    starts = vol_series.index[change == 1]
    ends   = vol_series.index[change == -1]
    if regime.iloc[0] == 1:
        starts = starts.insert(0, vol_series.index[0])
    durations = []
    for start in starts:
        future_ends = ends[ends > start]
        if len(future_ends) == 0:
            continue
        durations.append(len(vol_series.loc[start:future_ends[0]]))
    return np.array(durations), threshold

percentiles_to_test = [0.70, 0.75, 0.80, 0.85, 0.90, 0.92, 0.95]
sensitivity_rows_srw = []
for p in percentiles_to_test:
    durations, thresh = crisis_episode_stats_srw(master_srw['GARCH_VOL_FULL_SAMPLE_SRW'].dropna(), p)
    sensitivity_rows_srw.append({
        'Percentile cutoff': p, 'Volatility threshold': round(thresh, 4),
        'Num episodes': len(durations),
        'Mean duration': round(durations.mean(), 1) if len(durations) else np.nan,
        'Median duration': int(np.median(durations)) if len(durations) else np.nan,
        'Min / Max duration': f"{durations.min()} / {durations.max()}" if len(durations) else "n/a",
    })
sensitivity_df_srw = pd.DataFrame(sensitivity_rows_srw)
print("=== SRW: crisis-definition threshold sensitivity ===")
print(sensitivity_df_srw.to_string(index=False))
```

```python
from scipy.stats import genpareto, skew

vol_srw = master_srw['GARCH_VOL_FULL_SAMPLE_SRW'].dropna()

def episodes_from_threshold_srw(vol_series, threshold):
    regime = (vol_series > threshold).astype(int)
    change = regime.diff()
    starts = vol_series.index[change == 1]
    ends   = vol_series.index[change == -1]
    if regime.iloc[0] == 1:
        starts = starts.insert(0, vol_series.index[0])
    durations = []
    for start in starts:
        future_ends = ends[ends > start]
        if len(future_ends) == 0:
            continue
        end = future_ends[0]
        durations.append(len(vol_series.loc[start:end]))
    return np.array(durations)

def threshold_percentile(vol_series, p=0.90):
    return vol_series.quantile(p)

def threshold_mad_zscore(vol_series, z_cutoff=3.5):
    median = vol_series.median()
    mad = (vol_series - median).abs().median()
    return median + (z_cutoff / 0.6745) * mad

def threshold_log_sigma(vol_series, n_sigma=3):
    log_vol = np.log(vol_series)
    print(f"    Skewness -- raw: {skew(vol_series):.3f} | log: {skew(log_vol):.3f}")
    mu, sigma = log_vol.mean(), log_vol.std()
    return np.exp(mu + n_sigma * sigma)

def threshold_evt_pot(vol_series, low=0.80, high=0.96, step=0.01, window=3):
    vals = vol_series.values
    candidates = np.quantile(vals, np.arange(low, high, step))
    shapes = []
    for u in candidates:
        exceed = vals[vals > u] - u
        if len(exceed) < 15:
            shapes.append(np.nan)
            continue
        try:
            c, _, _ = genpareto.fit(exceed, floc=0)
            shapes.append(c)
        except Exception:
            shapes.append(np.nan)
    shapes = np.array(shapes)
    rolling_std = pd.Series(shapes).rolling(window, center=True).std().values
    stable_idx = np.nanargmin(rolling_std)
    selected = candidates[stable_idx]
    diagnostics = pd.DataFrame({
        'candidate_threshold': candidates.round(4),
        'gpd_shape_param': np.round(shapes, 3),
        'rolling_std_of_shape': np.round(rolling_std, 4),
    })
    return selected, diagnostics

def summarize(name, threshold, durations):
    return {
        'Method': name, 'Threshold': round(threshold, 4),
        'Num episodes': len(durations),
        'Mean duration': round(durations.mean(), 1) if len(durations) else np.nan,
        'Median duration': int(np.median(durations)) if len(durations) else np.nan,
        'Min / Max': f"{durations.min()}/{durations.max()}" if len(durations) else "-",
    }

rows_srw = []

thr = threshold_percentile(vol_srw, 0.90)
rows_srw.append(summarize('Percentile (90th) -- current', thr, episodes_from_threshold_srw(vol_srw, thr)))

print("SRW Method 3 (log + 3sigma) diagnostics:")
thr = threshold_log_sigma(vol_srw, n_sigma=3)
rows_srw.append(summarize('Log-transform + 3sigma', thr, episodes_from_threshold_srw(vol_srw, thr)))

thr = threshold_mad_zscore(vol_srw, z_cutoff=3.5)
rows_srw.append(summarize('MAD robust z-score (|z|>3.5)', thr, episodes_from_threshold_srw(vol_srw, thr)))

thr, evt_diagnostics_srw = threshold_evt_pot(vol_srw)
rows_srw.append(summarize('EVT / Peaks-over-threshold', thr, episodes_from_threshold_srw(vol_srw, thr)))

comparison_df_srw = pd.DataFrame(rows_srw)
print("\n=== SRW crisis-definition method comparison ===")
print(comparison_df_srw.to_string(index=False))

print("\n=== SRW EVT threshold-stability diagnostics ===")
print(evt_diagnostics_srw.to_string(index=False))
```

We tried four ways to define a "crisis" for SRW: a straight percentile cutoff, a robust MAD z-score, a log+3σ rule, and EVT. Episode counts ranged from 2 (log+3σ) to 57 (EVT) — and unlike HRW or Gold, SRW's crisis count is unusually sensitive to the exact threshold chosen: sweeping the percentile cutoff from 0.70 to 0.95 alone moves the count from 76 down to 27, a much wider swing than we saw for any other series.

Log+3σ can be ruled out again, despite showing the largest skewness improvement of any series we've tested (2.919 raw → 0.965 log) — its result is just 2 episodes, nowhere near enough to train a duration model on. MAD z-score (10 episodes) has the same problem, just less severely.

EVT selects a threshold (2.1977, 57 episodes) close to the 90th percentile's (2.4911, 45 episodes) — but the shape-parameter sequence never stabilizes, climbing steadily from 0.170 to 0.497 across the full candidate range with no flattening point. This is the same pattern WTI, Brent, and HRW showed: EVT's own diagnostic doesn't confirm a genuine threshold here, so its proximity to the percentile choice is coincidental rather than corroborating.

That leaves the 90th percentile as the practical choice, same reasoning as every other series: simple, no fragile distributional assumption, and — notably — 45 episodes is the largest usable sample of any commodity in this analysis so far, more than Brent (45), comparable to it, and well above Gold (22), HRW (28), or WTI (42). We keep this threshold, consistent with the rest of the pipeline.

```python
high_vol_threshold_srw = master_srw['GARCH_VOL_FULL_SAMPLE_SRW'].quantile(0.90)
master_srw['HIGH_VOL_REGIME_SRW'] = (master_srw['GARCH_VOL_FULL_SAMPLE_SRW'] > high_vol_threshold_srw).astype(int)
print("SRW 90th-percentile threshold:", round(high_vol_threshold_srw, 4))

regime = master_srw['HIGH_VOL_REGIME_SRW']
change = regime.diff()
crisis_starts = master_srw.index[change ==  1]
crisis_ends   = master_srw.index[change == -1]
if regime.iloc[0] == 1:
    crisis_starts = pd.Index([master_srw.index[0]]).append(crisis_starts)

crisis_periods = []
for i, start in enumerate(crisis_starts):
    future_ends = crisis_ends[crisis_ends > start]
    censored = len(future_ends) == 0
    end = future_ends[0] if not censored else master_srw.index[-1]
    duration = len(master_srw.loc[start:end])
    crisis_periods.append({'crisis_id': i+1, 'start_date': start, 'end_date': end,
                            'duration_days': duration, 'log_duration': np.log(duration), 'censored': censored})

crisis_table_full_srw = pd.DataFrame(crisis_periods)
n_censored_srw = crisis_table_full_srw['censored'].sum()
crisis_table_srw = crisis_table_full_srw[~crisis_table_full_srw['censored']].drop(columns='censored').reset_index(drop=True)
print(f"SRW crisis episodes (raw): {len(crisis_table_full_srw)} | Censored: {n_censored_srw} | Final: {len(crisis_table_srw)}")
```

```python
feature_cols_market_srw = ['OVX', 'DXY', 'FSI', 'IGREA', 'GPR',
                            'SRW_REAL', 'SRW_PRICE_SHOCK', 'WTI_REAL', 'WHEAT_SUR']

rows = []
for _, crisis in crisis_table_srw.iterrows():
    start = pd.Timestamp(crisis['start_date'])
    window = master_srw.loc[:start - pd.Timedelta(days=1)].tail(5)
    feat = {col: window[col].mean() for col in feature_cols_market_srw if col in window.columns}

    srw_sub = srw_ret.loc[:start].iloc[:-1]
    if len(srw_sub) < 100:
        feat['GARCH_VOL'] = np.nan
        feat['GARCH_PERSISTENCE'] = np.nan
    else:
        try:
            garch_roll = arch_model(srw_sub, mean='Zero', vol='GARCH', p=1, q=1,
                                     dist='skewt', rescale=True).fit(disp='off')
            feat['GARCH_VOL'] = float(garch_roll.conditional_volatility[-1])
            feat['GARCH_PERSISTENCE'] = float(garch_roll.params['alpha[1]'] + garch_roll.params['beta[1]'])
        except Exception as e:
            print(f"Crisis #{crisis['crisis_id']} ({start.date()}): GARCH refit failed — {e}")
            feat['GARCH_VOL'] = np.nan
            feat['GARCH_PERSISTENCE'] = np.nan

    # ── Initial drop: same-day (signed) return on crisis start date ──
    if start in srw_ret.index:
        feat['INITIAL_DROP'] = srw_ret.loc[start]
    else:
        prior = srw_ret.loc[:start]
        feat['INITIAL_DROP'] = prior.iloc[-1] if len(prior) else np.nan

    feat['crisis_id'] = crisis['crisis_id']
    feat['start_date'] = start
    feat['duration_days'] = crisis['duration_days']
    feat['log_duration'] = crisis['log_duration']
    rows.append(feat)

df_model_srw = pd.DataFrame(rows)
n_before = len(df_model_srw)
meta_cols = ['crisis_id', 'start_date', 'duration_days', 'log_duration']
feature_cols_srw = [c for c in df_model_srw.columns if c not in meta_cols]
df_model_srw = df_model_srw.dropna(subset=feature_cols_srw).reset_index(drop=True)
print(f"Dropped {n_before - len(df_model_srw)} row(s). {len(df_model_srw)} remain.")

df_model_srw = df_model_srw.set_index('crisis_id')
X_srw     = df_model_srw[feature_cols_srw].values
y_srw     = df_model_srw['duration_days'].values
y_log_srw = df_model_srw['log_duration'].values
print(f"df_model_srw shape: {df_model_srw.shape} | Features: {feature_cols_srw}")
```

```python
df_model_srw_eval, y_srw_eval, pred_log_srw_eval, pred_raw_srw_eval = run_timeseries_cv(
    df_model_srw, X_srw, y_srw, y_log_srw
)
y_pred_final_srw, target_used_srw, used_log_srw, r2_final_srw = pick_best(
    y_srw_eval, pred_log_srw_eval, pred_raw_srw_eval
)
plot_predictions(df_model_srw_eval, y_srw_eval, y_pred_final_srw, r2_final_srw)
shap_values_srw = plot_shap(xgb.XGBRegressor(**XGB_PARAMS), X_srw,
                             y_log_srw if used_log_srw else y_srw, feature_cols_srw)
results_srw = report_results(df_model_srw_eval, y_srw_eval, y_pred_final_srw)
```

Weaker start than HRW — R²=-0.314 (log-target chosen, skew 1.847→0.665). Only 25 of 35 crises get evaluated after the TS split. SHAP is led by SRW_REAL and INITIAL_DROP (nearly tied, ~0.31 each) — closer to Brent/WTI's pattern than HRW's OVX-led ranking. WHEAT_SUR again ranks dead last (0.0022) — confirms across both wheat series it's not adding real information. Next: nested SHAP and VIF-trim to see if either rescues this

```python
df_eval_nested_srw, y_nested_srw, pred_nested_srw, selected_per_fold_srw = nested_shap_feature_selection_ts(
    df_model_srw, X_srw, y_log_srw, y_srw, feature_cols_srw, top_n=top_n
)

print("\n=== SRW: nested feature selection, TimeSeriesSplit (honest, leakage-free) ===")
_, _, r2_nested_srw = evaluate(y_nested_srw, pred_nested_srw, "SRW XGB -- nested (honest) feature selection")

from collections import Counter
n_folds_effective_srw = len(selected_per_fold_srw)
selection_counts_srw = Counter(f for fold in selected_per_fold_srw for f in fold)
print(f"\nFeature selection frequency across {n_folds_effective_srw} outer folds:")
for feat, count in sorted(selection_counts_srw.items(), key=lambda kv: -kv[1]):
    print(f"  {feat:<25} selected in {count}/{n_folds_effective_srw} folds")

results_nested_srw = report_results(df_eval_nested_srw, y_nested_srw, pred_nested_srw)
plot_predictions(df_eval_nested_srw, y_nested_srw, pred_nested_srw, r2_nested_srw)
```

Small improvement over the full-feature baseline — R²=-0.263 vs -0.314. INITIAL_DROP stays consistent (4/5 folds), but GARCH_PERSISTENCE jumps to the top (4/5), edging out SRW_REAL (down to 3/5) — a somewhat different mix than the full-feature SHAP ranking suggested. The predictions table shows the same recurring failure mode across every commodity: short 2-7 day crises get badly overpredicted (12-25 days), while the one genuinely long episode (63 days) gets badly underpredicted (7 days) — the model keeps defaulting toward a "typical" mid-length guess instead of actually discriminating extremes. Worth seeing if VIF-trim does any better before drawing conclusions.

```python
from statsmodels.stats.outliers_influence import variance_inflation_factor

corr_features_srw = [c for c in feature_cols_srw if c in df_model_srw.columns]
corr_matrix_srw = df_model_srw[corr_features_srw].corr()

print("=== SRW feature correlation matrix ===")
print(corr_matrix_srw.round(2))

fig, ax = plt.subplots(figsize=(7, 6))
im = ax.imshow(corr_matrix_srw, vmin=-1, vmax=1, cmap='RdBu_r')
ax.set_xticks(range(len(corr_matrix_srw))); ax.set_yticks(range(len(corr_matrix_srw)))
ax.set_xticklabels(corr_matrix_srw.columns, rotation=45, ha='right')
ax.set_yticklabels(corr_matrix_srw.columns)
for i in range(len(corr_matrix_srw)):
    for j in range(len(corr_matrix_srw)):
        val = corr_matrix_srw.iloc[i, j]
        ax.text(j, i, f"{val:.2f}", ha='center', va='center',
                color='white' if abs(val) > 0.5 else 'black', fontsize=8)
plt.colorbar(im, label='Pearson correlation')
plt.title('SRW feature correlation -- volatility/stress proxies')
plt.tight_layout(); plt.show()

X_vif_srw = df_model_srw[corr_features_srw].dropna()
vif_srw = pd.Series(
    [variance_inflation_factor(X_vif_srw.values, i) for i in range(X_vif_srw.shape[1])],
    index=corr_features_srw
).sort_values(ascending=False)
print("\n=== SRW Variance Inflation Factor ===")
print(vif_srw.round(2))
```

```python
df_eval_vif_srw, y_vif_srw, pred_vif_srw, selected_per_fold_vif_srw = nested_vif_trim_ts(
    df_model_srw, X_srw, feature_cols_srw, y_srw, y_log_srw
)

print("\n=== SRW: nested VIF-trim (honest, leakage-free) ===")
_, _, r2_vif_nested_srw = evaluate(y_vif_srw, pred_vif_srw, "SRW XGB -- nested VIF-trim, TimeSeriesSplit")

print(f"\nFeature sets selected per fold:")
for i, feats in enumerate(selected_per_fold_vif_srw):
    print(f"  Fold {i+1}: {feats}")

print(f"\n=== Comparison: full vs. SHAP-nested vs. VIF-nested (SRW) ===")
print(f"Full, fixed features:  R² = {r2_final_srw:.3f}")
print(f"Nested SHAP selection: R² = {r2_nested_srw:.3f}")
print(f"Nested VIF-trim:       R² = {r2_vif_nested_srw:.3f}")

results_vif_srw = report_results(df_eval_vif_srw, y_vif_srw, pred_vif_srw)
```

VIF-trim gives the best result of the three — R²=-0.208, improving steadily from full-feature (-0.314) through nested SHAP (-0.263). Folds 3-5 converge on the same reliable core seen elsewhere (FSI, IGREA, SRW_PRICE_SHOCK, INITIAL_DROP), plus WTI_REAL here. Still solidly negative and the 63-day crisis remains badly underpredicted, but the consistent step-by-step improvement (unlike HRW's flat ~-0.07 across all three methods) makes this worth checking against the bootstrap CI and permutation test next.

```python
from scipy.stats import bootstrap

res_srw = bootstrap(
    (y_vif_srw, pred_vif_srw),
    statistic=lambda yt, yp: r2_score(yt, yp),
    n_resamples=10000, method='BCa', paired=True, random_state=42
)

print(f"Nested VIF-trim R² (SRW) = {r2_vif_nested_srw:.3f}")
print(f"Bootstrap 95% BCa CI: [{res_srw.confidence_interval.low:.3f}, {res_srw.confidence_interval.high:.3f}]  (n_resamples=10000)")
```

This is actually the most decisive negative result we've seen — the entire 95% CI sits below zero ([-0.400, -0.019]), unlike every other commodity where the interval crossed zero. So unlike Brent (positive lean), Gold (positive lean), WTI (crosses zero), or HRW (crosses zero, slightly positive-leaning), SRW's confidence interval rules out a genuinely positive effect with reasonable confidence. This aligns with — and actually sharpens — the permutation test's near-coin-flip result (p=0.47): between the two, SRW looks like the clearest "no real signal" case in the whole analysis so far, more so than WTI even.

```python
n_permutations = 200
rng = np.random.default_rng(42)
null_r2_srw = []

for _ in range(n_permutations):
    y_shuffled_srw = rng.permutation(y_srw)
    y_shuffled_log_srw = np.log(y_shuffled_srw)
    df_model_shuffled_srw = df_model_srw.copy()
    df_model_shuffled_srw['duration_days'] = y_shuffled_srw
    df_model_shuffled_srw['log_duration']  = y_shuffled_log_srw

    _, y_eval_shuf, pred_shuf, _ = nested_vif_trim_ts(
        df_model_shuffled_srw, X_srw, feature_cols_srw, y_shuffled_srw, y_shuffled_log_srw
    )
    null_r2_srw.append(r2_score(y_eval_shuf, pred_shuf))

null_r2_srw = np.array(null_r2_srw)
print(f"Real R² (nested VIF-trim, SRW): {r2_vif_nested_srw:.3f}")
print(f"Null distribution: mean={null_r2_srw.mean():.3f}, 95th pct={np.percentile(null_r2_srw, 95):.3f}")
print(f"Real R² beats {np.mean(null_r2_srw < r2_vif_nested_srw)*100:.1f}% of random permutations")

n_at_least_as_extreme_srw = np.sum(null_r2_srw >= r2_vif_nested_srw)
p_value_srw = (n_at_least_as_extreme_srw + 1) / (n_permutations + 1)
print(f"Exact permutation p-value: p = {p_value_srw:.4f} (Phipson & Smyth, 2010 correction)")
```

Not significant, and this time barely better than chance at all — real R² beats only 53.5% of permutations, essentially a coin flip, with p=0.4677. The null distribution's mean (-0.258) is close to the real R² (-0.208), unlike HRW where the gap was much larger (-0.394 vs -0.073). So despite VIF-trim's steady improvement across the three feature-selection methods, this permutation test is the most decisive "no real signal" result we've seen yet — even weaker than WTI's. Combined with the earlier steady improvement trend, the honest read is: VIF-trim found the least-bad feature set, but "least-bad" here still isn't distinguishable from noise.

```python
from sklearn.model_selection import GridSearchCV, KFold
from statsmodels.stats.outliers_influence import variance_inflation_factor

# single-pass VIF trim, for the search only
vif_features_srw_single = list(feature_cols_srw)
while True:
    X_vif_check_srw = df_model_srw[vif_features_srw_single].dropna()
    vif_check_srw = pd.Series(
        [variance_inflation_factor(X_vif_check_srw.values, i) for i in range(X_vif_check_srw.shape[1])],
        index=vif_features_srw_single)
    worst_feature, worst_vif = vif_check_srw.idxmax(), vif_check_srw.max()
    if worst_vif <= 10 or len(vif_features_srw_single) <= 2:
        break
    vif_features_srw_single.remove(worst_feature)

feature_cols_trimmed_srw = vif_features_srw_single
print(f"SRW single-pass VIF trim kept: {feature_cols_trimmed_srw}")

order_srw = np.argsort(df_model_srw['start_date'].values)
X_trimmed_srw     = df_model_srw[feature_cols_trimmed_srw].values[order_srw]
y_trimmed_srw     = df_model_srw['duration_days'].values[order_srw]
y_log_trimmed_srw = df_model_srw['log_duration'].values[order_srw]

final_search_srw = GridSearchCV(
    xgb.XGBRegressor(random_state=42, n_jobs=-1, verbosity=0),
    param_grid=XGB_PARAM_GRID, cv=KFold(n_splits=5, shuffle=True, random_state=42),
    scoring='neg_mean_squared_error', n_jobs=-1
)
final_search_srw.fit(X_trimmed_srw, y_log_trimmed_srw)
print("Best params found (SRW):", final_search_srw.best_params_)

XGB_PARAMS_TUNED_SRW = dict(final_search_srw.best_params_, random_state=42, n_jobs=-1, verbosity=0)

_, y_srw_final, pred_log_srw_final, _ = run_timeseries_cv(
    df_model_srw, X_trimmed_srw, y_trimmed_srw, y_log_trimmed_srw, xgb_params=XGB_PARAMS_TUNED_SRW)
_, _, r2_final_tuned_srw = evaluate(y_srw_final, pred_log_srw_final, "SRW -- tuned, honest check")
print(f"\nUntuned nested VIF-trim: R² = {r2_vif_nested_srw:.3f}")
print(f"Tuned (honest re-check): R² = {r2_final_tuned_srw:.3f}")

seed_results_srw = []
for seed in [7, 42, 123, 2024, 99]:
    search = GridSearchCV(xgb.XGBRegressor(random_state=seed, n_jobs=-1, verbosity=0),
                           param_grid=XGB_PARAM_GRID, cv=KFold(n_splits=5, shuffle=True, random_state=seed),
                           scoring='neg_mean_squared_error', n_jobs=-1)
    search.fit(X_trimmed_srw, y_log_trimmed_srw)
    params_seed = dict(search.best_params_, random_state=42, n_jobs=-1, verbosity=0)
    _, y_seed, pred_seed, _ = run_timeseries_cv(df_model_srw, X_trimmed_srw, y_trimmed_srw, y_log_trimmed_srw, xgb_params=params_seed)
    _, _, r2_seed = evaluate(y_seed, pred_seed, f"seed={seed}")
    seed_results_srw.append({'seed': seed, 'r2_honest': r2_seed})

seed_df_srw = pd.DataFrame(seed_results_srw)
print(seed_df_srw.to_string(index=False))
print(f"Mean R²: {seed_df_srw['r2_honest'].mean():.3f} | Range: [{seed_df_srw['r2_honest'].min():.3f}, {seed_df_srw['r2_honest'].max():.3f}]")
```

```python
N_HOLDOUT = 6
df_sorted_srw = df_model_srw.sort_values('start_date').reset_index(drop=True)
holdout_dates_srw = df_sorted_srw['start_date'].iloc[-N_HOLDOUT:].values
holdout_mask_srw = df_model_srw['start_date'].isin(holdout_dates_srw)
dev_mask_srw = ~holdout_mask_srw

print(f"Development set: {dev_mask_srw.sum()} crises | Holdout: {holdout_mask_srw.sum()} crises")
print("Holdout crises (untouched until this cell):")
print(df_model_srw.loc[holdout_mask_srw, ['start_date', 'duration_days']].to_string())

X_dev_srw     = df_model_srw.loc[dev_mask_srw, feature_cols_srw].values
y_dev_srw     = df_model_srw.loc[dev_mask_srw, 'duration_days'].values
y_dev_log_srw = df_model_srw.loc[dev_mask_srw, 'log_duration'].values
X_holdout_srw = df_model_srw.loc[holdout_mask_srw, feature_cols_srw].values
y_holdout_srw = df_model_srw.loc[holdout_mask_srw, 'duration_days'].values

# VIF-based feature selection using DEV data only
dev_df_srw = pd.DataFrame(X_dev_srw, columns=feature_cols_srw)
vif_features_holdout_srw = list(feature_cols_srw)
while len(vif_features_holdout_srw) > 2:
    X_vif_h = dev_df_srw[vif_features_holdout_srw].dropna()
    vif_h = pd.Series([variance_inflation_factor(X_vif_h.values, i) for i in range(X_vif_h.shape[1])],
                       index=vif_features_holdout_srw)
    worst_f, worst_v = vif_h.idxmax(), vif_h.max()
    if worst_v <= 10:
        break
    vif_features_holdout_srw.remove(worst_f)

dev_idx_srw = [feature_cols_srw.index(c) for c in vif_features_holdout_srw]
final_model_srw = xgb.XGBRegressor(**XGB_PARAMS)
final_model_srw.fit(X_dev_srw[:, dev_idx_srw], y_dev_log_srw)
holdout_pred_srw = np.exp(final_model_srw.predict(X_holdout_srw[:, dev_idx_srw]))

_, _, r2_holdout_srw = evaluate(y_holdout_srw, holdout_pred_srw, "SRW -- confirmatory holdout")
print(pd.DataFrame({'start_date': df_model_srw.loc[holdout_mask_srw, 'start_date'].values,
                     'actual_days': y_holdout_srw, 'predicted_days': np.round(holdout_pred_srw).astype(int)}))
```

Negative on the holdout too (R²=-0.176). Short crises land roughly in the right range (9→6, 2→3, 4→6), but the model badly misses the 63-day crisis (predicted just 8) and, oddly, overshoots on two other short 3-4 day crises (→15, →19) — a noisier, less consistent error pattern than Gold/HRW's holdouts showed, more like the model hasn't found any real generalizable rule at all. With 29 dev crises (the largest of any commodity tested), more data didn't translate into better generalization here.

```python
gaps_srw = df_model_srw.sort_values('start_date')['start_date'].diff().dt.days
close_dates_srw = df_model_srw.sort_values('start_date').loc[gaps_srw <= 5, 'start_date']
keep_mask_srw = ~df_model_srw['start_date'].isin(close_dates_srw)

df_model_srw_declustered = df_model_srw.loc[keep_mask_srw].copy()
_, y_decl_srw, pred_decl_srw, _ = nested_vif_trim_ts(
    df_model_srw_declustered, df_model_srw.loc[keep_mask_srw, feature_cols_srw].values, feature_cols_srw,
    df_model_srw.loc[keep_mask_srw, 'duration_days'].values, df_model_srw.loc[keep_mask_srw, 'log_duration'].values
)
_, _, r2_declustered_srw = evaluate(y_decl_srw, pred_decl_srw, "SRW -- declustered")
print(f"As-is (N={len(df_model_srw)}): R² = {r2_vif_nested_srw:.3f}")
print(f"Declustered (N={len(df_model_srw_declustered)}): R² = {r2_declustered_srw:.3f}")
```

Removing the 7 close-pair crises makes things worse, not better — R² drops from -0.208 to -0.305. Unlike HRW (barely moved) but similar to Gold (also worsened on declustering), SRW's already-weak result was getting modest support from those closely-timed episodes. Combined with the entirely-negative bootstrap CI and near-coin-flip permutation test, this is one more consistent piece of evidence that SRW shows no robust signal — the clearest "no real signal" case of the four commodities so far.

```python
censor_lookup_srw = crisis_table_srw.set_index('crisis_id')['end_date'] == master_srw.index[-1]
is_censored_srw = df_model_srw.index.to_series().map(censor_lookup_srw).fillna(False).values

y_lower_srw = df_model_srw['duration_days'].values.astype(float)
y_upper_srw = np.where(is_censored_srw, np.inf, y_lower_srw)
print(f"Right-censored episodes: {is_censored_srw.sum()} of {len(is_censored_srw)}")

df_eval_aft_srw, y_aft_srw, pred_aft_srw = nested_vif_trim_ts_aft(
    df_model_srw, X_srw, feature_cols_srw, y_lower_srw, y_upper_srw, y_srw
)
_, _, r2_aft_srw = evaluate(y_aft_srw, pred_aft_srw, "SRW -- nested VIF-trim AFT, TimeSeriesSplit")

abs_err_plain_srw = np.abs(y_vif_srw - pred_vif_srw)
abs_err_aft_srw   = np.abs(y_aft_srw - pred_aft_srw)

stat, p_value = wilcoxon(abs_err_plain_srw, abs_err_aft_srw, alternative='two-sided')
print(f"Wilcoxon signed-rank, plain vs AFT, n={len(y_aft_srw)}: W={stat:.1f}, p={p_value:.4f}")
print(f"Mean |error| -- Plain: {abs_err_plain_srw.mean():.2f} days | AFT: {abs_err_aft_srw.mean():.2f} days")
```
